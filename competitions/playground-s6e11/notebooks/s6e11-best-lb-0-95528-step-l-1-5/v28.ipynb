<div style="color: #ffffff; font-size: 12px; letter-spacing: 3px; text-transform: uppercase; opacity: 0.9;">Playground Series S6E11 &middot; Predicting Credit Card Fraud &middot; ROC-AUC</div>
<div style="color: #ffffff; font-size: 36px; font-weight: 800; line-height: 1.12; margin-top: 10px;">S6E11 | LB 0.95484 | Top Public, Fully Explained</div>
<div style="color: #ffffff; font-size: 17px; margin-top: 8px;">Public ladders, their noise, the winner's curse, and the nested CV that tells you which steps will survive the private 80%</div>
<div style="margin-top: 18px; line-height: 2.0;">
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">public LB = 20% of test</span>
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">private = 80%</span>
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">nested CV</span>
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">every number computed here</span>
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">public inputs only, SHA-256 checked</span>
<span style="display: inline-block; white-space: nowrap; background: #ffffff; color: #0b1f4b; border-radius: 12px; padding: 3px 11px; font-size: 12.5px; font-weight: 700; margin: 0 6px 6px 0;">CPU, about 15 min</span>
</div>
</div>

<div style="background: #0b1f4b; color: #ffffff; padding: 12px 16px; border-radius: 8px; margin: 10px 0; font-size: 15px; line-height: 1.5;">
<b style="color: #ffffff; font-size: 16px;">This notebook's <code style="color: #0b1f4b; background: #ffffff;">submission.csv</code> scores 0.95484 on the public board. We built it by steering on a model of the public board, and below we show why most of that score will likely not survive the private leaderboard.</b><br>
The claim is about our file only. It builds on Areeb Ali's <i>Steer, Then Mix</i> (0.95423) and goodpjw2008's 0.95421 file (credits in <a style="color: #ffd9c2; text-decoration: underline;" href="#s17">section v12</a>). It is a pure public-board file with no OOF reading, so we also write a CV-backed alternative.
</div>

The top of the public board is a dense band of files a few hundred-thousandths apart. Many of them are **ladders**:
a strong public file plus small steps, each kept because the public score went up by one or two points
(1 point = 0.00001 AUC). But the public score is computed on only ~60,000 rows with about a thousand frauds. This
notebook measures, with public files only, **how big the noise of one such step is, how much of a ladder's gain
nested CV confirms, and how much of the top band is likely to move on the private 80%**. It takes no side: the ladder
authors themselves list every probe and check their steps on CV, which is what makes this analysis possible.

<div style="background: #fff4ec; color: #1a1a1a; border-left: 5px solid #eb6834; padding: 12px 16px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a; font-size: 15px;">The v7 story in five lines (15 Nov)</b><br>
&bull; <b style="color: #1a1a1a;">What the new top public notebook does.</b> goodpjw2008's <i>Forecast the Board</i> (LB 0.95417) climbs from the 0.95388 file in ten small re-weightings of 87 public models. Most steps are chosen by a <i>public model</i>: a model of the public score, fitted on the scores of 69 files submitted before. The author says all of this openly.<br>
&bull; <b style="color: #1a1a1a;">Public versus OOF.</b> 4 of the 5 kept forecast-led steps went up on public (+23 points in all) while OOF went down on at least 4 of 5 folds (-85 points of the author's nested CV). The 2 CV-led steps went up on OOF on 5/5, 5/5 folds and moved the board by -1, +4.<br>
&bull; <b style="color: #1a1a1a;">Expected private change</b> (from the 0.95388 file, 90% interval): the 0.95417 file -70 points [-89, -52], the CV-led file (public 0.95387) +7 points [+3, +12]. Between the two: the 0.95417 file is expected 78 points (90% interval 59 to 96) below the CV-led file on private (54 on in-sample OOF), although it is 30 points higher on public.<br>
&bull; <b style="color: #1a1a1a;">Why.</b> Each scored file is one linear equation about the 60,000 public labels; a step aimed along those equations wins public points that the 240,000 private rows do not share. In a simulation on OOF rows, a step steered by a public model fitted on 69 scored files reads +5.0 on public and 0.0 on private (first order); steered by CV, +0.8 and +0.3.<br>
&bull; <b style="color: #1a1a1a;">Conclusion.</b> The 0.95417 file is an open, well-documented public-board fit; for the private rows the OOF evidence favours the CV-led file, as its author also says. Since v12 the notebook's submission.csv is a pure public-board file (public 0.95424, no OOF reading available); it also writes the v11 public-board-steered file, a CV-backed alternative and the CV-led file, all rebuilt byte for byte from public files and credited.
</div>

<div style="background: #f3f5f9; color: #1a1a1a; border-left: 5px solid #8a8984; padding: 10px 14px; border-radius: 6px; margin: 6px 0; font-size: 13px;">
<b style="color: #1a1a1a;">Words used.</b> <b style="color: #1a1a1a;">OOF</b> (out-of-fold): predictions for training rows made by models that did not see those rows, our stand-in for the test set.
<b style="color: #1a1a1a;">Nested CV</b>: every choice made on four folds and scored on the fifth. <b style="color: #1a1a1a;">Noise SD</b>: how much a score gap moves by chance from one random 60,000-row public part to another.
<b style="color: #1a1a1a;">Winner's curse</b>: keeping the best of several noisy scores makes the kept score look better than it is. 1 point = 0.00001 AUC.
</div>

<a id="setup"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">Setup &middot; public inputs, checked byte for byte</div>

Every input is a public file: the competition data, public OOF/test predictions and two public ladder base files.
Each one is found anywhere under `/kaggle/input` by name and checked by its SHA-256 (the `find()` idea is
goodpjw2008's). The notebook runs on CPU in about ten minutes and never trains a big model.

````python
# Imports, input discovery by SHA-256, the shared scale for predictions, and a fast AUC for repeated random splits.
import os, time, hashlib, warnings, math
from pathlib import Path
import numpy as np, pandas as pd
from scipy.stats import rankdata, norm
from scipy.special import logit
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore")
from IPython.display import display, HTML
T0 = time.time()
def log(*a): print(f"[{(time.time() - T0) / 60:4.1f} min]", *a, flush=True)

# Reader-facing output is HTML with EXPLICIT colours (Kaggle's dark theme would otherwise paint light-grey text on the
# pale boxes). Every verification result goes to one "Checks" table, shown once after the computations.
TX, TX2, NAVY_BG, ROW_A, ROW_B, LINE_C = "#1a1a1a", "#4a4a46", "#0b1f4b", "#ffffff", "#f3f5f9", "#d6d5d0"
CHECKS = []
def check(what, result, ok=True):
    """Record one verification (shown later in the Checks table)."""
    CHECKS.append((what, str(result), bool(ok)))

def _cell(v):
    return "&ndash;" if v is None or (isinstance(v, float) and v != v) else str(v)

def table(df, title=None, note=None, right=None, wrap=None):
    """A compact styled HTML table: navy header, explicit dark text on white rows, right-aligned numeric columns."""
    df = df.reset_index(drop=True)
    right = set(right if right is not None else [c for c in df.columns if df[c].astype(str).str.match(r"^[\s+\-\u2212\u00b1\[\]()0-9.,%/e<>=x]*$").all()])
    wrap = set(wrap or [])
    th = "".join(f'<th style="background:{NAVY_BG};color:#ffffff;padding:6px 10px;font-weight:600;text-align:{"right" if c in right else "left"};'
                 f'white-space:nowrap;border:none;">{c}</th>' for c in df.columns)
    rows = []
    for i, r in df.iterrows():
        bg = ROW_A if i % 2 == 0 else ROW_B
        tds = "".join(f'<td style="background:{bg};color:{TX};padding:5px 10px;border:none;border-bottom:1px solid #e6e5df;'
                      f'text-align:{"right" if c in right else "left"};{"" if c in wrap else "white-space:nowrap;"}'
                      f'font-variant-numeric:tabular-nums;">{_cell(r[c])}</td>' for c in df.columns)
        rows.append(f"<tr>{tds}</tr>")
    head = f'<div style="color:{TX};font-weight:700;font-size:14px;margin:0 0 6px 2px;">{title}</div>' if title else ""
    foot = f'<div style="color:{TX2};font-size:12px;margin:6px 2px 0 2px;">{note}</div>' if note else ""
    return HTML(f'<div style="background:#ffffff;color:{TX};border:1px solid {LINE_C};border-radius:8px;padding:10px 12px;'
                f'margin:6px 0;max-width:100%;overflow-x:auto;font-family:-apple-system,Segoe UI,Helvetica,Arial,sans-serif;">{head}'
                f'<table style="border-collapse:collapse;font-size:13px;color:{TX};margin:0;">'
                f'<thead><tr>{th}</tr></thead><tbody>{"".join(rows)}</tbody></table>{foot}</div>')

def checks_table():
    ok = sum(c[2] for c in CHECKS)
    df = pd.DataFrame({"check": [c[0] for c in CHECKS], "result": [c[1] for c in CHECKS],
                       "": ['<span style="color:#0f7a55;font-weight:700;">&#10003; ok</span>' if c[2] else
                            '<span style="color:#b3261e;font-weight:700;">&#9888; see note</span>' for c in CHECKS]})
    return table(df, f"Checks: {ok} of {len(CHECKS)} passed", right=[], wrap=["check", "result"])

INPUT = Path(os.environ.get("KAGGLE_INPUT", "/kaggle/input"))
WORK = Path(os.environ.get("KAGGLE_WORKING", "/kaggle/working"))
WRITE_SUBMISSION = False   # set True to keep submission.csv and the other submission files in the output
_sha = {}
def sha256(p):
    p = Path(p)
    if p not in _sha: _sha[p] = hashlib.sha256(p.read_bytes()).hexdigest()
    return _sha[p]

def find(names, digest=None, inside=None, required=True):
    """A file anywhere under /kaggle/input, by (one of its) names and, if given, its SHA-256.
    If no file has the expected hash but one has the name, use it and say so (the author may have updated it)."""
    names = [names] if isinstance(names, str) else names
    hits = [p for n in names for p in sorted(INPUT.rglob(n)) if inside is None or inside in str(p)]
    good = [p for p in hits if digest is None or sha256(p) == digest]
    if good: return good[0]
    if hits:
        check(f"input {names[0]}", "found, but its SHA-256 differs from the one this notebook was written with (the author may have updated it); numbers may differ slightly", False)
        return hits[0]
    if required: raise FileNotFoundError(names)
    check(f"input {names[0]}", "not attached; the cells that need it are skipped", False)
    return None

def L(v):
    """Logit rank: AUC only sees order, so every prediction goes on one scale (as the public ladders do)."""
    return logit((rankdata(v) - 0.5) / len(v))

# daoviet removed his GP + DVAE OOF/test files from his dataset on 13 Nov; Aman Atar's q17 bundle carries the same
# predictions (identical ranks, checked below by a hash of the ranks), so they are read from there when needed.
GPD_IN_BUNDLE = {"oof_ours.npy": ("oof_daoviet_gpdvae", "1c7c7fa5e49aaf8f6f63f40bb258bdff864dd49f15239fa10bb2a9c55b800688"),
                 "test_ours.npy": ("test_daoviet_gpdvae", "10f7dec317a923cc4bb67b3aff770db528abd330b53da8c53be23e8b636d050d")}
def load(names, digest, required=True):
    nm = names if isinstance(names, str) else names[0]
    if nm in GPD_IN_BUNDLE and not any(sha256(q) == digest for q in INPUT.rglob(nm)):
        key, rank_sha = GPD_IN_BUNDLE[nm]
        v = np.load(find("rank1_q17_streams.npz", "0fe4ab1263a673c8065d7145590b92da440263161bf860ce45f2976b5c4fa56f"), allow_pickle=False)[key].astype(np.float64)
        assert hashlib.sha256(rankdata(v).tobytes()).hexdigest() == rank_sha, "bundle copy of GP + DVAE differs in rank order"
        check(f"daoviet's GP + DVAE ({'OOF' if 'oof' in nm else 'test'} side), read from Aman Atar's q17 bundle",
              "rank order identical to daoviet's original file (hash of the ranks)")
        return v
    p = find(names, digest, required=required)
    return None if p is None else np.load(p, allow_pickle=False).astype(np.float64)

def auc_prep(score):
    """Pre-sort a score once: the negatives in score order, and for every positive the number of negatives below it."""
    o = np.argsort(score, kind="stable"); yo = y[o]
    neg_o = o[~yo].astype(np.int32); pos_rows = o[yo].astype(np.int32)
    kpos = np.cumsum(~yo)[yo].astype(np.int32)
    return neg_o, pos_rows, kpos

def auc_counts(prep, counts, denom):
    """AUC on a bootstrap sample (counts[i] = how often row i was drawn): O(n) per sample, no sort (ties ignored)."""
    neg_o, pos_rows, kpos = prep
    cn = np.concatenate(([0], np.cumsum(counts[neg_o], dtype=np.int64)))
    return float((counts[pos_rows].astype(np.int64) * cn[kpos]).sum()) / denom

def auc_sort(yr, s):
    """Plain AUC by one sort (Mann-Whitney), ~3x faster than sklearn on 700k rows; ties ignored."""
    idx = np.flatnonzero(yr[np.argsort(s, kind="stable")]); npos = len(idx)
    return (idx.sum() - npos * (npos - 1) / 2) / (npos * (len(yr) - npos))

COMP = "playground-series-s6e11"
train = pd.read_csv(find("train.csv", inside=COMP))
test_ids = pd.read_csv(find("test.csv", inside=COMP), usecols=["id"])["id"].to_numpy()
sample = pd.read_csv(find("sample_submission.csv", inside=COMP))
y = train["is_fraud"].to_numpy().astype(bool)
FOLDS = [b for _, b in StratifiedKFold(5, shuffle=True, random_state=42).split(y, y)]   # the split every public OOF uses
N, NT = len(y), len(test_ids)
log(f"train {N:,} rows ({y.mean():.2%} fraud), test {NT:,} rows")
````

````python
# One style for every figure: a colorblind-checked palette with fixed roles, recessive axes, titles that state the
# takeaway, KPI tiles, and the helpers that place text without clutter.
from matplotlib.patches import FancyBboxPatch, Patch
from matplotlib.lines import Line2D
from matplotlib.ticker import FuncFormatter, MaxNLocator
INK, INK2, MUTED, RULE, SURF, PAGE, NAVY = "#0b0b0b", "#52514e", "#8a8984", "#d6d5d0", "#fcfcfb", "#f9f9f7", "#0b1f4b"
C = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#e87ba4", "#008300", "#6250d6", "#e34948"]
PUB, PRIV, CV, BAND = C[0], C[1], C[2], "#e6e5df"      # fixed roles: public = blue, private = orange, CV = aqua
CVTXT = "#0f7a55"                                        # darker aqua for text next to aqua marks
plt.rcParams.update({"figure.facecolor": PAGE, "axes.facecolor": SURF, "savefig.facecolor": PAGE, "axes.edgecolor": "#c3c2b7",
                     "axes.labelcolor": INK2, "xtick.color": INK2, "ytick.color": INK2, "axes.grid": True,
                     "grid.color": "#e8e7e3", "grid.linewidth": 0.7, "axes.spines.top": False, "axes.spines.right": False,
                     "font.family": "DejaVu Sans", "font.size": 12, "axes.titlesize": 13.5, "axes.titleweight": "bold",
                     "axes.titlecolor": INK, "axes.titlelocation": "left", "axes.titlepad": 10, "lines.linewidth": 2,
                     "legend.frameon": False, "axes.axisbelow": True, "figure.dpi": 110, "xtick.labelsize": 11, "ytick.labelsize": 11, "legend.fontsize": 11, "axes.labelsize": 12})
E5 = FuncFormatter(lambda v, _: f"{v:+g}" if v else "0")    # axis ticks in units of 1e-5 AUC

def headline(fig, title, sub=None, top_pad=0.0):
    """Figure title that states the takeaway (left aligned, wrapped to the figure width), with an optional grey subtitle."""
    import textwrap
    w, h = fig.get_figwidth(), fig.get_figheight()
    tl = textwrap.wrap(title, int(w * 72 / (16 * 0.56)))
    sl = textwrap.wrap(sub, int(w * 72 / (12 * 0.52))) if sub else []
    fig.text(0.012, 1 - 0.10 / h, "\n".join(tl), ha="left", va="top", fontsize=16, fontweight="bold", color=INK, linespacing=1.15)
    y_sub = 0.10 + 0.30 * len(tl) + 0.06
    if sl: fig.text(0.012, 1 - y_sub / h, "\n".join(sl), ha="left", va="top", fontsize=12, color=INK2, linespacing=1.2)
    used = y_sub + (0.23 * len(sl) if sl else 0) + 0.12
    fig.tight_layout(rect=(0, 0, 1, 1 - (used + top_pad) / h))

def tile(fig, rect, label, value, note, color=NAVY):
    """A KPI tile: label, big value, small note, with a coloured accent bar."""
    ax = fig.add_axes(rect); ax.set_axis_off()
    ax.add_patch(FancyBboxPatch((0, 0), 1, 1, boxstyle="round,pad=0,rounding_size=0.08", transform=ax.transAxes,
                                facecolor=SURF, edgecolor="#e6e5df", lw=1.2))
    ax.add_patch(FancyBboxPatch((0, 0.10), 0.018, 0.80, boxstyle="round,pad=0,rounding_size=0.01", transform=ax.transAxes,
                                facecolor=color, edgecolor="none"))
    ax.text(0.07, 0.76, label, transform=ax.transAxes, fontsize=11.5, color=INK2, va="center")
    ax.text(0.07, 0.45, value, transform=ax.transAxes, fontsize=22, fontweight="bold", color=INK, va="center")
    ax.text(0.07, 0.15, note, transform=ax.transAxes, fontsize=10.5, color=MUTED, va="center")

def panel_title(ax, title, sub=None):
    ax.set_title(title, loc="left", fontsize=13.5, fontweight="bold", color=INK, pad=26 if sub else 8)
    if sub: ax.text(0, 1.02, sub, transform=ax.transAxes, fontsize=11, color=INK2, va="bottom")

def zero_line(ax, axis="y"):
    (ax.axhline if axis == "y" else ax.axvline)(0, color="#b9b8b3", lw=1, zorder=1)
````

**How the numbers below are made.** The test labels are hidden, so we use the next best thing: the **nested OOF of
goodpjw2008's public wide LM stack** (Part 12, AUC 0.9526 on 699,960 training rows). It is a real top-band
prediction with honest out-of-fold scores. We draw hundreds of random public parts (~60k rows) and private parts
(~240k rows) from it, as two parts of a fresh test set would be drawn, and watch how scores move. Every ladder step that has public OOF files
behind it is applied to the same OOF, exactly as the ladder authors did in their own CV checks.

````python
# Load the public OOF predictions (train side). W = the nested OOF of the public wide LM stack.
OOF_FILES = {   # key: (file name(s), SHA-256, who published it)
    "W":   ("wide_lm_stack_nested.npy", "69375871698f1adf2b4c666fc7f5180edf00ac6836668353b8e7e4ca43eb2722", "goodpjw2008, wide LM stack (nested OOF)"),
    "LM":  (["lm_oof.npy", "megalm_oof_sha181c.npy"], "181ceac8f3a48fcf9a1702e1222ee13178f0688a28c1aff22c8f881532ea981a", "megayak, target-first LM"),
    "L7F": ("lm7_oof_5fold.npy", "c56d6b7282e7ea8c4c5927ac27e8b286d539ffbd66d5795564323c8b92b65242", "goodpjw2008, compact LM seed 7"),
    "GPD": ("oof_ours.npy", "91be476ddb00fe1232ba98b9a30f3c6215e81769499e3594058e4e99703c706a", "daoviet, GP + DVAE"),
    "MLP": ("oof_gbn_bag5_w512.npy", "e8ef910425641b7be8f3bf8f964b1637e9fa39937a772df56ef0cb660d3776f3", "goodpjw2008, OOF bank MLP bag"),
    "GB":  ("oof_gb_bag5.npy", "4b4bc47571ee389a54a7e70d6ac2d716207c329a1ca619a37e577de1f30aea40", "goodpjw2008, OOF bank GB bag"),
    "MS":  ("oof.npy", "3468aaaeb7851739ec0de1e06b780da6d2777fcaeef638a6852c56e68e67a206", "masayakawamata, target-first GPT-2"),
    "G26": ("gpt2_s2026_oof_5fold.npy", "4a902ecd3d725d21123f5b6239ac2190a9fd94f3de37e3157be391fcf3fe996a", "goodpjw2008, GPT-2 seed 2026"),
}
O = {k: L(load(n, h)) for k, (n, h, _) in OOF_FILES.items()}
W = O["W"]
def cv(s): return np.mean([roc_auc_score(y[b], s[b]) for b in FOLDS])
def folds_up(s, base): return sum(roc_auc_score(y[b], s[b]) > roc_auc_score(y[b], base[b]) for b in FOLDS)
CV_W = cv(W)

# The OOF bank's 20 'core' members (one per recipe), used as extra probe directions and for the hill-climbing demo.
bank_dir = find("manifest.csv", inside="oof-bank").parent
bank_man = pd.read_csv(bank_dir / "manifest.csv")
CORE = [n for n in bank_man.loc[bank_man["core"], "name"] if (bank_dir / f"oof_{n}.npy").exists()]
BANK = {n: L(np.load(bank_dir / f"oof_{n}.npy", allow_pickle=False).astype(np.float64)) for n in CORE}
log(f"wide LM stack: nested CV {CV_W:.6f}; OOF bank: {len(BANK)} core members")
display(table(pd.DataFrame({"key": list(O), "public OOF predictions (author, model)": [OOF_FILES[k][2] for k in O],
                            "OOF AUC": [f"{roc_auc_score(y, O[k]):.6f}" for k in O]}),
              "Public OOF inputs on the shared five folds", "W, the nested OOF of the wide LM stack, is the stand-in test set of sections 2 to 7."))
````

````python
# Every public probe that has OOF files behind it, applied to W as a move  W + w * direction.
# 'pub' = the public-LB change the authors report for that single step (x 1e-5), with the source.
MOVES = [  # (short name, direction on the OOF, reported public change, source)
    ("+0.12 megayak LM",          0.12 * (O["LM"] - W),                       +1, "goodpjw2008 v1: 0.95342 -> 0.95343"),
    ("+0.06 seed-7 LM",           0.06 * (O["L7F"] - W),                      +2, "goodpjw2008 v1: 0.95343 -> 0.95345"),
    ("+0.12 public GPT-2",        0.12 * (O["MS"] - W),                      -14, "goodpjw2008 v1 probe: -14 points"),
    ("+0.05 GP + DVAE",           0.05 * (O["GPD"] - W),                      +3, "daoviet: 0.95348 -> 0.95351"),
    ("LM weight 0.12 -> 0.09",   -0.03 * (O["LM"] - W),                       +1, "Areeb Ali v19: 0.95347 -> 0.95348"),
    ("more of both LMs",          0.04 * (O["LM"] - W) + 0.03 * (O["L7F"] - W), -2, "Areeb Ali probe: 0.95347 -> 0.95345"),
    ("less seed-7 LM",           -0.02 * (O["L7F"] - W),                       0, "Areeb Ali probe: 0.95347 -> 0.95347"),
    ("GPT-2 seed swap 0.06",      0.06 * (O["G26"] - O["MS"]),                +2, "goodpjw2008 v2: 0.95352 -> 0.95354"),
    ("GPT-2 seed swap 0.10",      0.10 * (O["G26"] - O["MS"]),                +2, "goodpjw2008 v2 probe: 0.95354"),
]
TREE = -0.10 * (O["MLP"] - W) - 0.05 * (O["GB"] - W)        # Areeb Ali's full tree move (v15); no clean single-step public number
# A family of plausible probes: a small step towards (or away from) each bank member and each public LM.
rng = np.random.default_rng(11)
FAMILY = []
for name, X in list(BANK.items()) + [("megayak LM", O["LM"]), ("seed-7 LM", O["L7F"]), ("public GPT-2", O["MS"]),
                                     ("GPT-2 seed 2026", O["G26"]), ("GP + DVAE", O["GPD"])]:
    w = rng.choice([-0.10, -0.06, -0.03, 0.03, 0.06, 0.10])
    FAMILY.append((f"{w:+.2f} {name}", w * (X - W)))
SCORES = [("base", np.zeros(N))] + [(m[0], m[1]) for m in MOVES] + [("tree move", TREE)] + FAMILY

RHO, TRUE, UP = {}, {}, {}
RW = rankdata(W)
for name, d in SCORES[1:]:
    s = W + d
    RHO[name] = np.corrcoef(rankdata(s), RW)[0, 1]
    TRUE[name] = (cv(s) - CV_W) * 1e5
    UP[name] = folds_up(s, W)
log(f"nested-CV change of {len(SCORES) - 1} moves computed")
````

````python
# Random public/private splits: the public part is a bootstrap sample of NPUB rows from the OOF, the private part an
# independent bootstrap sample of the other NT - NPUB rows (independent draws, as two parts of a fresh test set would be).
R = int(os.environ.get("NB_DRAWS", 300))
NPUB = int(round(0.2 * NT))
preps = [auc_prep(W + d) for _, d in SCORES]
AUC = np.zeros((len(SCORES), R, 2))
rng = np.random.default_rng(2026)
for r in range(R):
    cp = np.bincount(rng.integers(0, N, NPUB), minlength=N).astype(np.int32)
    cq = np.bincount(rng.integers(0, N, NT - NPUB), minlength=N).astype(np.int32)
    dp = float(cp[y].sum()) * float(cp[~y].sum()); dq = float(cq[y].sum()) * float(cq[~y].sum())
    for i, pr in enumerate(preps):
        AUC[i, r, 0] = auc_counts(pr, cp, dp); AUC[i, r, 1] = auc_counts(pr, cq, dq)
del preps
DELTA = (AUC[1:] - AUC[:1]) * 1e5           # (move, draw, public/private) change vs the base file, in 1e-5
NAMES = [n for n, _ in SCORES[1:]]
SD_PUB = DELTA[:, :, 0].std(1); SD_PRV = DELTA[:, :, 1].std(1)
AGREE = (np.sign(DELTA[:, :, 0]) == np.sign(DELTA[:, :, 1])).mean(1)
ONE_RHO = np.array([1 - RHO[n] for n in NAMES])
K_RULE = np.exp(np.mean(np.log(SD_PUB * 1e-5 / np.sqrt(ONE_RHO))))   # sigma_public(delta) ~= K_RULE * sqrt(1 - rho)
SD_BASE_PUB, SD_BASE_GAP = AUC[0, :, 0].std(), (AUC[0, :, 0] - AUC[0, :, 1]).std()
log(f"{R} random 20/80 splits scored; noise rule K = {K_RULE:.4f}")
````

````python
# Rebuild the public ladder files on the TEST set from the authors' public inputs (formulas copied from their notebooks),
# and measure how different consecutive rungs are (rank correlation). The last file is checked byte for byte against
# the published 0.95354 output (Areeb Ali's notebook asserts the same SHA-256).
TEST_FILES = {
    "A":   ("submission_095340.csv", "12b525b84f29e32d6511e9d6105f6dc5eac79ae41b10140eed3353ea57afc159"),   # Strike 10, amanatar
    "F":   ("submission_095330.csv", "f4a7a4f8636c99fabd82945c7d81956c5559c609dcb55cde6d4e03f80b0ea105"),   # Strike 9, Moran Z
    "LM":  (["lm_test.npy", "megalm_test_sha3168.npy"], "3168779be644a530c6bdc4c141cae168bfc2feb01ac21090c917cc978a7687d3"),
    "L7F": ("lm7_test_5fold.npy", "68639fec79f081e97d40185443fa3903c9db3e62698abb01dfe31dcee217462a"),
    "GPD": ("test_ours.npy", "96f3bdde816e604438e61eac9722f7a7402f1c39bb8efe8999793b23f91d02f2"),
    "MLP": ("test_gbn_bag5_w512.npy", "035d6544d7c03108fc42ee296a8660a45825e3b1b19b7cabce1c3f42cac4e7cb"),
    "GB":  ("test_gb_bag5.npy", "7d54e5f17e0707d88fec1a2cccb05885978ae6cc2a595c5f4575c7b7d45a9bb8"),
    "MS":  ("test.npy", "6c27aff7be744f7a0cc17809e0cd04fcdea0664234af88e3040a25235fb67906"),
    "G26": ("gpt2_s2026_test_f0.npy", "40e08058ed7632ee328c2fe7d97040b856b36d2636f32cc6d2781ec9db080695"),
}
T = {}
for k, (n, h) in TEST_FILES.items():
    if isinstance(n, str) and n in GPD_IN_BUNDLE: T[k] = L(load(n, h)); continue
    p = find(n, h, required=False)
    if p is None: continue
    if str(p).endswith(".csv"):
        d = pd.read_csv(p); assert (d["id"].to_numpy() == sample["id"].to_numpy()).all(); T[k] = L(d["is_fraud"].to_numpy())
    else: T[k] = L(np.load(p, allow_pickle=False).astype(np.float64))
HAVE_LADDER = len(T) == len(TEST_FILES)
RUNGS = []   # (label, test-file score, reported public LB)
if HAVE_LADDER:
    A, F = T["A"], T["F"]
    a2 = A + 0.6 * (A - F)
    a3 = a2 + 0.12 * (T["LM"] - F)
    s352 = A + 0.6 * (A - F) + 0.09 * (T["LM"] - F) + 0.09 * (T["L7F"] - F) + 0.05 * (T["GPD"] - F) - 0.05 * (T["MLP"] - F) - 0.025 * (T["GB"] - F)
    s354 = L(s352) + 0.06 * (T["G26"] - T["MS"])
    s354b = L(s352) + 0.10 * (T["G26"] - T["MS"])
    out = (rankdata(s354) - 0.5) / NT
    chk = WORK / "rebuild_095354.csv"; sample[["id"]].assign(is_fraud=out).to_csv(chk, index=False)
    _ok = sha256(chk) == "1e2e72fa26309ce630a19e0df780ae63d525901f33dc23f9058581ced5e9d6e0"
    check("0.95354 ladder file rebuilt from public inputs", "byte-identical to the published file (SHA-256)" if _ok else "SHA-256 differs (float rounding can swap tied ranks)", _ok)
    RUNGS = [("Strike 9 (Moran Z)", F, 95330), ("Strike 10 (amanatar)", A, 95340), ("+ 0.6 (S10 - S9)", a2, 95342),
             ("+ 0.12 megayak LM", a3, 95343), ("0.95352 file (several steps)", s352, 95352),
             ("+ 0.06 GPT-2 seed swap", s354, 95354)]
    rk = {lab: rankdata(s) for lab, s, _ in RUNGS}
    STEPS = []   # (label, public gain e-5, 1 - rho vs previous rung, predicted public noise SD e-5)
    for (l0, _, p0), (l1, _, p1) in zip(RUNGS[:-1], RUNGS[1:]):
        one_rho = 1 - np.corrcoef(rk[l0], rk[l1])[0, 1]
        STEPS.append((l1, p1 - p0, one_rho, K_RULE * np.sqrt(one_rho) * 1e5))
    one_rho = 1 - np.corrcoef(rankdata(s352), rankdata(s354b))[0, 1]
    STEPS_EXTRA = ("0.10 seed swap vs 0.95352 file", 2, one_rho, K_RULE * np.sqrt(one_rho) * 1e5)
    del rk
else:
    check("0.95354 ladder inputs", "not all attached; the rung figure uses the OOF moves only", False)
SIG_RUNG = float(np.median([s[3] for s in STEPS])) if HAVE_LADDER else float(np.median(SD_PUB))
log(f"typical public noise of one ladder rung: {SIG_RUNG:.1f}e-5")
````

````python
# Winner's curse, two ways.
# (1) Textbook: K probes with zero true gain, public noise SD sigma: the best public score is sigma * E[max of K normals].
KS = np.arange(1, 101)
_z = np.linspace(-6, 8, 4001)
_trap = getattr(np, "trapezoid", None) or np.trapz
EMAX = np.array([_trap(_z * k * norm.pdf(_z) * norm.cdf(_z) ** (k - 1), _z) for k in KS])
# (2) On real data: choose the best of k probes from the 'family' (and the ladder moves) by the public part of each
#     random split, then read the private part of the same split for the chosen probe.
rng = np.random.default_rng(7)
KGRID = [1, 2, 3, 5, 8, 12, 20, 30]
ALLI = np.arange(len(NAMES))
CURSE = {k: [] for k in KGRID}
for r in range(R):
    for k in KGRID:
        for _ in range(4):
            idx = rng.choice(ALLI, k, replace=False)
            j = idx[np.argmax(DELTA[idx, r, 0])]
            CURSE[k].append((DELTA[j, r, 0], DELTA[j, r, 1], TRUE[NAMES[j]]))
CURSE = {k: np.array(v) for k, v in CURSE.items()}
# A frozen snapshot of the public leaderboard (2026-11-12 12:11 UTC): scores >= 0.95200 in units of 1e-5 above 0.95,
# and each team's number of submissions. Team names are left out on purpose.
LB_TOP = np.array([int(v) for v in """370,369,369,369,364,364,363,363,360,357,356,355,355,355,355,354,354,354,354,354,354,354,354,354,354,354,354,354,354,354,354,354,353,352,352,352,352,351,351,351,351,351,351,351,351,350,350,349,349,349,349,349,349,348,348,348,348,348,348,348,348,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,346,346,346,346,346,346,346,346,346,346,346,346,346,346,346,346,346,345,345,345,345,344,344,344,344,343,343,343,343,343,343,343,342,340,340,340,340,339,339,339,337,337,337,335,334,334,333,333,333,332,331,331,331,331,331,331,331,331,331,331,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,328,327,326,323,322,322,322,320,319,317,316,316,315,314,313,313,311,310,310,310,309,309,308,307,307,307,306,306,305,305,305,305,303,303,302,302,301,301,300,300,300,300,300,300,300,300,300,300,300,300,300,300,299,298,298,298,298,298,297,295,295,295,294,294,294,293,291,291,291,290,288,288,288,287,286,286,285,285,285,285,285,285,285,285,285,285,285,285,285,285,285,284,280,278,277,277,276,275,275,274,273,273,271,268,267,267,267,266,266,265,265,263,262,262,260,259,258,258,258,258,256,256,255,255,255,254,251,250,249,249,247,247,246,246,245,245,244,243,243,242,242,241,241,240,239,239,239,239,239,239,239,239,239,239,239,239,239,239,238,238,238,238,237,237,236,235,234,234,234,233,233,231,231,231,229,229,229,227,226,225,224,224,224,222,222,222,222,222,218,217,217,216,213,213,213,212,212,212,211,210,209,209,209,208,206,206,205,205,203,203,202,201,201,200,200,200""".split(",")])
LB_SUBS = np.array([int(v) for v in """101,31,111,102,10,55,93,18,111,2,85,20,18,20,1,96,60,10,42,96,20,11,20,10,104,17,10,13,5,2,118,13,58,20,11,84,48,91,17,66,47,17,27,18,7,48,6,10,51,39,24,15,12,4,90,9,65,10,71,4,14,95,19,30,6,57,66,5,18,8,7,1,26,5,11,34,9,2,7,14,24,64,7,11,7,18,4,10,36,10,43,15,17,6,8,7,12,6,102,45,10,27,16,1,14,8,1,20,12,38,8,7,5,15,13,22,27,42,8,120,11,42,15,12,7,23,73,73,10,12,11,30,29,10,23,19,98,78,22,31,6,2,13,11,8,16,5,8,10,2,11,7,2,11,7,7,4,15,80,27,21,2,13,8,2,6,9,8,9,5,11,10,15,3,3,7,30,40,5,33,5,4,14,26,13,5,13,46,18,23,7,10,28,28,12,16,51,1,9,30,79,6,2,23,17,11,6,30,38,31,21,51,36,22,47,6,25,16,5,3,4,4,2,1,19,10,2,3,28,19,16,1,32,2,8,77,30,16,6,5,6,5,21,31,40,3,39,4,34,10,5,26,10,2,9,11,3,2,10,7,12,5,2,9,1,3,14,3,69,14,19,15,58,3,4,120,16,100,11,10,4,3,5,7,1,9,27,79,5,12,4,78,14,3,9,4,12,5,1,8,6,16,3,5,9,20,38,1,6,5,33,5,9,5,5,25,12,10,17,1,59,6,4,6,12,2,16,2,9,26,4,36,55,31,16,3,9,9,25,15,6,19,1,5,10,4,8,11,24,5,4,30,5,16,32,7,33,39,5,1,1,1,16,9,13,4,7,7,9,7,1,9,19,3,4,1,3,11,3,4,3,7,25,5,13,7,30,31,6,41,7,7""".split(",")])
LB_TEAMS = 1577
TOP_SCORE = LB_TOP.max()
N_2SIG = int((LB_TOP >= TOP_SCORE - 2 * SIG_RUNG).sum())
log("winner's-curse draws done")
````

**New in v4: the inputs of the two new ladders.** Both build on Aman Atar's public *rank-1 q17 bundle*: the public
0.95370 file plus 16 public models with out-of-fold (OOF) and test predictions on the shared five folds. goodpjw2008's
*Public and CV Up* replays the 0.95370 file on OOF rows and moves it towards 87 public members (the 16 bundle models,
the wide LM stack, 70 OOF-bank models). Areeb Ali's *Bold, Then Bolder* takes half a step further along one of
those moves. The cells below rebuild both from the public files and score every step on the same OOF rows.

````python
# v4 inputs: the q17 bundle (Aman Atar), the wide LM stack's test side, and the two LMs' test files (all checked by SHA-256).
QZ = np.load(find("rank1_q17_streams.npz", "0fe4ab1263a673c8065d7145590b92da440263161bf860ce45f2976b5c4fa56f"), allow_pickle=False)
FOLD_ID = np.zeros(N, np.int8)
for k_, b_ in enumerate(FOLDS): FOLD_ID[b_] = k_
assert (QZ["y_train"].astype(bool) == y).all() and (QZ["folds"] == FOLD_ID).all(), "bundle rows or folds differ from the shared split"
QN = [str(s) for s in QZ["stream_names"]]
W_TE = L(load("wide_lm_stack_test.npy", "b37471e6c34a50715284e807d68e15e5f3ad033e72400904f27876b24041f8cc"))
LM7_TE = L(load("lm7_test_5fold.npy", "68639fec79f081e97d40185443fa3903c9db3e62698abb01dfe31dcee217462a"))
LM_TE = L(load(["megalm_test_sha3168.npy", "lm_test.npy"], "3168779be644a530c6bdc4c141cae168bfc2feb01ac21090c917cc978a7687d3"))
_bank_files = sorted(p for p in bank_dir.glob("*.npy") if p.name.startswith(("oof_", "test_")))
BANK_OK = hashlib.sha256("".join(sha256(p) for p in _bank_files).encode()).hexdigest() == "ddd0b8e8112d49846e415a4ddbead968bce6d50fb8f351785f21a8053f420538"
check("OOF bank (goodpjw2008): all 140 files", "identical to the ones the ladders used (SHA-256 of the file hashes)" if BANK_OK else "differ from the ones the ladders used", BANK_OK)

# The 0.95370 file has no OOF of its own. goodpjw2008's replay: fit it on the test rows from the 16 bundle models and the
# wide LM stack (least squares), apply the same weights to their OOF predictions, then the same 0.06 LM move.
AT = L(QZ["test_anchor_095370"])
_one = lambda X: np.column_stack([X, np.ones(len(X))])
_XT = np.column_stack([L(QZ["test_" + n]) for n in QN] + [W_TE])
_XO = np.column_stack([L(QZ["oof_" + n]) for n in QN] + [W])
_b = np.linalg.lstsq(_one(_XT), AT, rcond=None)[0]
REPLAY_R2 = 1 - ((AT - _one(_XT) @ _b) ** 2).mean() / AT.var()
A_OOF = _one(_XO) @ _b
del _XT, _XO
REF_OOF = A_OOF + 0.06 * (L(QZ["oof_lm7_5fold"]) - L(QZ["oof_megayak_5f_lm"]))
U0 = (rankdata(AT + 0.06 * (LM7_TE - LM_TE)) - 0.5) / NT
REF_TE = logit(U0)
check("0.95370 file replayed on OOF rows (goodpjw2008's least-squares replay)", f"R\u00b2 {REPLAY_R2:.4f} on the test rows; replay OOF AUC {roc_auc_score(y, REF_OOF):.6f}")
log("0.95370 file replayed on the OOF rows")
````

````python
# Public noise of every step (the one-line rule on the TEST files) and the private give-back of public selection,
# simulated on the OOF rows with the real public/private sizes: on each random split the probe with the best public gap
# is kept, as the authors did, and its private gap is read.
TE = {"0.95354 file": L(out) if HAVE_LADDER else None, "0.95370 file": REF_TE}
for n, (c, _, _) in PROBES.items(): TE[n] = combo(c, "test")
def rho_te(a, b): return np.corrcoef(rankdata(TE[a]), rankdata(TE[b]))[0, 1]
def sig(r): return K_RULE * np.sqrt(max(1 - r, 0)) * 1e5

FAMILIES = {   # the probes each author compared on the board before keeping one, and the base they compared against
    "goodpjw2008": ("ref", ["careful", "balanced", "bold", "10% wide LM stack", "25% wide LM stack"]),
    "Areeb Ali": ("bold", ["1.25 x bold", "1.5 x bold", "bold + 0.5 (bold - careful)", "bold + 0.5 (bold - balanced)",
                           "bold + 1.0 (bold - balanced)", "bold + 1.5 (bold - balanced)", "bold + 2.0 (bold - balanced)",
                           "1.25 x (bold + 1.0 (bold - balanced))"]),
}
_sc = {"ref": REF_OOF, **PO}
_names = ["ref"] + [n for n in PO]
_preps = [auc_prep(_sc[n]) for n in _names]
QAUC = np.zeros((len(_names), R, 2))
rng = np.random.default_rng(2611)
for r_ in range(R):
    cp = np.bincount(rng.integers(0, N, NPUB), minlength=N).astype(np.int32)
    cq = np.bincount(rng.integers(0, N, NT - NPUB), minlength=N).astype(np.int32)
    dp = float(cp[y].sum()) * float(cp[~y].sum()); dq = float(cq[y].sum()) * float(cq[~y].sum())
    for i, pr in enumerate(_preps): QAUC[i, r_, 0] = auc_counts(pr, cp, dp); QAUC[i, r_, 1] = auc_counts(pr, cq, dq)
del _preps
QI = {n: i for i, n in enumerate(_names)}
def qgap(a, b): return (QAUC[QI[a]] - QAUC[QI[b]]) * 1e5        # (draw, public/private)
GIVE = {}
for fam, (base, probes) in FAMILIES.items():
    G = np.stack([qgap(p, base) for p in probes])                 # (probe, draw, 2)
    j = G[:, :, 0].argmax(0)
    pick_pub, pick_prv = G[j, np.arange(R), 0], G[j, np.arange(R), 1]
    GIVE[fam] = dict(pub=pick_pub, prv=pick_prv, back=pick_pub - pick_prv, sd_pub=G[:, :, 0].std(1), K=len(probes))
# check of the noise rule on these new, larger steps: measured public SD vs K_RULE * sqrt(1 - rho) on the OOF
RULE_CHECK = []
for n in ["careful", "balanced", "bold"]:
    rr = np.corrcoef(rankdata(PO[n]), rankdata(REF_OOF))[0, 1]
    RULE_CHECK.append((n, 1 - rr, qgap(n, "ref")[:, 0].std(), sig(rr)))
for n in ["bold + 0.5 (bold - balanced)", "bold + 1.5 (bold - balanced)", "1.5 x bold"]:
    rr = np.corrcoef(rankdata(PO[n]), rankdata(PO["bold"]))[0, 1]
    RULE_CHECK.append((n + " vs bold", 1 - rr, qgap(n, "bold")[:, 0].std(), sig(rr)))
_rel = max(abs(m - p) / m for _, _, m, p in RULE_CHECK)
check("noise rule on the larger new steps (6 moves)", "measured public SD vs rule: " + ", ".join(f"{m:.1f} vs {p:.1f}" for _, _, m, p in RULE_CHECK) + f" (largest relative gap {_rel:.0%})", _rel < 0.35)
log("give-back draws done")
````

````python
# ERROR ANALYSIS on the OOF rows (the stand-in test set): the honest wide LM stack W, the 0.95370 replay REF, the bold mix
# and the 0.95388 file's OOF counterpart. Every change below is computed by applying a step to a SUBSET of rows only.
EA = {"wide LM stack": W, "0.95370 replay": REF_OOF, "bold mix": L(PO["bold"]), "0.95388 file": B_O}
EA = {k: L(v) for k, v in EA.items()}
EA_AUC = {k: roc_auc_score(y, v) for k, v in EA.items()}
UR = {k: (rankdata(v) - 0.5) / N for k, v in EA.items()}
STEPS_EA = [("0.95370 replay", "bold mix"), ("bold mix", "0.95388 file"), ("wide LM stack", "0.95388 file")]
def masked_gain(a, b, m):
    """AUC change (x1e-5) when file a takes file b's logit ranks on the rows in mask m only."""
    s = EA[a].copy(); s[m] = EA[b][m]; return (auc_sort(y, s) - auc_sort(y, EA[a])) * 1e5
# (a) where the files disagree: the most-moved rows first
QGRID = [0.005, 0.01, 0.02, 0.05, 0.10, 0.20, 0.50, 1.0]
DISAG = {}
for a, b in STEPS_EA:
    mv = np.abs(UR[b] - UR[a]); order_ = np.argsort(-mv)
    gains = []
    for q in QGRID:
        m = np.zeros(N, bool); m[order_[:int(round(q * N))]] = True; gains.append(masked_gain(a, b, m))
    DISAG[(a, b)] = dict(gain=np.array(gains), fraud_top1=y[order_[:N // 100]].mean(), mv=mv)
# (b) segments: score band of the base file, how common the transaction amount is (value count over train + test), channel
_amt_te = pd.read_csv(find("test.csv", inside=COMP), usecols=["amount_usd"])["amount_usd"]
_vc = pd.concat([train["amount_usd"], _amt_te]).value_counts()
AMT_CNT = train["amount_usd"].map(_vc).to_numpy()
SEGS = {
    "score band (rank in the base file)": lambda base: [("bottom 80%", UR[base] < 0.80), ("80-95%", (UR[base] >= 0.80) & (UR[base] < 0.95)),
                                                        ("95-99%", (UR[base] >= 0.95) & (UR[base] < 0.99)), ("top 1%", UR[base] >= 0.99)],
    "how common the amount is": lambda base: [("unique", AMT_CNT == 1), ("2-3 rows", (AMT_CNT >= 2) & (AMT_CNT <= 3)),
                                              ("4-20 rows", (AMT_CNT >= 4) & (AMT_CNT <= 20)), ("> 20 rows", AMT_CNT > 20)],
    "channel": lambda base: [(str(c), (train["channel"] == c).to_numpy()) for c in sorted(train["channel"].unique())],
}
SEG_TAB = []
for a, b in STEPS_EA:
    for sname, fn in SEGS.items():
        for lab, m in fn(a):
            SEG_TAB.append(dict(step=f"{a} -> {b}", segmentation=sname, segment=lab, rows=m.mean(), frauds=y[m].sum() / y.sum(),
                                gain=masked_gain(a, b, m), moved=np.abs(UR[b] - UR[a])[m].mean() * 1e3,
                                lift=((UR[b] - UR[a])[m & y].mean() - (UR[b] - UR[a])[m & ~y].mean()) * 1e3))
    for lab, m in [("frauds only", y), ("legit rows only", ~y), ("frauds ranked below 80%", y & (UR[a] < 0.80))]:
        SEG_TAB.append(dict(step=f"{a} -> {b}", segmentation="by label (analysis only)", segment=lab, rows=m.mean(), frauds=y[m].sum() / y.sum(),
                            gain=masked_gain(a, b, m), moved=np.abs(UR[b] - UR[a])[m].mean() * 1e3, lift=np.nan))
SEG_TAB = pd.DataFrame(SEG_TAB)
# (c) rank shift of frauds and legit rows along the base file's score (calibration of the order)
EDGES = np.r_[0, 0.5, 0.8, 0.9, 0.95, 0.97, 0.98, 0.99, 0.995, 0.998, 1.0]
SHIFT = {}
for a, b in STEPS_EA:
    d = UR[b] - UR[a]; band = np.clip(np.searchsorted(EDGES, UR[a], side="right") - 1, 0, len(EDGES) - 2)
    SHIFT[(a, b)] = np.array([[d[(band == i) & y].mean() * 1e3, d[(band == i) & ~y].mean() * 1e3, (band == i).sum(), (y & (band == i)).sum()]
                              for i in range(len(EDGES) - 1)])
CAPTURE = {k: [y[UR[k] >= 1 - q].sum() / y.sum() for q in (0.01, 0.02, 0.05)] for k in EA}
log("error analysis done")
````

**New in v6: the inputs of the 0.95390 notebook.** Aman Atar's *2nd-Order Curved Ladder Probit* (LB 0.95390, 14 Nov)
loads its final file from a dataset that is not public, so its intermediate files cannot be rebuilt. Its **output

rebuilt the way goodpjw2008 rebuilt the 0.95370 file: the file is fitted on the **test rows** (no labels) from the
0.95370 file and the 87 public members of the mixes, and the same weights are applied to their OOF predictions.

````python
# v6: the 0.95390 file (Aman Atar), its OOF replay, its geometry, and the verdicts of its step.
from scipy.special import ndtri, expit
from scipy.optimize import minimize
S390_SHA = "6d73414ff3d3330020bebbe952177fa1bef3baf242b2748d41da8180311a96a1"
_p390 = find(["submission.csv", "amanatar_curved-ladder_submission.csv"], S390_SHA, inside="curved-ladder")   # or the pinned copy
_d390 = pd.read_csv(_p390)
assert (_d390["id"].to_numpy() == sample["id"].to_numpy()).all(), "row order of the 0.95390 file differs"
V390 = _d390["is_fraud"].to_numpy().astype(np.float64)
check("0.95390 file (Aman Atar), the output of his public notebook", f"SHA-256 {sha256(_p390)[:16]}&hellip; as published; "
      f"{len(np.unique(V390)):,} distinct values, no ties", sha256(_p390) == S390_SHA)
RHO390 = np.corrcoef(rankdata(V390), rankdata(S388))[0, 1]
SD390 = sig(RHO390)                                           # public noise SD of the step 0.95388 -> 0.95390

# Replay on the OOF rows: least squares on the TEST rows (no labels), the 0.95370 file + the 87 members, logit ranks.
KEYS = list(RECIPES["bold"])
_XT = np.column_stack([REF_TE] + [QM_T[k] for k in KEYS] + [np.ones(NT)])
_tg = L(V390)
_b = np.linalg.lstsq(_XT, _tg, rcond=None)[0]
R2_390 = 1 - ((_tg - _XT @ _b) ** 2).mean() / _tg.var()
del _XT
W390 = dict(zip(KEYS, _b[1:-1] / _b[:-1].sum()))             # weights on the same scale as the 0.95388 weights
O390, T390 = mix(W390, "oof"), mix(W390, "test")
RHO_REPLAY = np.corrcoef(rankdata(T390), rankdata(V390))[0, 1]
OOFAUC["390"], FA["390"] = roc_auc_score(y, O390), fold_aucs(O390)
_dt, _dr = L(V390) - L(S388), L(T390) - L(S388)                # the real step and the replayed step, on the test rows
STEP_R2 = 1 - ((_dt - _dr) ** 2).sum() / (_dt ** 2).sum(); del _dt, _dr
check("0.95390 file replayed from the 0.95370 file and the 87 public members (test rows, no labels)",
      f"R\u00b2 {R2_390:.5f}; rank correlation with the real file {RHO_REPLAY:.6f} (the step itself: {RHO390:.6f}); "
      f"the replay reproduces {STEP_R2:.0%} of the step's variance; replay OOF AUC {OOFAUC['390']:.6f}")

# The author's own printed numbers (his public notebook log). His bundle is not public, so only the four ladder
# files can be recomputed here; they match to the sixth decimal.
AUTH390 = {"0.95370 file": 0.952036, "careful": 0.952368, "balanced": 0.952274, "bold": 0.952412, "0.95388": 0.952455,
           "Sat B2": 0.952501, "twin (B2 + B3) / 2": 0.952531, "Sat B4 (published)": 0.952552}
AUTH390_FOLDS = {"0.95388": [0.951410, 0.952511, 0.957397, 0.952525, 0.948416], "Sat B2": [0.951468, 0.952546, 0.957435, 0.952591, 0.948447],
                 "twin (B2 + B3) / 2": [0.951510, 0.952573, 0.957456, 0.952633, 0.948462], "Sat B4 (published)": [0.951534, 0.952597, 0.957467, 0.952646, 0.948494]}
_ours = {"careful": OOFAUC["careful"], "balanced": OOFAUC["balanced"], "bold": OOFAUC["bold"], "0.95388": OOFAUC["bold + 0.5 (bold - balanced)"]}
_ours["0.95370 file"] = roc_auc_score(y, A_OOF)                   # his anchor = the replay without the 0.06 LM move
_dev = max(abs(_ours[k] - AUTH390[k]) for k in _ours)
check("Aman Atar's OOF AUCs of careful / balanced / bold / 0.95388 / 0.95370 file, recomputed from public files",
      " / ".join(f"{_ours[k]:.6f}" for k in _ours) + f" (his: " + " / ".join(f"{AUTH390[k]:.6f}" for k in _ours) + ")", _dev < 1.5e-6)

# Geometry in the author's probit space z = Phi^-1(u), on the test rows (his Experiment 1, recomputed from public files).
def Pz(v): return ndtri(np.clip((rankdata(v) - 0.5) / len(v), 1e-7, 1 - 1e-7))
ZT = {"careful": Pz(MT["careful"]), "balanced": Pz(MT["balanced"]), "bold": Pz(MT["bold"]), "0.95388": Pz(S388), "0.95390": Pz(V390)}
def sp_dist(a, b): return 1 - np.corrcoef(rankdata(ZT[a]), rankdata(ZT[b]))[0, 1]
_v1, _v2 = ZT["bold"] - ZT["balanced"], (ZT["bold"] - ZT["balanced"]) - (ZT["balanced"] - ZT["careful"])
COS12 = float(np.dot(_v1 - _v1.mean(), _v2 - _v2.mean()) / (np.linalg.norm(_v1 - _v1.mean()) * np.linalg.norm(_v2 - _v2.mean())))
E1_OURS = [sp_dist("careful", "balanced"), sp_dist("balanced", "bold"), sp_dist("careful", "bold"), sp_dist("bold", "0.95388")]
E1_HIS = [6.2214e-05, 1.5535e-04, 3.1796e-04, 3.8761e-05]
check("Aman Atar's Experiment 1 (ladder distances, cos of tangent and curvature), recomputed from public files",
      "distances " + ", ".join(f"{a:.4e}" for a in E1_OURS) + f"; cos {COS12:.4f} (his: " + ", ".join(f"{a:.4e}" for a in E1_HIS) + "; 0.7464)",
      max(abs(a / b - 1) for a, b in zip(E1_OURS, E1_HIS)) < 0.01 and abs(COS12 - 0.7464) < 0.002)
# 2-D map of the path: x along the tangent (bold - balanced), y the part of (balanced - careful) orthogonal to it.
_c = {k: v - ZT["balanced"] for k, v in ZT.items()}
_e1 = _v1 / np.linalg.norm(_v1)
_u = (ZT["balanced"] - ZT["careful"]); _u = _u - (_u @ _e1) * _e1; _e2 = _u / np.linalg.norm(_u)
_rms = lambda v: np.linalg.norm(v) / np.sqrt(NT) * 1e3
GEO = {k: (v @ _e1 / np.sqrt(NT) * 1e3, v @ _e2 / np.sqrt(NT) * 1e3, _rms(v - (v @ _e1) * _e1 - (v @ _e2) * _e2)) for k, v in _c.items()}
STEP_RMS = _rms(ZT["0.95390"] - ZT["0.95388"])
# How much the probit-vs-logit choice matters: a 50/50 blend of the 0.95388 and 0.95390 files built in both spaces.
_bz, _bl = 0.5 * ZT["0.95388"] + 0.5 * ZT["0.95390"], 0.5 * L(S388) + 0.5 * L(V390)
PROBIT_VS_LOGIT = 1 - np.corrcoef(rankdata(_bz), rankdata(_bl))[0, 1]
del _c, _u, _v1, _v2, _bz, _bl

# Where the step goes (label-free, on the test rows): Delta = replay - 0.95388 mix, regressed on the ladder tangent
# v1 = bold - balanced and the curvature v2 = (bold - balanced) - (balanced - careful); the same coefficients on OOF.
_D = T390 - MT["388"]; _V = np.column_stack([MT["bold"] - MT["balanced"], MT["bold"] - 2 * MT["balanced"] + MT["careful"]])
_ab = np.linalg.lstsq(_V - _V.mean(0), _D - _D.mean(), rcond=None)[0]
_r = (_D - _D.mean()) - (_V - _V.mean(0)) @ _ab
SHARE_PLANE = 1 - (_r @ _r) / ((_D - _D.mean()) @ (_D - _D.mean()))
del _D, _V, _r
_DO = O390 - MO["388"]
_V1O, _V2O = MO["bold"] - MO["balanced"], MO["bold"] - 2 * MO["balanced"] + MO["careful"]
PARTS390 = {"along the ladder (tangent)": _ab[0] * _V1O, "curvature (2nd order)": _ab[1] * _V2O,
            "off the ladder plane (re-weighting)": _DO - _ab[0] * _V1O - _ab[1] * _V2O, "the whole step": _DO}
_f388 = FA["bold + 0.5 (bold - balanced)"]
PART_GAIN = {n: (fold_aucs(MO["388"] + d) - _f388) * 1e5 for n, d in PARTS390.items()}
del _V1O, _V2O

# Nested step size along the step's direction: t chosen on four folds, scored on the fifth.
TGRID3 = np.round(np.arange(0, 3.01, 0.25), 2)
TG3 = np.array([[auc_sort(y[b], (MO["388"] + t * _DO)[b]) for b in FOLDS] for t in TGRID3]) * 1e5
TG3 = TG3 - TG3[0]
T3_PICK, T3_OUT = np.zeros(5), np.zeros(5)
for f in range(5):
    j = TG3[:, [g for g in range(5) if g != f]].mean(1).argmax(); T3_PICK[f], T3_OUT[f] = TGRID3[j], TG3[j, f]
del _DO

# Nested re-weighting of the same 87 members on top of the 0.95388 file: a ridge logistic model with the 0.95388 OOF
# as an input (its weight free), fitted on four folds and scored on the fifth, for three fixed penalties (all shown).
_X = np.column_stack([QM_O[k] - REF_OOF for k in KEYS]).astype(np.float32)
_X = (_X - _X.mean(0)) / _X.std(0)
_B = (MO["388"] - MO["388"].mean()) / MO["388"].std()
_yf = y.astype(np.float64)
def _ridge_fit(rows, lam):
    Xr, br, yr = _X[rows], _B[rows], _yf[rows]; n = len(rows)
    def f(p):
        z = p[0] + p[1] * br + Xr @ p[2:].astype(np.float32)
        r = (expit(z) - yr) / n
        return (np.logaddexp(0, z) - yr * z).mean() + lam * p[2:] @ p[2:], np.r_[r.sum(), r @ br, (Xr.T @ r.astype(np.float32)).astype(np.float64) + 2 * lam * p[2:]]
    p0 = np.zeros(2 + _X.shape[1]); p0[0] = np.log(yr.mean() / (1 - yr.mean())); p0[1] = 1.0
    return minimize(f, p0, jac=True, method="L-BFGS-B", options=dict(maxiter=300)).x
LAMS = [3e-2, 1e-2, 3e-3]
RIDGE_OUT = np.zeros((len(LAMS), 5))
for i, lam in enumerate(LAMS):
    for f, b in enumerate(FOLDS):
        p = _ridge_fit(np.setdiff1d(np.arange(N), b), lam)
        RIDGE_OUT[i, f] = (auc_sort(y[b], p[1] * _B[b] + _X[b] @ p[2:].astype(np.float32)) - auc_sort(y[b], MO["388"][b])) * 1e5
del _X, _B, _yf

# Expected private change of the public +2: no selection (K = 1), and the best of the three files the author reports
# at 0.95390 (Sat B2, B3, B4; K = 3, the textbook give-back of section 4, which assumes probes of equal true value).
_wide = np.sqrt(SD390 ** 2 + (SD390 * np.sqrt(NPUB / (NT - NPUB))) ** 2) * 1.645
GIVE390 = SD390 * EMAX[2]
EXP390 = {"K1": (2, 2 - _wide, 2 + _wide), "K3": (2 - GIVE390, 2 - GIVE390 - _wide, 2 - GIVE390 + _wide)}
log(f"0.95390 file: step noise SD {SD390:.1f}e-5; replay OOF change {(OOFAUC['390'] - OOFAUC['bold + 0.5 (bold - balanced)']) * 1e5:+.1f}e-5")
# Leaderboard snapshot 14 Nov 2026 10:06 UTC (scores >= 0.95200 as points above 0.95, and submission counts; no names).
LB14B_TOP = np.array([int(v) for v in """784,604,568,442,437,432,408,404,403,400,394,394,393,392,391,390,390,389,389,389,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,388,387,387,386,386,386,386,386,386,386,385,384,383,383,382,376,372,372,371,370,370,370,370,370,370,370,370,370,370,370,370,370,370,370,370,369,367,364,364,364,363,361,361,361,358,357,357,356,356,355,355,355,355,355,354,354,354,354,354,354,354,354,354,354,354,354,354,354,354,354,353,352,351,351,351,351,351,350,350,349,349,349,348,348,348,347,347,347,347,347,347,347,347,347,347,347,347,347,347,347,346,346,346,346,346,346,346,346,346,346,346,346,345,345,345,345,344,344,343,343,343,343,342,341,340,340,340,339,339,337,337,333,333,333,331,331,331,331,331,331,331,331,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,330,326,325,323,322,322,322,319,317,316,315,313,313,311,311,310,310,310,309,309,309,308,308,307,307,307,306,306,305,303,303,302,302,302,301,300,300,300,300,300,300,300,300,300,300,300,300,300,300,299,298,298,298,298,297,296,295,295,295,295,294,294,293,291,291,291,290,290,288,288,288,287,286,286,285,285,285,285,285,285,285,285,285,285,285,285,285,285,285,285,284,280,278,277,277,277,276,275,275,275,274,274,274,273,272,271,268,268,268,267,267,267,267,266,266,265,265,262,262,262,262,260,258,258,258,258,257,256,256,255,255,255,254,252,251,250,250,249,249,247,247,246,246,245,245,244,243,243,242,242,241,241,240,240,239,239,239,239,239,239,239,239,239,239,239,239,239,238,238,238,238,238,237,237,237,236,235,235,234,234,233,233,231,231,231,230,230,229,229,228,226,225,224,224,224,224,222,222,222,222,222,222,219,218,217,217,216,216,213,213,212,212,211,210,209,209,209,208,206,205,205,203,203,203,202,201,201,200,200,200,200""".split(",")])
LB14B_SUBS = np.array([int(v) for v in """131,34,122,61,10,42,117,31,30,117,81,41,32,102,120,114,95,64,82,111,41,90,12,42,9,19,111,108,26,113,95,18,17,2,12,11,93,7,112,7,138,28,34,1,3,8,23,3,5,7,9,24,28,74,67,13,86,18,3,19,3,30,22,12,13,15,15,27,24,30,24,16,104,12,8,1,49,18,6,50,10,54,24,140,68,2,30,21,18,30,65,62,1,21,9,4,1,13,1,84,11,1,20,67,19,20,33,18,10,55,71,14,46,48,66,57,6,17,22,6,16,30,12,20,7,11,12,5,2,26,21,25,3,1,3,14,9,6,15,2,7,84,18,11,19,7,22,10,6,10,39,13,90,10,4,30,7,8,5,11,2,7,15,24,9,4,15,36,15,6,8,7,9,27,16,1,14,8,1,21,12,8,7,5,15,13,43,8,12,42,15,12,6,31,10,13,11,30,10,23,19,78,6,8,8,16,5,8,10,11,7,2,7,7,5,15,27,21,2,13,8,2,6,13,8,9,5,11,10,10,3,3,7,7,5,10,33,5,4,14,13,18,13,18,7,10,7,28,28,13,16,51,2,2,36,9,30,87,6,3,24,11,17,38,21,13,51,36,47,6,25,16,5,3,4,4,2,1,19,10,2,3,28,16,1,32,2,9,5,10,77,30,16,6,7,5,21,31,40,3,28,39,4,34,10,5,26,10,2,9,11,3,2,10,7,12,5,2,3,9,1,3,14,4,82,16,19,4,16,58,5,3,4,116,140,13,16,24,92,86,29,13,4,3,5,1,7,1,5,9,5,5,18,14,1,14,3,9,4,12,13,5,1,8,6,19,6,3,28,5,9,20,38,1,6,5,33,5,9,5,5,25,12,10,17,5,1,59,6,4,6,12,2,16,2,9,4,36,65,31,16,2,3,9,9,25,6,15,6,19,5,5,10,9,59,12,24,5,10,17,4,30,8,32,7,33,24,39,5,1,8,1,1,16,9,3,14,4,7,7,8,9,2,19,3,4,1,3,11,3,4,3,25,5,13,7,6,30,31,6,41,4,7,7""".split(",")])
LB14B_TEAMS = 1756
````

````python
# Step anatomy: public change, its noise, OOF change (in-sample and the author's nested), our nested gate, and two readings
# of the expected private change: the board-only selection model of sections 4 and 9, and the OOF reading.
RT7 = {r: rankdata(v) for r, v in ST7.items()}
def rho7(a, b): return np.corrcoef(RT7[a], RT7[b])[0, 1]
SD7 = {r: sig(rho7(r, FROM7[r])) for r in STEPS7}                    # public noise SD of each step (one-line rule)
PRIV_F, OOF_F = np.sqrt(NPUB / (NT - NPUB)), np.sqrt(NPUB / N)        # noise scale of the same gap on 240k private / 700k OOF rows
def iv_board(sd): return 1.645 * np.sqrt(sd ** 2 + (sd * PRIV_F) ** 2)
def iv_oof(sd): return 1.645 * sd * np.sqrt(PRIV_F ** 2 + OOF_F ** 2)
# Our nested gate on each step: the step size t in base + t (step - base) chosen on four folds, scored on the fifth.
TGRID7 = np.round(np.arange(0, 2.01, 0.25), 2)
NEST7 = {}
for r in STEPS7:
    b_ = L(SO7[FROM7[r]]); d_ = SO7[r] - b_
    G_ = np.array([[auc_sort(y[f], (b_ + t * d_)[f]) for f in FOLDS] for t in TGRID7]) * 1e5
    G_ = G_ - G_[0]; pick, out = np.zeros(5), np.zeros(5)
    for f in range(5):
        j = G_[:, [g for g in range(5) if g != f]].mean(1).argmax(); pick[f], out[f] = TGRID7[j], G_[j, f]
    NEST7[r] = dict(pick=pick, out=out, take=int((pick > 0).sum()))
del b_, d_, G_
ROWS7 = []
for r in STEPS7:
    pub = LB7[r] - LB7[FROM7[r]]; sd = SD7[r]; K = KEPT7.get(r, 1)
    gb = sd * EMAX[K - 1]; up = int((FOLD7[r] > 0).sum()); nest = AUTH_NESTED7[r]
    if nest > 0 and up >= 4: verdict = "real"
    elif nest < 0 and up <= 1: verdict = "public-fit" if pub > 0 else "public-fit, missed"
    else: verdict = "noise"
    ROWS7.append(dict(step=r, led=LED7[r], pub=pub, fcst=AUTH_FCST7[r], sd=sd, ins=INS7[r], up=up, nested=nest, K=K, give=gb,
                      board=pub - gb, board_lo=pub - gb - iv_board(sd), board_hi=pub - gb + iv_board(sd),
                      exp=nest, lo=nest - iv_oof(sd), hi=nest + iv_oof(sd), takes=NEST7[r]["take"], verdict=verdict))
T7 = pd.DataFrame(ROWS7).set_index("step")

# Whole files against the 0.95388 file: the public change, the OOF change (in-sample, exact moves) and the author's nested
# changes added up along the path each file was built through.
def path7(r):
    p = [];
    while r != "388": p.append(r); r = FROM7[r]
    return p[::-1]
FILES7 = [("0.95388 file (Areeb Ali)", "388"), ("0.95390 file (Aman Atar)", "390"), ("CV-led step: submission_cv.csv", "cv_led"),
          ("step 6", "step6"), ("0.95417 file: submission.csv", "step7")]
RT7["390"] = rankdata(V390)
FROWS = []
for lab, r in FILES7:
    if r == "388": ins, nst, sd = 0.0, 0.0, 0.0
    elif r == "390": ins, nst, sd = (OOFAUC["390"] - AUC7["388"]) * 1e5, np.nan, sig(rho7("390", "388"))
    else: ins, nst, sd = (AUC7[r] - AUC7["388"]) * 1e5, sum(AUTH_NESTED7[s] for s in path7(r)), sig(rho7(r, "388"))
    pub = (95390 if r == "390" else LB7[r]) - 95388
    gb = sum(SD7[s] * EMAX[KEPT7[s] - 1] for s in path7(r) if s in KEPT7) if r not in ("388", "390") else (GIVE390 if r == "390" else 0.0)
    cen = nst if nst == nst else ins
    FROWS.append(dict(file=lab, key=r, pub=pub, ins=ins, nested=nst, sd=sd, exp=cen, lo=cen - iv_oof(sd), hi=cen + iv_oof(sd),
                      board=pub - gb, board_lo=pub - gb - iv_board(sd), board_hi=pub - gb + iv_board(sd), give=gb))
F7 = pd.DataFrame(FROWS).set_index("key")
# The pair that matters for a final pick: the 0.95417 file against the CV-led file.
SD_PAIR = sig(rho7("step7", "cv_led"))
PAIR_INS = (AUC7["step7"] - AUC7["cv_led"]) * 1e5
PAIR_NEST = F7.loc["step7", "nested"] - F7.loc["cv_led", "nested"]
PAIR_IV = iv_oof(SD_PAIR)
RHO_PAIR = rho7("step7", "cv_led")
log(f"step anatomy done; 0.95417 vs CV-led file: OOF {PAIR_INS:+.1f} in-sample, {PAIR_NEST:+.1f} nested (sum)")
````

````python
# STEERING BY A MODEL OF THE PUBLIC BOARD, simulated on the OOF rows (general logic, first order, no new model).
# Each split draws three disjoint parts of the OOF rows: a public part (60k rows), a private part (240k) and a training
# part (240k). On each part, the first-order response of the AUC of the 0.95388 file to a move towards each of the same
# 87 public members is measured (finite difference). A "scored file" is a random combination of those moves; its score
# change is one exact linear equation (the toy example). A model fitted on m scored files recovers the part of the
# response that the m files span, and one step is taken along it:
#   * steered by the public part: the files are scored on the public rows (what a public-board model sees);
#   * steered by CV: the same files scored on the training rows instead.
# The step is read on the public and on the private part. To first order every number scales with the step size; it
# is set so that the public-steered step gains, with 69 files, as much as the kept forecast-led steps of the ladder did.
B7 = L(SO7["388"])
DM = np.column_stack([QM_O[k] - B7 for k in KEYS]).astype(np.float32)
NSPLIT, EPS7, MGRID = int(os.environ.get("NB_STEER_SPLITS", 12)), 0.05, [5, 10, 20, 40, 69, 87]
UNIT = {m: np.zeros((NSPLIT, 2, 2)) for m in MGRID}               # (split, steered by public/CV, read on public/private), per unit step
COS_PQ, COS_TQ = np.zeros(NSPLIT), np.zeros(NSPLIT)
rng = np.random.default_rng(2715)
for s_ in range(NSPLIT):
    perm = rng.permutation(N); parts = [perm[:NPUB], perm[NPUB:NT], perm[NT:NT + (NT - NPUB)]]
    G = np.zeros((3, len(KEYS)))
    for a, I in enumerate(parts):
        yI, BI = y[I], B7[I]; b0 = auc_sort(yI, BI)
        for k in range(len(KEYS)): G[a, k] = (auc_sort(yI, BI + EPS7 * DM[I, k]) - b0) / EPS7 * 1e5
    COS_PQ[s_] = G[0] @ G[1] / np.linalg.norm(G[0]) / np.linalg.norm(G[1]); COS_TQ[s_] = G[2] @ G[1] / np.linalg.norm(G[2]) / np.linalg.norm(G[1])
    X = rng.normal(0, 1, (len(KEYS), len(KEYS)))                    # the scored files: random combinations of the 87 moves
    for m in MGRID:
        for a, src in enumerate((0, 2)):
            Xm = X[:m]; g = Xm.T @ np.linalg.lstsq(Xm @ Xm.T, Xm @ G[src], rcond=None)[0]; u = g / np.linalg.norm(g)
            UNIT[m][s_, a] = [G[0] @ u, G[1] @ u]
del DM
_kfc = T7[(T7.led == "forecast") & T7.index.isin(list(KEPT7))]
STEER_TARGET = float(_kfc.pub.mean())                             # mean public gain of the kept forecast-led steps
STEER_SCALE = STEER_TARGET / UNIT[69][:, 0, 0].mean()
STEER = {m: UNIT[m] * STEER_SCALE for m in MGRID}
ST_MEAN = {m: STEER[m].mean(0) for m in MGRID}
ST_SE = {m: STEER[m].std(0) / np.sqrt(NSPLIT) for m in MGRID}
REAL_SHARE = ST_MEAN[69][0, 1] / ST_MEAN[69][0, 0]
log(f"steering simulation: {NSPLIT} splits; with 69 files the public-steered step reads {ST_MEAN[69][0, 0]:+.1f} public, "
    f"{ST_MEAN[69][0, 1]:+.1f} private; CV-steered {ST_MEAN[69][1, 0]:+.1f} / {ST_MEAN[69][1, 1]:+.1f}")
````

````python
# The toy example of section v7: 10 rows, two scored files, and the forecast of a third file, all computed exactly.
TOY_Y = np.array([0, 0, 1, 0, 0, 1, 0, 1, 0, 1])                    # hidden labels, rows in the base file's order (lowest score first)
TOY_BASE = np.arange(10, dtype=float)
def toy_swap(s, i, j): s = s.copy(); s[i], s[j] = s[j], s[i]; return s
TOY_F = {"file 1": toy_swap(TOY_BASE, 2, 3), "file 2": toy_swap(TOY_BASE, 7, 8)}
TOY_F["file 3 = both moves"] = toy_swap(toy_swap(TOY_BASE, 2, 3), 7, 8)
TOY_AUC0 = roc_auc_score(TOY_Y, TOY_BASE)
TOY_D = {k: roc_auc_score(TOY_Y, v) - TOY_AUC0 for k, v in TOY_F.items()}
TOY_PN = int(TOY_Y.sum() * (1 - TOY_Y).sum())
assert abs(TOY_D["file 3 = both moves"] - TOY_D["file 1"] - TOY_D["file 2"]) < 1e-12
````

<a id="s15"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">New in v7 &middot; Steering a ladder by a model of the public board</div>

Thank you to **goodpjw2008** for
[S6E11 LM Ladder | Forecast the Board | LB 0.95417](https://www.kaggle.com/code/goodpjw2008/s6e11-lm-ladder-forecast-the-board-lb-0-95417)
(15 Nov), the new top public notebook. It is unusually open: it publishes the weights of every step, the public score
**and** a forecast written down before each submission, its own nested CV for every step, and the files that lost. It
also states the limits that this section measures: the 0.95417 file is tuned on the public rows, and the author would
take the CV-led file for the private rows. This section is built on that openness, and it does not single anyone out:
probing the public board is how every ladder near the top was built, including the ones analysed in earlier versions.

**What the notebook does, in plain words.** It starts from the 0.95388 file (Areeb Ali, section 9) and takes ten
small steps, each a re-weighting of the same 87 public members. Each step is chosen to maximise
`OOF change + mu x forecast of the public change`. The forecast comes from a *public model*: a model fitted on the
public scores of 69 files the author had scored. Of the 10 steps, 7 were led by the forecast, 2 by CV and 1 by both.

Words used below. **OOF** (out-of-fold): predictions for training rows from models that did not see those rows; the
best stand-in for a test set. **Nested CV**: every choice made on four folds and scored on the fifth, so the score
never sees the rows that made the choice. **In-sample**: the choice saw all five folds (optimistic). **Noise SD**: how
much a score difference moves by chance between random 60,000-row public parts. **Winner's curse**: keeping the best of
several noisy scores makes the kept score look better than it is.

### Each scored file is one linear equation about the public labels

AUC only counts pairs: how often a fraud is ranked above a legitimate row. Move one row up past its neighbour and the
score changes by exactly `(label of the row that moved up - label of the row it passed) / (frauds x legit rows)`. So the
public score of a file is a **known** combination of **unknown** public labels: one linear equation. The toy below has
10 rows (4 frauds, 6 legitimate, so 24 fraud-legit pairs) and two scored files.

````python
fig, ax = plt.subplots(figsize=(13, 4.9))
ax.set_axis_off(); ax.set_xlim(-3.4, 15.2); ax.set_ylim(-1.55, 4.1)
_rows = [("base file", TOY_BASE), ("file 1", TOY_F["file 1"]), ("file 2", TOY_F["file 2"]), ("file 3 = both moves", TOY_F["file 3 = both moves"])]
_swaps = {"file 1": {2, 3}, "file 2": {7, 8}, "file 3 = both moves": {2, 3, 7, 8}}
for yi, (lab, s) in zip([3.3, 2.3, 1.3, 0.3], _rows):
    order = np.argsort(s)
    ax.text(-0.15, yi, lab, ha="right", va="center", fontsize=12, color=INK, fontweight="bold" if lab.startswith("file 3") else "normal")
    for pos, row in enumerate(order):
        moved = row in _swaps.get(lab, set())
        ax.add_patch(FancyBboxPatch((pos + 0.06, yi - 0.32), 0.88, 0.64, boxstyle="round,pad=0,rounding_size=0.08",
                                    facecolor="#fde7d9" if moved else SURF, edgecolor=PRIV if moved else RULE, lw=1.6 if moved else 1))
        ax.text(pos + 0.5, yi + 0.06, f"r{row + 1}", ha="center", va="center", fontsize=11, color=INK)
        ax.text(pos + 0.5, yi - 0.2, "?", ha="center", va="center", fontsize=9.5, color=MUTED)
    if lab != "base file":
        d = TOY_D[lab] * TOY_PN
        txt = (f"scored: {d:+.0f}/{TOY_PN}" if lab != "file 3 = both moves" else
               f"forecast {TOY_D['file 1'] * TOY_PN:+.0f}/{TOY_PN} {TOY_D['file 2'] * TOY_PN:+.0f}/{TOY_PN} = {(TOY_D['file 1'] + TOY_D['file 2']) * TOY_PN:+.0f}/{TOY_PN};  scored: {d:+.0f}/{TOY_PN}")
        ax.text(10.25, yi, txt, ha="left", va="center", fontsize=11.5, color=CVTXT if lab.startswith("file 3") else INK,
                fontweight="bold" if lab.startswith("file 3") else "normal")
ax.annotate("", xy=(10.0, 3.75), xytext=(0.0, 3.75), arrowprops=dict(arrowstyle="->", color=MUTED, lw=1.2))
ax.text(5.0, 3.88, "rows in score order: lowest score left, highest right", ha="center", va="bottom", fontsize=10.5, color=INK2)
ax.text(0.0, -0.55, "hidden labels (shown only to check the toy):  " + "   ".join(f"r{i + 1}={v}" for i, v in enumerate(TOY_Y)),
        ha="left", va="center", fontsize=10.5, color=INK2)
ax.text(0.0, -0.95, f"file 1 tells:  y(r3) - y(r4) = {TOY_D['file 1'] * TOY_PN:+.0f}        file 2 tells:  y(r8) - y(r9) = {TOY_D['file 2'] * TOY_PN:+.0f}",
        ha="left", va="center", fontsize=11, color=INK)
ax.text(0.0, -1.32, "two equations, ten unknowns: the labels stay unknown, but any file built from the same moves is forecast exactly",
        ha="left", va="center", fontsize=11, color=INK)
headline(fig, "Two scored files are two equations: they forecast a third file exactly, without knowing a single label",
         "Toy example, 10 rows; orange boxes = rows that moved. The real board has 60,000 rows and a top file has been scored dozens of times")
plt.show()
````

**Takeaway:** a file built from moves that were already scored can be forecast without any label, and a file built
against that forecast gains on the public rows by construction.

<div style="font-size: 12.5px; color: #1a1a1a; background: #f3f5f9; border-left: 4px solid #8a8984; padding: 8px 12px; border-radius: 6px;">
<b style="color: #1a1a1a;">The same in one formula (the author's).</b> To first order, for a file whose rank fractions are <code style="color: #1a1a1a;">u_file</code> and a reference file with <code style="color: #1a1a1a;">u</code>,
<code style="color: #1a1a1a;">AUC(file) - AUC(ref) = mean over public rows of (y - p(u)) (u_file - u) / (pi (1 - pi))</code>, where <code style="color: #1a1a1a;">p(u)</code> is the fraud rate at rank <code style="color: #1a1a1a;">u</code> and
<code style="color: #1a1a1a;">pi</code> the fraud rate. The only unknowns are the 60,000 numbers <code style="color: #1a1a1a;">y - p</code>. Each scored file adds one equation; 69 equations
cannot identify 60,000 unknowns, but they do identify the score of any file whose move lies in the span of the scored moves.
A step taken along the direction those equations favour raises the public score whether or not it raises the private one.
</div>

### Every step of the 0.95417 ladder: public score against OOF

One point per file. The x-axis is the public score; the y-axis is the OOF change from the 0.95388 file, as the
**author's nested CV** (filled; added up along the path a file was built through) and as our **in-sample** recomputation
from the public files (hollow; it reproduces the author's in-sample numbers to 0.05 points).

````python
_P7 = ["bold", "388", "joint", "cv_led", "step1", "step2_half", "step2", "step3", "step4", "step5", "step6", "step7"]
_COL7 = {"forecast": C[6], "CV": CV, "forecast + CV": C[3]}
def _nest_cum(r):
    if r in ("388", "bold"): return np.nan if r == "bold" else 0.0
    return sum(AUTH_NESTED7[s] for s in path7(r))
fig, ax = plt.subplots(figsize=(13, 7.4))
XY7 = {}
for r in _P7:
    x = LB7[r] - 95000; yi = (AUC7[r] - AUC7["388"]) * 1e5; yn = _nest_cum(r)
    col = _COL7.get(LED7.get(r), NAVY); XY7[r] = (x, yn if yn == yn else yi)
    if yn == yn and r != "388":
        ax.plot([x, x], [yi, yn], color=col, lw=1.2, alpha=0.5, zorder=2)
        ax.plot(x, yn, "o", ms=12 if r == "step7" else 10, color=col, mec=INK if r == "step7" else SURF, mew=1.6, zorder=5)
    ax.plot(x, yi, "o", ms=8, mfc=SURF, mec=col, mew=1.8, zorder=4)
x390, y390 = 390, (OOFAUC["390"] - AUC7["388"]) * 1e5
ax.plot(x390, y390, "s", ms=8, mfc=SURF, mec=PRIV, mew=1.8, zorder=4)
ax.plot(388, 0, "o", ms=10, color=NAVY, mec=SURF, mew=1.5, zorder=5)
for a, b in [("bold", "388"), ("388", "step1"), ("step1", "step2"), ("step2", "step3"), ("step3", "step4"), ("step4", "step6"), ("step6", "step7")]:
    ax.annotate("", xy=XY7[b], xytext=XY7[a], arrowprops=dict(arrowstyle="-|>", color=INK2, lw=1.3, mutation_scale=13, shrinkA=7, shrinkB=7), zorder=3)
for a, b in [("388", "joint"), ("388", "cv_led"), ("step1", "step2_half"), ("step4", "step5")]:
    ax.annotate("", xy=XY7[b], xytext=XY7[a], arrowprops=dict(arrowstyle="-", color=MUTED, lw=1, ls=(0, (2, 2)), shrinkA=7, shrinkB=7), zorder=3)
_OFF7 = {"bold": ((-10, -2), "right", "v4 bold mix\n(0.95386)"), "388": ((6, -20), "left", "0.95388 file"),
         "joint": ((12, 2), "left", "joint"), "cv_led": ((-12, 8), "right", "CV-led step\nsubmission_cv.csv"),
         "step1": ((12, 10), "left", "step 1"), "step2_half": ((12, 0), "left", "step 2, half"), "step2": ((12, 0), "left", "step 2"),
         "step3": ((12, 0), "left", "step 3"), "step4": ((-12, -4), "right", "step 4"), "step5": ((12, 0), "left", "step 5 (not kept)"),
         "step6": ((12, 6), "left", "step 6 (CV-led)"), "step7": ((14, 0), "left", "step 7 = 0.95417\nsubmission.csv")}
for r, (off, ha, t_) in _OFF7.items():
    ax.annotate(t_, XY7[r], xytext=off, textcoords="offset points", ha=ha, va="center", fontsize=11, color=INK,
                fontweight="bold" if r in ("step7", "cv_led") else "normal")
ax.annotate("0.95390 file (Aman Atar)\nreplay, in-sample", (x390, y390), xytext=(14, 18), textcoords="offset points", ha="left", va="bottom",
            fontsize=10.5, color=PRIV, arrowprops=dict(arrowstyle="-", color=PRIV, lw=0.8))
ax.text(0.985, 0.97, "led by the forecast", transform=ax.transAxes, ha="right", va="top", fontsize=12, color=C[6], fontweight="bold")
ax.text(0.985, 0.92, "led by CV", transform=ax.transAxes, ha="right", va="top", fontsize=12, color=CVTXT, fontweight="bold")
ax.text(0.985, 0.87, "filled = author's nested CV; hollow = in-sample", transform=ax.transAxes, ha="right", va="top", fontsize=10.5, color=INK2)
zero_line(ax)
ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"0.95{int(v):03d}")); ax.yaxis.set_major_formatter(E5)
ax.set_xlim(383.0, 424); ax.set_ylim(None, 22); ax.set_xlabel("public LB score"); ax.set_ylabel("OOF AUC change from the 0.95388 file (x 1e-5)")
headline(fig, f"Up the board, down on OOF: from the 0.95388 file, the 0.95417 file is {LB7['step7'] - LB7['388']:+d} on public and {_nest_cum('step7'):+.0f} on nested OOF",
         "Every scored file of the 0.95417 ladder (goodpjw2008); arrows = the path to the published file; dotted = files that were not kept")
plt.show()
````

### How far can steering by the public board go? A simulation on the OOF rows

The mechanism can be measured without any public score. On each of 12 random splits, the OOF rows are cut
into a public part (60,000 rows), a private part (240,000) and a training part (240,000). On each part we measure how
the AUC of the 0.95388 file responds, to first order, to a move towards each of the same 87 members. A "scored file"
is a random combination of those moves, and its score change is one exact linear equation, as in the toy. A model
fitted on m scored files recovers the part of the response those files span; one step is then taken along it. The
model sees the files' scores on the **public** part (a public model) or on the **training** part (CV), and the step
is read on the **private** part. To first order every number scales with the step size; it is set so that the
public-steered step with 69 files gains on public what the ladder's kept forecast-led steps gained on average (+5.0 points).

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 5.4), gridspec_kw=dict(width_ratios=[1.45, 1]))
mm = np.array(MGRID)
for (a, b), col, ls, lab in [((0, 0), PUB, "-", "steered by public: read on public"), ((0, 1), PRIV, "-", "steered by public: read on private"),
                             ((1, 1), CV, "-", "steered by CV: read on private"), ((1, 0), CV, (0, (3, 2)), "steered by CV: read on public")]:
    mu = np.array([ST_MEAN[m][a, b] for m in MGRID]); se = np.array([ST_SE[m][a, b] for m in MGRID])
    ax.fill_between(mm, mu - 1.645 * se, mu + 1.645 * se, color=col, alpha=0.12, lw=0)
    ax.plot(mm, mu, ls=ls, color=col, lw=2.4, marker="o", ms=6, mec=SURF)
    dy = {(0, 0): 0, (0, 1): -11, (1, 1): 3, (1, 0): 10}[(a, b)]
    ax.annotate(lab, (mm[-1], mu[-1]), xytext=(8, dy), textcoords="offset points", fontsize=11, color=CVTXT if col == CV else col, va="center")
zero_line(ax); ax.set_xscale("log"); ax.set_xticks(MGRID); ax.set_xticklabels([str(m) for m in MGRID]); ax.set_xlim(4, 420)
ax.axvline(N_EQ7, color=MUTED, lw=1, ls=(0, (3, 3))); ax.text(N_EQ7 * 0.95, ax.get_ylim()[1] * 0.97, f"{N_EQ7} files:\nthe 0.95417 ladder", ha="right", va="top", fontsize=10.5, color=INK2)
ax.yaxis.set_major_formatter(E5); ax.set_xlabel("number of scored files the model is fitted on"); ax.set_ylabel("change of the steered step (x 1e-5)")
ax.grid(axis="x", visible=False)
panel_title(ax, "More scored files, more public points; the private part grows far less", f"mean of {NSPLIT} splits, band = 90% interval of the mean")
share = np.array([ST_MEAN[m][0, 1] / ST_MEAN[m][0, 0] for m in MGRID]); share_cv = np.array([ST_MEAN[m][1, 1] / ST_MEAN[m][1, 0] for m in MGRID])
xx = np.arange(len(MGRID))
ax2.bar(xx - 0.19, 100 * share, color=PRIV, width=0.36, zorder=3, label="steered by public")
ax2.bar(xx + 0.19, 100 * share_cv, color=CV, width=0.36, zorder=3, label="steered by CV")
for i, s in enumerate(share): ax2.text(i - 0.19, max(100 * s, 0) + 1.5, f"{100 * s:+.0f}%", ha="center", va="bottom", fontsize=10.5, fontweight="bold", color=INK)
ax2.set_xticks(xx); ax2.set_xticklabels([str(m) for m in MGRID]); ax2.set_xlabel("scored files")
ax2.set_ylabel("private change / public change (%)"); ax2.grid(axis="x", visible=False); zero_line(ax2); ax2.legend(loc="upper right", fontsize=10.5)
panel_title(ax2, "How much of the public gain is real", "the share the private rows repeat")
headline(fig, f"A model of the public board turns scored files into public points: with {N_EQ7} files the private rows repeat {SHARE_TXT} of the steered public gain",
         "First-order simulation on the OOF rows of the public files: random 60k / 240k / 240k parts, the 87 members of the 0.95417 ladder")
plt.show()
````

**Takeaway:** with 69 scored files a public-steered step gains +5.0 on public and 0.0 on private (the private rows repeat none of the public gain); a CV-steered step of the same size reads +0.8 on public and +0.3 on private.

**How far can it go?** (general logic only.) The public part has about 1,000 frauds. Its labels are fixed, so each
scored file is a fresh, exact equation about the same unknowns, and the public gain of the best steered file keeps
rising while the scored files span new directions. Two things limit it: the number of submissions (each equation
costs one) and the number of directions the files can move in. The private part is 240,000 *other* rows; nothing in
these equations is about them, so the fitted part of the gain is worth zero there in expectation. What a steered step
keeps on private is its real change, which the OOF rows estimate. This first-order picture also leaves out the cost of
large moves (curvature), which is where the OOF losses of the real steps come from.

### What the files are likely to score on private

````python
fig, ax = plt.subplots(figsize=(13, 5.6))
yy = np.arange(len(F7))[::-1]
for yi, r in zip(yy, F7.itertuples()):
    if r.Index != "388":
        ax.plot([r.board_lo, r.board_hi], [yi + 0.2, yi + 0.2], color=MUTED, lw=3, alpha=0.45, solid_capstyle="round", zorder=2)
        ax.plot(r.board, yi + 0.2, "o", ms=8, mfc=SURF, mec=MUTED, mew=1.8, zorder=4)
        ax.plot([r.lo, r.hi], [yi - 0.12, yi - 0.12], color=PRIV, lw=3, alpha=0.5, solid_capstyle="round", zorder=2)
    ax.plot(r.exp, yi - 0.12, "o", ms=10, color=PRIV, mec=SURF, mew=1.4, zorder=5)
    ax.plot(r.pub, yi + 0.2, "o", ms=9, color=PUB, mec=SURF, mew=1.4, zorder=5)
    if r.Index not in ("388", "390"): ax.plot(r.ins, yi - 0.12, "D", ms=7, mfc=SURF, mec=CVTXT, mew=1.6, zorder=4)
    ax.text(1.01, yi, "reference" if r.Index == "388" else f"{r.exp:+.0f}  [{r.lo:+.0f}, {r.hi:+.0f}]", transform=ax.get_yaxis_transform(),
            ha="left", va="center", fontsize=11.5, color=INK, fontweight="bold" if r.Index in ("step7", "cv_led") else "normal")
ax.text(1.01, len(F7) - 0.35, "expected private\n[90% interval]", transform=ax.get_yaxis_transform(), ha="left", va="bottom", fontsize=10.5, color=INK2)
ax.set_yticks(yy); ax.set_yticklabels(F7.file, fontsize=11.5); ax.tick_params(axis="y", length=0)
zero_line(ax, "x"); ax.xaxis.set_major_formatter(E5); ax.grid(axis="y", visible=False)
ax.set_xlabel("change from the 0.95388 file (x 1e-5)")
ax.legend(handles=[Line2D([], [], marker="o", ls="", color=PUB, ms=8, label="public change"),
                   Line2D([], [], marker="o", ls="-", mfc=SURF, mec=MUTED, color=MUTED, ms=8, label="board-only reading (public minus give-back)"),
                   Line2D([], [], marker="o", ls="-", color=PRIV, ms=8, label="expected private: OOF reading"),
                   Line2D([], [], marker="D", ls="", mfc=SURF, mec=CVTXT, ms=7, label="OOF change, in-sample")],
          loc="upper center", bbox_to_anchor=(0.45, -0.15), ncol=2, fontsize=10.5)
headline(fig, f"Expected private change: the 0.95417 file {F7.loc['step7', 'exp']:+.0f}, the CV-led file {F7.loc['cv_led', 'exp']:+.0f} (points from the 0.95388 file)",
         "OOF reading: the author's nested CV added up along each file's path (0.95390: our in-sample replay); 90% interval from private and OOF noise")
plt.show()
````

**Takeaway:** on OOF the CV-led file is expected at +7 and the 0.95417 file at -70 points from the 0.95388 file; the board-only reading ranks them the other way (-1 and +19).

**Two readings of the same file.** The *board-only* reading (grey) takes the public change and subtracts the
winner's-curse give-back of keeping the best of the files compared at each stage. It assumes that the probes were
random, so it is right for a ladder steered by luck and badly wrong for a ladder steered by a public model, where the
public gain was fitted on purpose. The *OOF* reading (orange) ignores the public score and uses the out-of-fold change,
which nothing in the steering has seen except through the CV term. For the forecast-led steps the two readings point
in opposite directions; the simulation above says which one to trust.

<a id="s16"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">New in v11 &middot; A file steered by the public board, shown openly</div>

The previous section showed how a model of the public board works. Each scored file is one linear equation about the
hidden public labels. Fitted on enough scored files, the model forecasts the public score of a new file before it is
submitted. This section uses that tool openly, on top of the best public file. Two things are shown side by side:
what the board is expected to say, and what the OOF rows say.

**The starting point** is Areeb Ali's
[Two Families, One Blend](https://www.kaggle.com/code/areebbb/s6e11-two-families-one-blend-lb-0-95421) (public 0.95421).
It mixes, in logit-rank space:
- 90% of BhuvanNR's
  [step-8 file](https://www.kaggle.com/code/bhuvannr2006/s6e11-lm-ladder-forecast-the-board-lb-0-95417) (0.95419),
  which is one more rung on goodpjw2008's
  [Forecast the Board](https://www.kaggle.com/code/goodpjw2008/s6e11-lm-ladder-forecast-the-board-lb-0-95417) ladder;
- 10% of alexchilton's
  [shallow-tree stack](https://www.kaggle.com/code/alexchilton/credit-card-fraud-lb-0-9542-shallow-tree-stack).

A note on S. Areeb's table lists the stack at 0.95417. alexchilton's own notebook gives its `submission.csv` as
public 0.95260, and the public model agrees (left out, it forecasts S at 0.95260). This section uses 0.95260.

**The model** is goodpjw2008's public model, built openly from scratch the way BhuvanNR did it and fitted on
52 public files with a publicly stated score (and then on the two probes below). Forecast errors:
- 0.6 points when one file is left out;
- 1.4 points when each ladder file is forecast only from the files scored before it.

The second number is the honest one for a new probe.

**Two probes, then stop.** Two files picked by the model were scored on the public board. Both came in at
**0.95422**, as forecast. Refitted with both results, the model's best next file forecasts only 0.95422.15: less than
half a point above what is already scored, well inside the forecast error. So the probing stops there.

**Read this before you use the file.** `submission_v11_steered.csv` (this notebook's `submission.csv` in v11) is
**public-board-steered**:
- it is probe 1 (public 0.95422); its weights were picked by a forecast of the public score, then checked on the board;
- the OOF rows do not ask for it: on OOF it is 4.8 points (0.000048) below the 0.95421 file, with 0 of 5 folds up;
- its expected private change is **&minus;4.8 points [&minus;6.5, &minus;3.0]** versus the 0.95421 file, and about
  **&minus;83 points [&minus;104, &minus;62]** versus goodpjw2008's CV-led file;
- of the two 0.95422 files it is the one with the smaller OOF cost (&minus;4.8 against &minus;18.4).

**The CV-backed alternative** is `submission_cv_backed.csv` = 0.89 X + 0.11 S. Its public forecast is 0.95421 (no gain),
but it is +1.5 points above the 0.95421 file on OOF, up on 5 of 5 folds. For the private rows the OOF evidence points
further still, to the CV-led file (`submission_cv.csv`, about 77 points above this alternative on our reading).

````python
# v11: the public-board-steered file, rebuilt byte for byte from public files: a weighted sum in logit-rank space.
# Inputs: name -> (file names, SHA-256, a folder-name part that tells the attached inputs apart). alexchilton re-ran his
# notebook and two of the source notebooks were renamed, so S, his 90/10 files, O and A are read from pinned,
# byte-identical copies in the dataset s6e11-public-board-files (checked by SHA-256 as before).
V11_IN = {
    "X": (["submission_step8.csv", "submission.csv"], "29e5ab0ff9c7a1e25c8af42eede2418fdc5188ced7bf49c19a216b10974030d2", "forecast-the-board"),  # BhuvanNR step 8, 0.95419
    "S": ("alexchilton_v23_submission.csv", "36686835b6ca3bd805ef578fc50cd62b3cbbe4d2a70275e50d6c38c072d4ad77", "public-board-files"),                       # alexchilton's stack
    "0.95417 file": (["submission_0p95417.csv", "submission.csv"], "d143a60b14b4a8cc1ed963f846f5b0ebe6a28989e92b363d7e7c06595299ac15", "forecast-the-board"),           # goodpjw2008 step 7
    "CV-led file": (["submission_private.csv", "submission_cv.csv"], "2ebbbc99cc04c489909762fa1a56aed66e8fb9b5cec78ec811f4630d330eff08", "forecast-the-board"),
    "alex s417_wide10": ("alexchilton_v23_submission_s417_wide10.csv", "c403f3d93b269ef4fa89b1055dd4b069c119586f0957c7d4ceaeeab1edf65b14", "public-board-files"),
    "alex cvled_wide10": ("alexchilton_v23_submission_cvled_wide10.csv", "cfcc0c3cf6ed6b77ba8c83cb3825713a12724b923c799f937a9e8053d6b923bf", "public-board-files"),
    "O": ("areebali_v22_submission.csv", "b08485cffab7f66cac99ff7872c8d1d1563128c6d63d88f26dc72452feb1ce14", "public-board-files"),        # Areeb Ali, 0.95421
    "A": ("asteriospanoras_v11_submission.csv", "9ecd6ea1ed92bfaa69e609d5ad2d95f039d7728e1b797f4b537a948399c463fc", "public-board-files"),  # this notebook, v11
}
V11_REF = dict(weights={"X": 0.90, "S": 0.10}, sha="b08485cffab7f66cac99ff7872c8d1d1563128c6d63d88f26dc72452feb1ce14")   # Areeb Ali, 0.95421
# The pick: probe 1 (Phase-1 candidate 3), public 0.95422. Numbers from the offline public model and OOF:
# forecast and band in points of 0.00001; OOF change vs the 0.95421 file, folds up, and the expected private change
# [90% interval] vs the 0.95421 file and vs the CV-led file.
V11_PICK = dict(name="probe 1", weights={"X": 1.44, "S": 0.16, "0.95417 file": -0.30, "alex s417_wide10": -0.30},
                sha="9ecd6ea1ed92bfaa69e609d5ad2d95f039d7728e1b797f4b537a948399c463fc",
                forecast=95422.4, band=1.4, public=95422, oof=-4.8, folds="0/5", priv_ref=(-4.8, -6.5, -3.0), priv_cv=(-83, -104, -62))
# The CV-backed alternative (Phase-1 candidate 6): no forecast gain, up on OOF on every fold.
V11_CVB = dict(name="CV-backed alternative", weights={"X": 0.89, "S": 0.11},
               sha="d30021174e62f391725e4939944f0ac20b6a10d5143a896559c234a96a86b7aa",
               forecast=95421.0, band=1.4, oof=1.5, folds="5/5", priv_ref=(1.5, 1.0, 2.0), priv_cv=(-77, -96, -57))
# Every probe of this section: (file, forecast written down before the probe, public score), then the refit's best next forecast.
V11_PROBES = [("probe 1: +1.44 X +0.16 S -0.30 (0.95417 file) -0.30 (alex s417_wide10)", 95422.4, 95422),
              ("probe 2: +1.35 X +0.15 S -0.20 (CV-led file) -0.30 (0.95417 file)", 95422.4, 95422)]
V11_NEXT = 95422.15   # the best forecast after the refit with both probes: less than 0.5 above the best probed file, so the stopping rule ends probing

def _v11_file(weights):
    comp = {}
    for k in weights:
        names, digest, inside = V11_IN[k]
        d = pd.read_csv(find(names, digest, inside=inside)); assert (d["id"].to_numpy() == sample["id"].to_numpy()).all()
        comp[k] = L(d["is_fraud"].to_numpy())
    return sum(w * comp[k] for k, w in weights.items())          # the same order of operations as the offline build
def _v11_write(t, path):
    sample[["id"]].assign(is_fraud=(rankdata(t) - 0.5) / NT).to_csv(path, index=False)
    _sha.pop(Path(path), None); return sha256(path)   # the file was rewritten: hash it again, not from the cache
_h_ref = _v11_write(_v11_file(V11_REF["weights"]), WORK / "_v11_ref.csv")
if _h_ref != V11_REF["sha"]:   # the rebuild should match; if it ever does not, use the pinned copy (checked by SHA-256)
    shutil.copyfile(find(*V11_IN["O"][:2], inside=V11_IN["O"][2]), WORK / "_v11_ref.csv")
V12_O = L(pd.read_csv(WORK / "_v11_ref.csv")["is_fraud"].to_numpy(np.float64)); (WORK / "_v11_ref.csv").unlink()   # O of section v12
check("the 0.95421 file (Areeb Ali) rebuilt as 0.90 X + 0.10 S in logit-rank space",
      "byte-identical to his published file (SHA-256)" if _h_ref == V11_REF["sha"] else "SHA-256 differs", _h_ref == V11_REF["sha"])
SUB11_SHA = _v11_write(_v11_file(V11_PICK["weights"]), WORK / "submission_v11_steered.csv")
if SUB11_SHA != V11_PICK["sha"]:   # the same fallback for A, the v11 file
    shutil.copyfile(find(*V11_IN["A"][:2], inside=V11_IN["A"][2]), WORK / "submission_v11_steered.csv")
check(f"submission_v11_steered.csv = {V11_PICK['name']} (public-board-steered, v11)",
      f"SHA-256 {SUB11_SHA[:16]}&hellip; " + ("= the offline build, byte for byte" if SUB11_SHA == V11_PICK["sha"] else "differs from the offline build"),
      SUB11_SHA == V11_PICK["sha"])
SUB11CV_SHA = _v11_write(_v11_file(V11_CVB["weights"]), WORK / "submission_cv_backed.csv")
check("submission_cv_backed.csv = 0.89 X + 0.11 S (CV-backed alternative)",
      f"SHA-256 {SUB11CV_SHA[:16]}&hellip; " + ("= the offline build, byte for byte" if SUB11CV_SHA == V11_CVB["sha"] else "differs from the offline build"),
      SUB11CV_SHA == V11_CVB["sha"])
log(f"v11: submission_v11_steered.csv = {V11_PICK['name']} (public-board-steered), SHA {SUB11_SHA[:16]}; submission_cv_backed.csv SHA {SUB11CV_SHA[:16]}")
````

## v15: one step on top of the 0.95428 average

On 16 Nov the board moved again. Areeb Ali's *Two Ladders, One Average* (public **0.95428**) is the plain logit-rank

**The full analysis of these latest changes is pending.** This version only rebuilds the file from public inputs and

````python
# v15: T = 0.5 B + 0.5 G + 0.25 (B - B420) + 0.5 (A - O). B, B420: BhuvanNR's outputs; G: goodpjw2008's step 11.
# Every input is found by its SHA-256 anywhere under the inputs (attached notebooks can be mounted under older slugs).
V15_IN = {"B": ("submission.csv", "ce020d84c6279f5946f94983db6fe68b96a0dd96b5809d0a96ecd263fe586703"),
          "B420": ("submission_0p95420.csv", "a2363d42"),
          "G": ("submission.csv", "aba55fd729e87a1db78a4d3d64fa1e5f54fd8495c28490f121221133a3121e8d")}
V15 = dict(sha="f4f633e1ceb45ee90227969a9efff4c5213c2760f3400ca24ec474ca4cb8b385", forecast=95428.30, band=0.84)
def _v15_get(name, digest):
    for q in sorted(INPUT.rglob(name)) + sorted(INPUT.rglob("*submission*.csv")):   # then the pinned copies
        if sha256(q).startswith(digest):
            d = pd.read_csv(q); assert (d["id"].to_numpy() == sample["id"].to_numpy()).all(); return L(d["is_fraud"].to_numpy(np.float64))
    print("v15 input not found:", name, digest[:16]); return None
_v15 = {k: _v15_get(*v) for k, v in V15_IN.items()}
shutil.copyfile(WORK / "submission.csv", WORK / "submission_v12.csv")
if all(v is not None for v in _v15.values()):
    _C15 = 0.5 * _v15["B"] + 0.5 * _v15["G"]
    SUB15_SHA = _v11_write(_C15 + 0.25 * (_v15["B"] - _v15["B420"]) + 0.5 * (_A12 - V12_O), WORK / "submission.csv")
else:
    SUB15_SHA = "missing inputs"
check("submission.csv = 0.5 B + 0.5 G + 0.25 (B - B420) + 0.5 (A - O) (public-board file, v15)",
      f"SHA-256 {SUB15_SHA[:16]}&hellip; " + ("= the offline build, byte for byte" if SUB15_SHA == V15["sha"] else "differs from the offline build"),
      SUB15_SHA == V15["sha"])
log(f"v15: submission.csv SHA {SUB15_SHA[:16]}")
````

## v16: a real model update on top of the board files

On 17 Nov sriram00777 (*Top 20 CPU Blend + EDA*, public **0.95440**) found that the top files had not yet seen
a new model release: Vladimir Demidov's (yekenot) RealMLP, PyTabKit v15 (N), against his earlier PyTorch v28 run (C).
His recipe updates each of the two 0.95427 files by `0.3 (N &minus; C)` and averages them as rank fractions.
We did the same on goodpjw2008's newer 0.95432 file (G') and BhuvanNR's step 10 (B), with 0.4 of the update:

`U( [ U(L(G') + 0.4 D) + U(L(B) + 0.4 D) ] / 2 )`, with `D = L(N) &minus; L(C)`, L = logit rank, U = rank fraction.

It is this notebook's `submission.csv` (public **0.95442**). The v15 file is kept as `submission_v15.csv`.

**The full analysis of these latest changes is pending.** Unlike the earlier steps, part of this move comes from a
new model, not only from the board. How much of it should survive on the private rows is the open question.

````python
# v16: T = U( [U(L(G') + 0.4 D) + U(L(B) + 0.4 D)] / 2 ), D = L(N) - L(C); each U() goes through a CSV round trip.
V16_IN = {"G'": ("submission.csv", "4d88b6f6dc7612df8d5ff9d77799748815b79f9762103341c0be48aa3dd45c1d"),
          "N": ("submission.csv", "b1462d0c7f64a48a4739acc4e68673d63e4648660744acb734b6770bc5207055"),
          "C": ("submission.csv", "c8ff11aeedb8e7a6268a1617ae14133f2dc7fa347a581f3cceed0ac6d44bfc98")}
V16 = dict(w=0.40, sha="46eb96d06ea474a16029ee65df7fdbfcf44ee1b6197e8b2e0136ab908f5a6299", public=95442)
_v16 = {k: _v15_get(*v) for k, v in V16_IN.items()}
shutil.copyfile(WORK / "submission.csv", WORK / "submission_v15.csv")
import io
def _v16_rt(x):   # rank fraction written and read back as CSV text, exactly as the scored file was built
    _b = io.StringIO(); sample[["id"]].assign(is_fraud=(rankdata(x) - 0.5) / NT).to_csv(_b, index=False); _b.seek(0)
    return pd.read_csv(_b)["is_fraud"].to_numpy(np.float64)
if all(v is not None for v in _v16.values()) and _v15["B"] is not None:
    _D16 = L(_v16["N"]) - L(_v16["C"])
    _T16 = (_v16_rt(L(_v16["G'"]) + V16["w"] * _D16) + _v16_rt(L(_v15["B"]) + V16["w"] * _D16)) / 2
    SUB16_SHA = _v11_write(_T16, WORK / "submission.csv")
else:
    SUB16_SHA = "missing inputs"
check("submission.csv = U([U(L(G') + 0.4 D) + U(L(B) + 0.4 D)] / 2), D = L(N) - L(C) (v16)",
      f"SHA-256 {SUB16_SHA[:16]}&hellip; " + ("= the scored file, byte for byte" if SUB16_SHA == V16["sha"] else "differs from the scored file"),
      SUB16_SHA == V16["sha"])
log(f"v16: submission.csv SHA {SUB16_SHA[:16]}")
````

smaller one. The v17 file is kept as `submission_v17.csv`. As in v17, the extra half step comes from the public board only.

````python
# v18: P + 1.5 (L(S_C) - L(P)), P = the v17 file, S_C read from the pinned copy by SHA-256
V18 = dict(k=1.5, sha="4b5b85600bae3873", public=95452)
_v18_C = _v15_get("submission.csv", "2e984f68ae14a770")
shutil.copyfile(WORK / "submission.csv", WORK / "submission_v17.csv")
if _v18_C is not None and SUB17_SHA.startswith(V17["sha"]):
    _P17r = L(pd.read_csv(WORK / "submission_v17.csv")["is_fraud"].to_numpy(np.float64))
    SUB18_SHA = _v11_write(_P17r + V18["k"] * (_v18_C - _P17r), WORK / "submission.csv")
else:
    SUB18_SHA = "missing inputs"
check("submission.csv = P + 1.5 (L(S_C) - L(P)) (v18)",
      f"SHA-256 {SUB18_SHA[:16]}&hellip; " + ("= the scored file, byte for byte" if SUB18_SHA.startswith(V18["sha"]) else "differs from the scored file"),
      SUB18_SHA.startswith(V18["sha"]))
log(f"v18: submission.csv SHA {SUB18_SHA[:16]}")
````

<a id="s1"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">1 &middot; The competition in five minutes</div>

* **Data.** 699,960 synthetic card transactions to train on, 299,984 to predict; 24 columns (amount, timing, device,
  merchant, risk scores, flags). About **1.67%** of the training rows are fraud.
* **Metric.** ROC-AUC: the probability that a random fraud is ranked above a random legitimate transaction. Only the
  **order** of the predictions matters, which is why every public blend here works on logit ranks
  `L(v) = logit((rank(v) - 0.5) / n)`.
* **The split.** The public leaderboard scores **20%** of the test rows, the private leaderboard (the final ranking)
  the other **80%**. Both are fixed for everyone; nobody knows which rows are which.
* **Units.** Gaps at the top are a few hundred-thousandths of AUC. We write them as **points of 1e-5**:
  0.95354 - 0.95352 = +2 points.

````python
# The two boards in rows and expected frauds (train fraud rate applied to the test set).
fig, ax = plt.subplots(figsize=(13, 2.9))
rate = y.mean()
segs = [("train", 0, N, MUTED, f"train  {N:,} rows  |  {int(y.sum()):,} frauds"),
        ("public", 0, NPUB, PUB, f"public LB  ~{NPUB:,} rows\n~{NPUB * rate:,.0f} frauds"),
        ("private", NPUB, NT - NPUB, PRIV, f"private LB  ~{NT - NPUB:,} rows  |  ~{(NT - NPUB) * rate:,.0f} frauds")]
ax.barh(1, N, color="#e6e5df", height=0.55); ax.text(N / 2, 1, segs[0][4], ha="center", va="center", fontsize=12.5, color=INK)
ax.barh(0, NPUB, color=PUB, height=0.55); ax.barh(0, NT - NPUB, left=NPUB + 1500, color=PRIV, height=0.55)
ax.text(NT + 9000, 0, f"blue = public LB: ~{NPUB / 1000:.0f}k rows, ~{NPUB * rate:,.0f} frauds", ha="left", va="center", fontsize=12.5, color=PUB, fontweight="bold")
ax.text(NPUB + (NT - NPUB) / 2, 0, f"private LB: ~{(NT - NPUB) / 1000:.0f}k rows, ~{(NT - NPUB) * rate:,.0f} frauds", ha="center", va="center", fontsize=12.5, color="white", fontweight="bold")
ax.set_yticks([1, 0]); ax.set_yticklabels(["train", "test"], fontsize=13); ax.tick_params(axis="y", length=0)
ax.set_xlim(0, N * 1.01); ax.grid(False); ax.spines["left"].set_visible(False)
ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v / 1000:.0f}k")); ax.set_xlabel("rows")
headline(fig, f"The public board ranks everyone on about {NPUB * rate:,.0f} frauds",
         "Public = 20% of the test rows, private = the other 80%. Frauds are estimated from the training fraud rate.")
plt.show()
````

**Takeaway:** the public score rests on about a thousand frauds, so every public number carries real sampling noise.

<a id="s2"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">2 &middot; How noisy is the public LB?</div>

Two different questions hide behind "the public LB is noisy":

* **How far is one file's public score from its private score?** Far: the AUC of a fixed file on ~60k rows with about
  a thousand frauds wanders by a few *thousandths*. But every team is scored on the same rows, so this mostly shifts
  the whole board together.
* **How reliable is the public gap between two files?** This is what decides rankings and ladder steps. Two
  files that rank almost every row the same way share most of their noise, so the gap is far less noisy &mdash; but
  the gaps at the top are also tiny.

We measure both on the nested OOF of the public wide LM stack: draw a public part (~60k rows) and a private part
(~240k rows) at random, score every file on both, repeat a few hundred times.

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.4))
gap1 = (AUC[0, :, 0] - AUC[0, :, 1]) * 1e5
ax.hist(gap1, bins=30, color=MUTED, edgecolor=SURF)
ax.set_xlabel("public AUC minus private AUC of ONE file (x 1e-5)"); ax.set_ylabel("random splits"); ax.xaxis.set_major_formatter(E5)
panel_title(ax, f"One file: SD {np.std(gap1):.0f} points", "the whole board shares this shift")
iS = NAMES.index("GPT-2 seed swap 0.06")
bins_ = np.linspace(min(DELTA[iS, :, 0].min(), DELTA[iS, :, 1].min()), max(DELTA[iS, :, 0].max(), DELTA[iS, :, 1].max()), 36)
ax2.hist(DELTA[iS, :, 0], bins=bins_, color=PUB, edgecolor=SURF, alpha=0.9, label="public gap")
ax2.hist(DELTA[iS, :, 1], bins=bins_, color=PRIV, edgecolor=SURF, alpha=0.75, label="private gap")
ax2.axvline(TRUE["GPT-2 seed swap 0.06"], color=INK, lw=1.5)
ax2.text(TRUE["GPT-2 seed swap 0.06"] + 0.3, ax2.get_ylim()[1] * 0.95, f"nested CV {TRUE['GPT-2 seed swap 0.06']:+.1f}", fontsize=11, va="top")
zero_line(ax2, "x"); ax2.xaxis.set_major_formatter(E5); ax2.legend(loc="upper left", fontsize=11)
ax2.set_xlabel("gap between two consecutive ladder files (x 1e-5)")
panel_title(ax2, f"Two near-identical files: public SD {SD_PUB[iS]:.1f}, private SD {SD_PRV[iS]:.1f}",
            f"the GPT-2 seed swap: public gap < 0 in {np.mean(DELTA[iS, :, 0] < 0):.0%} of splits")
headline(fig, f"One public score swings by +-{np.std(gap1):.0f} points; the gap between two near-identical files by +-{SD_PUB[iS]:.1f}, still more than a rung",
         "Independent random public (~60k rows) and private (~240k rows) draws from the public wide-stack OOF")
plt.show()
````

**Takeaway:** even two near-identical files can trade places on the public board by chance, by more than one ladder rung.

### A one-line rule for the noise of a public gap
For two files `a` and `b`, the noise of their public gap depends almost only on **how differently they rank the rows**,
that is on `1 - rho`, where `rho` is the rank (Spearman) correlation of the two files. Over every move measured here the
relation is a clean square root:

````python
fig, ax = plt.subplots(figsize=(13, 5.2))
order_ = np.argsort(ONE_RHO)
is_move = np.array([n in MV or n == "tree move" for n in NAMES])
ax.scatter(ONE_RHO[~is_move], SD_PUB[~is_move], s=36, color=PUB, alpha=0.55, edgecolor="none", label="random probe steps (public part)")
ax.scatter(ONE_RHO[is_move], SD_PUB[is_move], s=60, color=PUB, edgecolor=SURF, lw=1.2, label="public ladder probes (public part)")
ax.scatter(ONE_RHO, SD_PRV, s=22, color=PRIV, alpha=0.6, edgecolor="none", label="the same steps, private part")
xx = np.logspace(np.log10(ONE_RHO.min() * 0.6), np.log10(ONE_RHO.max() * 1.6), 50)
ax.plot(xx, K_RULE * np.sqrt(xx) * 1e5, color=NAVY, lw=2)
ax.plot(xx, K_RULE * np.sqrt(xx) * 1e5 * np.sqrt(NPUB / (NT - NPUB)), color=PRIV, lw=1.5, ls=(0, (4, 3)))
ax.text(xx[-1], K_RULE * np.sqrt(xx[-1]) * 1e5, f"  public: {K_RULE:.4f} x sqrt(1 - rho)", va="center", fontsize=12, fontweight="bold", color=NAVY)
ax.text(xx[-1], K_RULE * np.sqrt(xx[-1]) * 1e5 * 0.5, "  private: half of it", va="center", fontsize=11.5, color=PRIV)
if HAVE_LADDER:
    for lab, g, orho, sd in STEPS:
        ax.axvline(orho, color=RULE, lw=1, zorder=0)
    ax.text(max(s[2] for s in STEPS) * 1.08, 0.04, "grey lines: consecutive\npublic ladder files", fontsize=10.5, color=INK2, va="bottom",
            transform=ax.get_xaxis_transform())
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("1 - rank correlation of the two files"); ax.set_ylabel("SD of the gap (x 1e-5)")
ax.set_xlim(xx[0], xx[-1] * 4.5); ax.legend(loc="upper left", fontsize=11)
headline(fig, f"Noise of a public gap = {K_RULE:.4f} x sqrt(1 - rho): about {K_RULE * np.sqrt(1e-4) * 1e5:.0f} points when rho = 0.9999",
         "Each dot: SD of the gap between the wide-stack OOF and one small move of it, over random 20/80 splits")
plt.show()
````

**Takeaway:** the noise of a public gap depends almost only on how similar the two files are, so it can be computed before submitting.

````python
# Shown: use the rule on your own two submission files (the constant K_RULE was fitted above on public OOF; it assumes
# a test set and fraud rate like this competition's, and two files of similar strength).
def public_gap_noise(file_a, file_b, k=K_RULE):
    """1 SD (in AUC) of the public-LB gap between two submissions, from their rank correlation alone."""
    a = pd.read_csv(file_a)["is_fraud"].to_numpy(); b = pd.read_csv(file_b)["is_fraud"].to_numpy()
    rho = np.corrcoef(rankdata(a), rankdata(b))[0, 1]
    return k * np.sqrt(max(1 - rho, 0)), rho
_rr = (0.99999, 0.9999, 0.9995, 0.999, 0.995)
table(pd.DataFrame({"rank correlation of the two files": [f"{r_:g}" for r_ in _rr],
                    "1 SD of the public gap (points)": [f"{K_RULE * np.sqrt(1 - r_) * 1e5:.1f}" for r_ in _rr],
                    "1 SD of the private gap (points)": [f"{K_RULE * np.sqrt(1 - r_) * 1e5 / 2:.1f}" for r_ in _rr]}),
      "The rule in numbers", "1 point = 0.00001 AUC. public_gap_noise(file_a, file_b) returns the first column for your own two files.")
````

<div style="background: #eef4fc; color: #1a1a1a; border-left: 5px solid #2a78d6; padding: 10px 14px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a;">How to read it.</b> Two files with rank correlation 0.9999 have a public gap noise of a few points, about the size
of a typical ladder step. A gain of one or two points between such files is <b style="color: #1a1a1a;">not evidence</b> of anything: the same
change would show a loss on public in a large share of possible splits. The private gap is half as noisy (four times
more rows), so the private board will not simply repeat the public order.
</div>

<a id="s3"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">3 &middot; Anatomy of a ladder</div>

A ladder starts from a strong public file and adds small steps in logit-rank space, keeping each step that the public
score rewards. The current public top (0.95354) is built like this, and its authors published **every rung and every
probe** (goodpjw2008's *LM Ladder*, Areeb Ali's *Three Ladders*):

```
s352  = A + 0.6 (A - F) + 0.09 (LM - F) + 0.09 (L7F - F) + 0.05 (GPD - F) - 0.05 (MLP - F) - 0.025 (GB - F)   # 0.95352
score = L(s352) + 0.06 (G26 - MS)                                                                            # 0.95354
```

`A` = Strike 10 (amanatar, 0.95340), `F` = Strike 9 (Moran Z, 0.95330), `LM` = megayak's target-first LM, `L7F` =
goodpjw2008's seed-7 LM, `GPD` = daoviet's GP + DVAE, `MLP`/`GB` = two OOF-bank bags, `MS` = masayakawamata's
target-first GPT-2, `G26` = a second seed of it (goodpjw2008). The notebook rebuilt this file above from the public
inputs and checked it byte for byte against the published one.

**The part no one can check.** `A` and `F` publish no OOF predictions, so the base and the extrapolation
`0.6 (A - F)` (move further along the direction from Strike 9 to Strike 10) have no out-of-fold check at all. Every
other ingredient has public OOF files, so each step can be replayed on a nested OOF, as the ladder authors did
themselves. The table collects all of them.

````python
rows = []
for n, d, pub, src in MOVES:
    i = NAMES.index(n)
    rows.append({"probe (step applied to the ladder file)": n, "public LB change": f"{pub:+d}",
                 "nested CV change": f"{TRUE[n]:+.1f}", "folds up": f"{UP[n]}/5",
                 "public noise SD": f"{SD_PUB[i]:.1f}", "P(public sign = private sign)": f"{AGREE[i]:.0%}", "source of the public number": src})
i = NAMES.index("tree move")
rows.append({"probe (step applied to the ladder file)": "full tree move (-0.10 MLP, -0.05 GB)", "public LB change": "part of v15",
             "nested CV change": f"{TRUE['tree move']:+.1f}", "folds up": f"{UP['tree move']}/5", "public noise SD": f"{SD_PUB[i]:.1f}",
             "P(public sign = private sign)": f"{AGREE[i]:.0%}", "source of the public number": "Areeb Ali v15 (with other changes)"})
TAB = pd.DataFrame(rows)
table(TAB, "Every public probe with OOF behind it", "All changes in points of 1e-5. Nested CV = the step applied to the nested OOF of the public "
      f"wide LM stack (CV {CV_W:.6f}).", wrap=["source of the public number"])
````

````python
fig, ax = plt.subplots(figsize=(13, 5.6))
names_ = [m[0] for m in MOVES] + ["tree move"]
pubv = [m[2] for m in MOVES] + [np.nan]
cvv = [TRUE[n] for n in names_]
sdv = [SD_PUB[NAMES.index(n)] for n in names_]
yy = np.arange(len(names_))[::-1]
for yi, s in zip(yy, sdv):
    ax.add_patch(FancyBboxPatch((-s, yi - 0.38), 2 * s, 0.76, boxstyle="round,pad=0,rounding_size=0.1", facecolor=BAND, edgecolor="none", zorder=0))
ax.barh(yy + 0.17, pubv, height=0.32, color=PUB, label="reported public LB change", zorder=3)
ax.barh(yy - 0.17, cvv, height=0.32, color=[CV if (c > 0 and UP[n] >= 4) else "#9ccfba" if c > 0 else PRIV for c, n in zip(cvv, names_)],
        label="nested CV change", zorder=3)
for yi, n, c in zip(yy, names_, cvv):
    ax.text(max(0, c) + 0.4 if c >= 0 else c - 0.4, yi - 0.17, f"{c:+.1f}  ({UP[n]}/5 folds)", va="center",
            ha="left" if c >= 0 else "right", fontsize=10.5, color=INK2)
ax.set_yticks(yy); ax.set_yticklabels([SHORT.get(n, "full tree move") for n in names_], fontsize=11.5); ax.tick_params(axis="y", length=0)
zero_line(ax, "x"); ax.xaxis.set_major_formatter(E5); ax.set_xlabel("change (x 1e-5)"); ax.grid(axis="y", visible=False)
ax.set_xlim(-17, 8)
ax.legend(handles=[Patch(color=PUB, label="reported public LB change"), Patch(color=CV, label="nested CV: up on >= 4 of 5 folds"),
                   Patch(color="#9ccfba", label="nested CV: up on < 4 folds"), Patch(color=PRIV, label="nested CV: down"), Patch(color=BAND, label="+-1 SD public noise of the step")],
          loc="upper center", bbox_to_anchor=(0.45, -0.13), ncol=3, fontsize=11)
headline(fig, f"Only the GPT-2 seed swap gains on both boards; {len(POS) - len(CONF)} other public gains are not confirmed by nested CV",
         "Every public probe that has OOF files behind it, replayed on the nested OOF of the public wide LM stack")
plt.show()
````

**Takeaway:** most public gains of the old ladders do not hold up on nested CV; the one that does has a clear mechanism.

**Reading the table.** The ladder authors saw the same thing and said so: Areeb Ali writes that *"every earlier ladder
step lost on CV"* and that the two-seed rung is the first one that holds up; goodpjw2008 calls a rung *"a bet, not a
measurement"*. Our replay agrees with their own CV checks: the two-seed rung gains about +1.9e-5 on 5 of 5 folds
(goodpjw2008: +1.9e-5) and the full tree move loses about 7.7e-5 (Areeb Ali: -7.4e-5).

The two-seed rung is also the one with a clear mechanism: averaging two seeds of the same model removes part of one
seed's noise, which is a gain one expects on any rows. The other steps tune *how much* of a member the file holds, and
there the public board and CV disagree; with a public noise SD of 3&ndash;7 points per step, the public board cannot
settle a one- or two-point question.

<a id="s4"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">4 &middot; The winner's curse</div>

Try `K` probes, keep the one with the best public score. Even if none of them is truly better, the best of `K` noisy
numbers is above zero, by about `sigma x E[max of K standard normals]`: about 0.8 sigma for K = 3, 1.5 sigma for K = 10,
2.0 sigma for K = 30, 2.5 sigma for K = 100. The private part of the same split does not share that luck, so the pick
falls back. This is the textbook *winner's curse* (selection bias) and, repeated over many rounds of public feedback,
*adaptive overfitting* of the public holdout.

Below, the probes are real: small steps from the wide stack towards (or away from) each OOF-bank member and each public
LM. On every random split we pick the best of `K` by its public gap and read its private gap.

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw=dict(width_ratios=[1.15, 1]))
kk = np.array(KGRID)
pub_m = np.array([CURSE[k][:, 0].mean() for k in KGRID]); prv_m = np.array([CURSE[k][:, 1].mean() for k in KGRID])
tru_m = np.array([CURSE[k][:, 2].mean() for k in KGRID])
ax.fill_between(kk, prv_m, pub_m, color=PUB, alpha=0.08, lw=0)
ax.plot(kk, pub_m, "o-", color=PUB, ms=7, mec=SURF, label="public gap of the pick")
ax.plot(kk, prv_m, "o-", color=PRIV, ms=7, mec=SURF, label="private gap of the same pick")
ax.plot(kk, tru_m, "o--", color=CV, ms=6, mec=SURF, lw=1.4, label="nested CV of the pick (its true size)")
ax.annotate(f"{pub_m[-1] - prv_m[-1]:.1f} points\nlost on private", xy=(kk[-1], (pub_m[-1] + prv_m[-1]) / 2), xytext=(-12, 0),
            textcoords="offset points", ha="right", va="center", fontsize=11.5, fontweight="bold", color=INK)
zero_line(ax); ax.set_xscale("log"); ax.set_xticks(KGRID); ax.set_xticklabels([str(k) for k in KGRID])
ax.yaxis.set_major_formatter(E5); ax.set_xlabel("probes compared on public (K)"); ax.set_ylabel("gap vs. the base file (x 1e-5)")
ax.legend(loc="upper left", fontsize=11)
panel_title(ax, "The more probes you compare, the more the pick is luck", "average over random 20/80 splits")
c20 = CURSE[20]
ax2.scatter(c20[:, 0], c20[:, 1], s=8, color=PUB, alpha=0.25, edgecolor="none")
lim = np.percentile(np.abs(c20[:, :2]), 99.5)
ax2.plot([-lim, lim], [-lim, lim], color=MUTED, lw=1, ls=(0, (4, 3)))
zero_line(ax2); zero_line(ax2, "x"); ax2.set_xlim(-lim, lim); ax2.set_ylim(-lim, lim)
ax2.xaxis.set_major_formatter(E5); ax2.yaxis.set_major_formatter(E5)
ax2.set_xlabel("public gap of the best of 20 (x 1e-5)"); ax2.set_ylabel("its private gap (x 1e-5)")
ax2.text(0.97, 0.05, f"public > 0 in {np.mean(c20[:, 0] > 0):.0%} of splits\nprivate > 0 in {np.mean(c20[:, 1] > 0):.0%}",
         transform=ax2.transAxes, ha="right", va="bottom", fontsize=11.5, color=INK)
panel_title(ax2, "Best of 20: almost always up on public", "each dot = one random split")
headline(fig, "Keeping the best public probe buys public points that the private board does not pay",
         "Real probes on the public wide-stack OOF; public and private parts drawn independently in each split")
plt.show()
````

**Takeaway:** choosing the best of many probes by public score buys public points that the private rows do not repeat.

<div style="background: #fff4ec; color: #1a1a1a; border-left: 5px solid #eb6834; padding: 10px 14px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a;">How many probes are behind a public file?</b> The version tables list ten versions for one ladder (v9&ndash;v20) and
about twenty scored probes for the other, and these ladders build on each other. On the leaderboard snapshot the
median top-50 team had made about 20 submissions on 12 Nov (28 on 14 Nov), and the top teams over 100. With a rung noise of a few points,
the theory says the best of 20 null probes shows about 1.9 noise SDs of public "gain". That is the size of the whole
visible spread between many of the top files.
</div>

<a id="s5"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">5 &middot; CV can be tuned too</div>

Nested CV is the referee in this notebook, so it is fair to ask how a CV number can itself be inflated. Three common
ways, each measured below on public data:

1. **Early stopping on the scored fold.** The fold that reports the score also picks the number of trees.
2. **Selecting on the reported folds.** A blend, a member set or a weight chosen on the same OOF rows it is scored on.
3. **A test inside one fold.** An inner CV run on the rows of a single outer fold measures something else.

````python
# 5.1  Early stopping on the scored fold: one 4-leaf LightGBM per outer fold on the raw columns (learning rate 0.1),
# trained on 90% of the training part. We record the AUC every 10 trees on the scored fold AND on the held-out 10%,
# so the two ways of choosing the number of trees are compared on the very same model.
import lightgbm as lgb
NTH = int(os.environ.get("OMP_NUM_THREADS", 4))
X = train.drop(columns=["id", "is_fraud"]).copy()
for c in X.columns:
    if X[c].dtype == object: X[c] = X[c].astype("category")
    elif X[c].dtype == bool: X[c] = X[c].astype(int)
ES_FOLDS, ES_ROUNDS, STEP = [0, 1, 2], 1500, 10
ES = []
for k in ES_FOLDS:
    va = FOLDS[k]; tr = np.setdiff1d(np.arange(N), va)
    inner = np.random.default_rng(k).random(len(tr)) < 0.10
    a, h = tr[~inner], tr[inner]
    bst = lgb.train(dict(objective="binary", num_leaves=4, learning_rate=0.1, min_data_in_leaf=100, verbose=-1,
                         num_threads=NTH, seed=k), lgb.Dataset(X.iloc[a], y[a]), num_boost_round=ES_ROUNDS)
    zh, zv = np.zeros(len(h)), np.zeros(len(va)); its, ah, av = [], [], []
    for s in range(0, ES_ROUNDS, STEP):
        zh += bst.predict(X.iloc[h], start_iteration=s, num_iteration=STEP, raw_score=True)
        zv += bst.predict(X.iloc[va], start_iteration=s, num_iteration=STEP, raw_score=True)
        its.append(s + STEP); ah.append(roc_auc_score(y[h], zh)); av.append(roc_auc_score(y[va], zv))
    its, ah, av = map(np.array, (its, ah, av))
    i_sc, i_in = av.argmax(), ah.argmax()
    ES.append(dict(fold=k, its=its, av=av, i_sc=i_sc, i_in=i_in, opt=(av[i_sc] - av[i_in]) * 1e5))
    log(f"fold {k} trained")
del X
table(pd.DataFrame({"fold": [e["fold"] for e in ES], "trees, chosen on the scored fold": [e["its"][e["i_sc"]] for e in ES],
                    "its AUC": [f"{e['av'][e['i_sc']]:.6f}" for e in ES], "trees, chosen on a held-out 10%": [e["its"][e["i_in"]] for e in ES],
                    "its AUC on the scored fold": [f"{e['av'][e['i_in']]:.6f}" for e in ES], "optimism (points)": [f"{e['opt']:+.1f}" for e in ES]}),
      "Early stopping: where the trees are chosen", "Optimism = the scored-fold AUC of the scored-fold pick minus that of the held-out pick, x 1e-5.")
````

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.6), gridspec_kw=dict(width_ratios=[1.7, 1]))
FC = [C[0], C[6], C[3]]
for e, col in zip(ES, FC):
    rel = (e["av"] - e["av"][e["i_sc"]]) * 1e5
    ax.plot(e["its"], rel, color=col, lw=1.6, label=f"fold {e['fold']}")
    ax.plot(e["its"][e["i_sc"]], 0, "o", ms=8, color=col, mec=SURF, mew=1.5, zorder=5)
    ax.plot(e["its"][e["i_in"]], rel[e["i_in"]], "D", ms=7, mfc=SURF, mec=col, mew=1.8, zorder=5)
low = min((e["av"][e["i_in"]] - e["av"][e["i_sc"]]) * 1e5 for e in ES)
ax.set_ylim(low * 1.6 - 5, 8); ax.set_xlim(100, ES_ROUNDS); ax.yaxis.set_major_formatter(E5)
ax.set_xlabel("number of trees"); ax.set_ylabel("scored-fold AUC minus its maximum (x 1e-5)")
ax.legend(handles=[Line2D([], [], color=c, lw=2, label=f"fold {e['fold']}") for e, c in zip(ES, FC)] +
          [Line2D([], [], marker="o", ls="", color=INK2, label="trees chosen on the scored fold"),
           Line2D([], [], marker="D", ls="", mfc=SURF, mec=INK2, label="trees chosen on a held-out 10%")], loc="lower right", fontsize=10.5)
panel_title(ax, "The top of a noisy curve is always the highest point", "scored-fold AUC along training, 4-leaf LightGBM, learning rate 0.1")
opts = [e["opt"] for e in ES]
ax2.bar(range(len(ES)), opts, color=PRIV, width=0.55)
for i, v in enumerate(opts): ax2.text(i, v + 0.3, f"+{v:.1f}", ha="center", va="bottom", fontsize=12, fontweight="bold", color=INK)
ax2.set_xticks(range(len(ES))); ax2.set_xticklabels([f"fold {e['fold']}" for e in ES]); ax2.yaxis.set_major_formatter(E5)
ax2.set_ylabel("optimism of the reported fold AUC (x 1e-5)"); ax2.set_ylim(0, max(opts) * 1.3 + 1)
panel_title(ax2, f"Free optimism: {np.mean(opts):+.1f}e-5 per fold", "scored-fold pick minus held-out pick")
headline(fig, f"Early stopping on the scored fold reports {np.mean(opts):.0f} points more than an honest choice of trees",
         "Same model, two ways to choose the number of trees: on the fold that is scored, or on a held-out 10% of the training part")
plt.show()
````

**Takeaway:** CV can be tuned too: any choice made on the scored rows makes the score look better than it is.

The held-out 10% is itself a noisy judge, so its pick lands away from the scored-fold peak: that is the point. The
scored-fold pick hides this noise and reports the peak; any choice made without the scored rows pays for it.
A fast learner (learning rate 0.1) is the bad case: its curve is jagged, so the maximum of the scored fold sits well
above what a choice made elsewhere achieves. With the slower learning rate most public tree members use (0.03) a
separate measurement of the same comparison gave about **+2 to +3e-5 per member**. Per member that is small next to
fold noise, but it is **systematic**: a stack of a hundred members
that all stopped on their scored fold inherits it, and stack weights then favour the members whose OOF is most
optimistic. Choosing the number of trees on a held-out slice of the training part
costs nothing and removes it.

````python
# 5.2  Selecting on the reported folds: greedy hill climbing (Caruana) over the OOF bank's 20 core members.
# Non-nested: choose the blend on all OOF rows, report its CV on the same rows.  Nested: choose on 4 folds, score the 5th.
names = list(BANK); M = np.stack([BANK[n] for n in names], 1).astype(np.float32)
HC_STEPS = 10
def hill_climb(rows):
    """Greedy forward selection with replacement on `rows`; returns the list of picks (indices into names)."""
    yr = y[rows]; Mr = M[rows]
    first = int(np.argmax([auc_sort(yr, Mr[:, j]) for j in range(M.shape[1])]))
    picks, s = [first], Mr[:, first].astype(np.float64).copy()
    for _ in range(HC_STEPS - 1):
        sc = [auc_sort(yr, s + Mr[:, j]) for j in range(M.shape[1])]
        j = int(np.argmax(sc)); picks.append(j); s += Mr[:, j]
    return picks
def blend(picks, upto): return M[:, picks[:upto]].mean(1)
full_picks = hill_climb(np.arange(N))
HC_NON = [cv(blend(full_picks, t)) for t in range(1, HC_STEPS + 1)]
HC_NEST = np.zeros((5, HC_STEPS))
for k, b in enumerate(FOLDS):
    p = hill_climb(np.setdiff1d(np.arange(N), b))
    for t in range(1, HC_STEPS + 1): HC_NEST[k, t - 1] = roc_auc_score(y[b], blend(p, t)[b])
HC_NEST_CV = HC_NEST.mean(0)
BEST_SINGLE = max(cv(M[:, j]) for j in range(M.shape[1]))
log(f"hill climbing: reported CV {HC_NON[-1]:.6f}, nested CV {HC_NEST_CV[-1]:.6f}")
````

````python
# 5.3  A test inside one fold: does adding megayak's LM (already a member of the wide stack) to W help?
# Outer test: weights fitted on the other four folds, scored on fold k.  Inner test: a 5-fold CV inside fold k's rows only.
Z = np.c_[W, O["LM"]]
INOUT = []
for k, b in enumerate(FOLDS):
    rest = np.setdiff1d(np.arange(N), b)
    lr_ = LogisticRegression(C=1.0, max_iter=500).fit(Z[rest], y[rest])
    outer = roc_auc_score(y[b], Z[b] @ lr_.coef_[0]) - roc_auc_score(y[b], W[b])
    ins = []
    for it, ib in StratifiedKFold(5, shuffle=True, random_state=0).split(b, y[b]):
        l2 = LogisticRegression(C=1.0, max_iter=500).fit(Z[b[it]], y[b[it]])
        ins.append(roc_auc_score(y[b[ib]], Z[b[ib]] @ l2.coef_[0]) - roc_auc_score(y[b[ib]], W[b[ib]]))
    INOUT.append((k, np.mean(ins) * 1e5, outer * 1e5))

log("inner and outer tests done")
````

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw=dict(width_ratios=[1.25, 1]))
t = np.arange(1, HC_STEPS + 1)
ax.plot(t, (np.array(HC_NON) - BEST_SINGLE) * 1e5, color=PRIV, marker="o", ms=6, mec=SURF, label="reported: chosen and scored on the same OOF")
ax.plot(t, (HC_NEST_CV - BEST_SINGLE) * 1e5, color=CV, marker="o", ms=6, mec=SURF, label="nested: chosen on 4 folds, scored on the 5th")
gap = (HC_NON[-1] - HC_NEST_CV[-1]) * 1e5
ax.annotate(f"gap {gap:+.1f}e-5", xy=(HC_STEPS, (HC_NON[-1] - BEST_SINGLE) * 1e5), xytext=(-8, -26), textcoords="offset points",
            ha="right", fontsize=12, fontweight="bold", color=INK)
ax.set_xlabel("hill-climbing step (members in the blend, with repeats)"); ax.set_ylabel("CV gain over the best single member (x 1e-5)")
ax.yaxis.set_major_formatter(E5); ax.set_xticks(t); ax.legend(loc="lower right", fontsize=11)
panel_title(ax, f"Selection on the scored rows reads {gap:.0f} points high", "greedy blend of the OOF bank's 20 core members")
ks = [r[0] for r in INOUT]
for k, i_, o_ in INOUT:
    ax2.plot([i_, o_], [k, k], color=RULE, lw=3, zorder=1)
ax2.scatter([r[1] for r in INOUT], ks, s=70, color=PRIV, zorder=3, label="inner CV inside the fold", edgecolor=SURF, lw=1.5)
ax2.scatter([r[2] for r in INOUT], ks, s=70, color=CV, zorder=3, label="outer test on the fold", edgecolor=SURF, lw=1.5)
zero_line(ax2, "x"); ax2.set_yticks(ks); ax2.set_yticklabels([f"fold {k}" for k in ks]); ax2.invert_yaxis()
ax2.xaxis.set_major_formatter(E5); ax2.set_xlabel("gain from adding a member already in the stack (x 1e-5)")
ax2.legend(loc="lower right", fontsize=11)
n_dis = sum(np.sign(i_) != np.sign(o_) for _, i_, o_ in INOUT)
panel_title(ax2, f"Inner and outer verdicts disagree on {n_dis} of 5 folds", "logistic re-weighting of the wide stack + megayak LM")
headline(fig, "Where you select and where you score must be different rows",
         "Left: a blend picked on the OOF it is scored on. Right: a decision taken inside one fold versus across folds.")
plt.show()
````

**Takeaway:** select on some rows, score on others; only then does a score tell you what to expect on new rows.

<div style="background: #fff4ec; color: #1a1a1a; border-left: 5px solid #eb6834; padding: 10px 14px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a;">Rules of thumb for an honest CV.</b> (1) Stop trees on a held-out slice of the training part, never on the scored
fold. (2) Every choice a pipeline makes (members, weights, C, gates, transforms) is made inside the training part of
each outer fold: that is <i>nested</i> CV. (3) Decide across all outer folds, not inside one: a fold-local test sees
that fold's own calibration and noise. (4) A stack fitted once on the full OOF and then reported on the same OOF reads
high; fit it again fold by fold to get the number you can compare.
</div>

<a id="s6"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">6 &middot; What did not work</div>

Negative results are rarely published and save everyone time. First, public probes that the ladder authors reported
as losses (all credit to them for publishing these):

| probe on a ladder file | public change | reported by |
|---|---|---|
| + 0.12 of the public target-first GPT-2 (already inside every ladder file) | -14 | goodpjw2008 (v1 probes) |
| + 5% / + 2% TabPFN-3.5 | -1 / -1 | Areeb Ali (v18 and probes) |
| 10% mix with another public stack | -1 | Areeb Ali |
| + 10% / + 20% of a GPT-2 + GLM fusion file | 0 / 0 | Areeb Ali |
| more of both LMs (+0.04, +0.03) | -2 | Areeb Ali |
| an extra LM step on daoviet's file, other LM kept at 0.12 | 0 | goodpjw2008 |

Areeb Ali's summary: *"new model families (TabPFN, GBDT files, other stacks) lose a point"*. On nested CV, adding more
language-model weight to a wide stack also loses (-2.8e-5, up on only 2 of 5 folds in our check), even though public
ladders gained points from LM steps.

Second, "smarter" ways to combine a bank of public OOF predictions, each compared on **nested CV over 5 outer folds**
with every setting fixed in advance, against simply keeping a strong logistic stack:

| method | nested CV vs. the plain stack | folds better |
|---|---|---|
| greedy hill climbing (Caruana), bank that already contains the stack | +0.0 (it picks the stack alone) | 0/5 |
| logistic regression on all ~80 deduplicated members, C by inner CV | -27.7e-5 | 0/5 |
| weights proportional to pairwise AUC | -40.6e-5 | 0/5 |
| two experts gated by a data segment | -30.9e-5 | 0/5 |
| stack + a small segment-gate correction | -1.9e-5 | 0/5 |
| non-linear meta-model (4-leaf LightGBM on members, member disagreement, raw columns) | -39e-5 to -56e-5 | 0&ndash;1/5 |

<div style="background: #eefaf4; color: #1a1a1a; border-left: 5px solid #1baf7a; padding: 10px 14px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a;">Pattern.</b> Once a linear stack holds the strong members, cleverer combiners mostly fit noise: they lose on
every outer fold by far more than any ladder step gains. The one kind of change that kept gaining on nested CV in the
public record is <b style="color: #1a1a1a;">new information</b> (another seed or another view of a strong model), not new weights.
</div>

<a id="s7"></a>
<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">7 &middot; What to expect on private, and how to pick two final files</div>

Put the pieces together for the current public top file. We replay its checkable steps on the nested OOF, in the
order they were built, and set them beside the public points each one earned.

````python
D_a3 = 0.12 * (O["LM"] - W)
D_352 = 0.09 * (O["LM"] - W) + 0.09 * (O["L7F"] - W) + 0.05 * (O["GPD"] - W) - 0.05 * (O["MLP"] - W) - 0.025 * (O["GB"] - W)
c_a3, c_352 = W + D_a3, W + D_352
c_354 = L(c_352) + 0.06 * (O["G26"] - O["MS"])
CHAIN = [("Strike 9 -> Strike 10", 10, None, None), ("+ 0.6 (S10 - S9)", 2, None, None),
         ("+ 0.12 megayak LM", 1, (cv(c_a3) - CV_W) * 1e5, folds_up(c_a3, W)),
         ("-> 0.95352 file\n(4 steps)", 9, (cv(c_352) - cv(c_a3)) * 1e5, folds_up(c_352, c_a3)),
         ("+ GPT-2 seed swap", 2, (cv(c_354) - cv(c_352)) * 1e5, folds_up(c_354, c_352))]
CHECK_PUB = sum(c[1] for c in CHAIN if c[2] is not None); CHECK_CV = sum(c[2] for c in CHAIN if c[2] is not None)
table(pd.DataFrame({"step of the 0.95354 ladder": [c[0].replace(chr(10), " ") for c in CHAIN], "public points": [f"{c[1]:+d}" for c in CHAIN],
                    "nested CV (points)": ["no OOF" if c[2] is None else f"{c[2]:+.1f}" for c in CHAIN],
                    "folds up": ["&ndash;" if c[3] is None else f"{c[3]}/5" for c in CHAIN]}),
      "The 0.95354 ladder, step by step", f"Checkable steps: public {CHECK_PUB:+d} points, nested CV {CHECK_CV:+.1f} points.",
      right=["public points", "nested CV (points)", "folds up"])
````

````python
fig, ax = plt.subplots(figsize=(13, 5.0))
x = np.arange(len(CHAIN))
ax.bar(x - 0.2, [c[1] for c in CHAIN], width=0.38, color=PUB, label="public LB points earned")
for i, (lab, p, c, u) in enumerate(CHAIN):
    ax.text(i - 0.2, p + 0.3, f"+{p}", ha="center", va="bottom", fontsize=12, fontweight="bold", color=INK)
    if c is None:
        ax.bar(i + 0.2, 4, width=0.38, color="none", edgecolor=MUTED, hatch="///", lw=0)
        ax.text(i + 0.2, 4.3, "no OOF:\nuntestable", ha="center", va="bottom", fontsize=10.5, color=INK2)
    else:
        ax.bar(i + 0.2, c, width=0.38, color=CV if (c > 0 and u >= 4) else PRIV)
        ax.text(i + 0.2, c + (0.3 if c >= 0 else -0.3), f"{c:+.1f}\n({u}/5)", ha="center", va="bottom" if c >= 0 else "top", fontsize=11, color=INK)
zero_line(ax); ax.set_xticks(x); ax.set_xticklabels([c[0] for c in CHAIN], fontsize=11.5); ax.yaxis.set_major_formatter(E5)
ax.set_ylabel("points (x 1e-5)"); ax.grid(axis="x", visible=False)
ax.set_ylim(min(-8, min(c[2] for c in CHAIN if c[2] is not None) - 3), 14)
ax.legend(handles=[Patch(color=PUB, label="public LB points earned"), Patch(color=CV, label="nested CV: confirmed"),
                   Patch(color=PRIV, label="nested CV: not confirmed"), Patch(facecolor="none", edgecolor=MUTED, hatch="///", label="no OOF published")],
          loc="upper right", fontsize=11)
headline(fig, f"Of the +24 public points from Strike 9 to the top file, the checkable +{CHECK_PUB} are worth {CHECK_CV:+.1f} on nested CV",
         "Each step of the 0.95354 ladder file replayed on the nested OOF of the public wide LM stack (moves applied in the same order)")
plt.show()
````

**Takeaway:** most of the public climb from Strike 9 cannot be checked, and the part that can is worth little on nested CV.

**What this means for the private board** (an estimate, not a forecast):

* The absolute private score of every file can sit a few *thousandths* away from its public score, in the same
  direction for everyone. Nobody can know that direction; it does not change the ranking.
* The **ranking** inside the top band depends on gaps of a few points, and each gap carries a public noise of about
  one ladder-rung SD. Expect the top of the private board to reshuffle by that much.
* Steps that nested CV did not confirm have an expected private value near their CV value, not their public value.
  Steps without any OOF (the base files and the extrapolation) are a genuine unknown: they may hold, or not.
* The two-seed rung is the exception that proves the method: it won on both boards and has a mechanism.

<div style="background: #eefaf4; color: #1a1a1a; border-left: 5px solid #1baf7a; padding: 10px 14px; border-radius: 6px; margin: 6px 0;">
<b style="color: #1a1a1a;">Picking two final files &mdash; general advice.</b><br>
<b style="color: #1a1a1a;">1. One CV-backed file.</b> Every choice in it (members, weights, early stopping) made inside the training folds; its
nested CV is the number you trust. Its public score may sit below the ladders; that is fine.<br>
<b style="color: #1a1a1a;">2. One different hypothesis.</b> If you believe a public ladder's untestable base carries real signal, that is a
legitimate second pick &mdash; but make the second file genuinely <i>different</i> from the first (lower rank
correlation), so the two picks hedge each other.<br>
<b style="color: #1a1a1a;">3. Never two near-copies.</b> Two files at rank correlation 0.9999 land within a few points of each other on
private: the second slot is wasted.<br>
<b style="color: #1a1a1a;">4. Do not let the public board choose between close files.</b> If two candidates differ by less than about two
noise SDs (use <code style="color: #1a1a1a; background: #d5efe2;">public_gap_noise()</code> above), the public board cannot tell them apart; decide by CV or by
reasoning.

<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">9 &middot; Anatomy of the 0.95386 and 0.95388 ladders (v4)</div>

Both new top notebooks publish every probe with its public score and an OOF check of each mix, which is what makes
the replay below possible. Thank you to **goodpjw2008** and **Areeb Ali** again, and to **Aman Atar** for the bundle
both are built on.

* **goodpjw2008, [LM Ladder | Public and CV Up | LB 0.95386](https://www.kaggle.com/code/goodpjw2008/s6e11-lm-ladder-public-and-cv-up-lb-0-95386).**
  Starts from Aman Atar's public 0.95370 file (with 0.06 of one LM moved to the other) and moves it towards 87 public
  members. The weights maximise *OOF change + mu x a forecast of the public change*, where the forecast is a model
  fitted on 44 of the author's own scored files (not shipped). Three settings were submitted: careful 0.95376,
  balanced 0.95380, **bold 0.95386**.
* **Areeb Ali, [Bold, Then Bolder | LB 0.95388](https://www.kaggle.com/code/areebbb/s6e11-bold-then-bolder-lb-0-95388).**
  Rebuilds the three mixes byte for byte and probes eight steps beyond bold on the board. Keeps
  **bold + 0.5 (bold - balanced) = 0.95388**.

```
U0    = rank(AT + 0.06 (LM7 - LM))                         # AT = the public 0.95370 file (Aman Atar)
mix   = (1 - sum_k w_k) logit(U0) + sum_k w_k member_k      # 87 members, logit ranks
w     = w_bold + 0.5 (w_bold - w_balanced)                  # 0.95388
```

Every mix is linear in its weights, so every probe of both notebooks can be rebuilt on the test rows (for its noise)
and on the OOF rows (for its CV) from the same public inputs. The base file has no OOF of its own; its OOF stand-in is
goodpjw2008's replay of it from the 16 bundle models and the wide LM stack (R&sup2; shown in the log above).

**What the anatomy says.**

* **The base switch (+16) is a different file, not a step.** The 0.95370 file is Aman Atar's; its rank correlation with
  the 0.95354 ladder file is far from 1, so its public noise is larger, and it has no OOF. It is the one link in the new
  ladders that nobody can check.
* **The three mixes are real moves with OOF behind them.** They are large moves (public noise SD of about 8 to 13
  points), and their OOF gains (+19 to +33 in-sample, +14 to +23 in the authors' nested CV) are larger than that noise

* **Areeb Ali's eight probes are inside the noise.** Their public changes (-1 to +2) are all within one noise SD of

  loses.

<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">10 &middot; "Public and CV up" &mdash; checked fold by fold</div>

goodpjw2008's claim is that the mixes raise **both** the public score and CV. Two checks on our replay:

1. **Fold by fold.** A CV-confirmed step should gain on at least 4 of 5 folds, by a margin larger than the
   fold-to-fold spread. The mix weights saw all five folds, so these are in-sample numbers (the authors' nested
   numbers, with weights refitted on four folds, are shown beside them).
2. **Nested choice of the step size.** Areeb Ali chose *how far* to go along `bold - balanced` on the public board.
   Here the four other folds choose it, and the fifth scores it, for each fold in turn.

````python
# (2) nested step size along d = bold - balanced: t chosen on four folds, scored on the fifth, gain vs the bold mix.
TGRID = np.round(np.arange(0, 2.01, 0.25), 2)
_d = PO["bold"] - MO["balanced"]
TG = np.array([[auc_sort(y[b], (PO["bold"] + t * _d)[b]) for b in FOLDS] for t in TGRID]) * 1e5
TG = TG - TG[0]                                          # gain over bold, (t, fold)
T_OUT, T_PICK = np.zeros(5), np.zeros(5)
for f in range(5):
    j = TG[:, [g for g in range(5) if g != f]].mean(1).argmax(); T_PICK[f], T_OUT[f] = TGRID[j], TG[j, f]
CHK10 = [("careful", "ref"), ("balanced", "ref"), ("bold", "ref"), ("bold + 0.5 (bold - balanced)", "bold"),
         ("1.5 x bold", "bold"), ("bold + 1.5 (bold - balanced)", "bold"), ("388 + 5% GP + DVAE", "bold + 0.5 (bold - balanced)"),
         ("388 + 5% shallow tree", "bold + 0.5 (bold - balanced)")]
LAB10 = {"careful": "careful mix vs 0.95370", "balanced": "balanced mix vs 0.95370", "bold": "bold mix vs 0.95370",
         "bold + 0.5 (bold - balanced)": "0.95388 step vs bold", "1.5 x bold": "1.5 x bold vs bold",
         "bold + 1.5 (bold - balanced)": "bold + 1.5 (bold - bal.) vs bold",
         "388 + 5% GP + DVAE": "+5% GP + DVAE on 0.95388", "388 + 5% shallow tree": "+5% shallow tree on 0.95388"}
V10 = []
for a, b in CHK10:
    c, d = chg(a, b); V10.append((LAB10[a], c, d, (d > 0).sum(), d.std(ddof=1)))
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 5.8), gridspec_kw=dict(width_ratios=[1.45, 1]))
yy = np.arange(len(V10))[::-1]
for yi, (lab, c, d, up, sd) in zip(yy, V10):
    hold = up >= 4 and abs(c) > sd / np.sqrt(5) * 2
    col = CV if (hold and c > 0) else PRIV if c < 0 and up <= 1 else MUTED
    ax.plot([d.min(), d.max()], [yi, yi], color=RULE, lw=2.5, solid_capstyle="round", zorder=1)
    ax.scatter(d, np.full(5, yi), s=26, color=col, alpha=0.8, edgecolor="none", zorder=3)
    ax.plot(c, yi, "|", ms=16, mew=2.5, color=INK, zorder=4)
    ax.text(1.01, yi, f"{c:+.1f}  {up}/5", va="center", ha="left", fontsize=11, color=CVTXT if col == CV else INK2, fontweight="bold" if col == CV else "normal",
            transform=ax.get_yaxis_transform(), clip_on=False)
ax.set_yticks(yy); ax.set_yticklabels([v[0] for v in V10], fontsize=11.5); ax.tick_params(axis="y", length=0)
zero_line(ax, "x"); ax.xaxis.set_major_formatter(E5); ax.set_xlim(-14, max(v[2].max() for v in V10) + 3); ax.grid(axis="y", visible=False)
ax.set_xlabel("OOF AUC change per fold (x 1e-5); black tick = change on all rows")
ax.legend(handles=[Line2D([], [], marker="o", ls="", color=CV, label="holds: >= 4/5 folds, clear of the fold spread"),
                   Line2D([], [], marker="o", ls="", color=MUTED, label="fold noise: small or mixed"),
                   Line2D([], [], marker="o", ls="", color=PRIV, label="down on >= 4 folds")], loc="upper center", bbox_to_anchor=(0.5, -0.16), ncol=3, fontsize=10.5)
panel_title(ax, "The mixes hold on every fold; small steps are mostly fold noise", "five dots = five folds")
for f in range(5): ax2.plot(TGRID, TG[:, f], color=C[[0, 6, 3, 4, 5][f]], lw=1.3, alpha=0.75, label=f"fold {f}")
ax2.plot(TGRID, TG.mean(1), color=INK, lw=2.4, label="mean of the folds")
ax2.axvline(0.5, color=PUB, lw=1.2, ls=(0, (3, 2))); ax2.text(0.55, ax2.get_ylim()[1] * 0.55, "public pick\nt = 0.5", color=PUB, fontsize=10.5, va="top")
zero_line(ax2); ax2.yaxis.set_major_formatter(E5)
ax2.set_xlabel("t in  bold + t (bold - balanced)"); ax2.set_ylabel("OOF change vs bold, per fold (x 1e-5)"); ax2.legend(loc="upper left", fontsize=10.5, ncol=2)
panel_title(ax2, f"Nested step: {T_OUT.mean():+.1f}, {(T_OUT > 0).sum()}/5 up", "inner folds pick t = " + ", ".join(f"{t:g}" for t in T_PICK))
headline(fig, "\"Public and CV up\" holds for the three mixes on every fold; the kept half step is small, and one fold carries about half of it",
         "OOF rows of the public files; the mix weights saw all five folds (in-sample), the step size on the right is chosen nested")
plt.show()
````

### What each new notebook offers, and what its public tuning is likely to cost on private

For every step that was **kept because it won on the board**, the public number is the best of several. The model of
section 4, now run on the real probes of each author: on each random 20/80 split of the OOF rows (real public and
private sizes), the probe with the best public gap is kept, and its private gap is read. The difference is the
**give-back**. A step's expected private change is its public change minus the give-back, with a 90% interval from the
same draws. Steps chosen without the board, or scored once, get the plain noise interval.

````python
def interval_k1(g, sd):
    s = np.sqrt(sd ** 2 + (sd * np.sqrt(NPUB / (NT - NPUB))) ** 2); return g, g - 1.645 * s, g + 1.645 * s
VERD = []
gp = GIVE["goodpjw2008"]; ga = GIVE["Areeb Ali"]
r_base = T9.iloc[0]
VERD.append(("Aman Atar / goodpjw2008", "0.95354 -> rank-1 q17 file", 16, *interval_k1(16, r_base.sd), np.nan, np.nan, "probes unknown: plain noise", -1))
for lab, gv, fam, oof_n, ref_n in [("goodpjw2008", 16, gp, "bold", "ref"), ("Areeb Ali", 2, ga, "bold + 0.5 (bold - balanced)", "bold")]:
    e = gv - fam["back"]
    VERD.append((lab, "bold mix (best of 5)" if lab == "goodpjw2008" else "half step (best of 8)", gv, e.mean(), np.percentile(e, 5), np.percentile(e, 95),
                 chg(oof_n, ref_n)[0], NESTED_AUTH.get(oof_n, np.nan), f"best of {fam['K']}: give-back {fam['back'].mean():+.1f}", int((chg(oof_n, ref_n)[1] > 0).sum())))
for lab, n_, pub in [("Sameer Shaikh (Part 3)", "388 + 5% GP + DVAE", -3), ("Sameer Shaikh (Part 3)", "388 + 5% shallow tree", -1)]:
    rr = np.corrcoef(rankdata(PO[n_]), rankdata(MO["388"]))[0, 1]
    VERD.append((lab, n_.replace("388", "0.95388"), pub, *interval_k1(pub, sig(rr)), chg(n_, "bold + 0.5 (bold - balanced)")[0], np.nan, "not kept: plain noise",
                 int((chg(n_, "bold + 0.5 (bold - balanced)")[1] > 0).sum())))
_c390, _d390 = chg("390", "bold + 0.5 (bold - balanced)")
VERD.append(("Aman Atar (new in v6)", "0.95388 -> 0.95390 (>= 3 files at 0.95390)", 2, *EXP390["K3"], _c390, np.nan,
             f"best of >= 3: give-back {GIVE390:+.1f}", int((_d390 > 0).sum())))
VT = pd.DataFrame(VERD, columns=["notebook", "step", "public", "exp_private", "lo", "hi", "oof", "nested", "model", "up"])
def verdict(r):
    if np.isnan(r.oof): return "unknown: no OOF"
    if r.oof >= 10 and r.up >= 4: return "real gain on OOF;\npublic alone is weak evidence"
    if r.oof > 0 and r.up >= 4: return "tiny gain on OOF;\npublic number is noise"
    if r.oof <= 0: return "no gain on OOF"
    return "fold noise"
VT["verdict"] = VT.apply(verdict, axis=1)
showv = pd.DataFrame({"notebook": VT.notebook, "step": VT.step, "public change": VT.public.map(lambda v: f"{v:+d}"),
                      "expected private change (90% interval)": [f"{m:+.1f}  [{l:+.1f}, {h:+.1f}]" for m, l, h in zip(VT.exp_private, VT.lo, VT.hi)],
                      "how": VT.model, "OOF change (in-sample)": [("-" if np.isnan(v) else f"{v:+.1f} ({u}/5 up)") for v, u in zip(VT.oof, VT.up)],
                      "authors' nested": VT.nested.map(lambda v: "-" if np.isnan(v) else f"{v:+.1f}"), "verdict": VT.verdict.str.replace("\n", " ")})
table(showv, "Verdicts: what each public step is likely to keep on private", "Points of 1e-5. 90% interval from the public and private noise of the step; "
      "give-back = the expected public luck of keeping the best of several probes (section 4). The 0.95390 OOF change uses the replayed OOF (section 12).")
````

````python
fig, ax = plt.subplots(figsize=(13, 6.3))
yy = np.arange(len(VT))[::-1]
for yi, r in zip(yy, VT.itertuples()):
    ax.plot([r.lo, r.hi], [yi - 0.12, yi - 0.12], color=PRIV, lw=3, solid_capstyle="round", alpha=0.45, zorder=2)
    ax.plot(r.exp_private, yi - 0.12, "o", ms=9, color=PRIV, mec=SURF, mew=1.4, zorder=4)
    ax.plot(r.public, yi + 0.14, "o", ms=8, color=PUB, mec=SURF, mew=1.4, zorder=4)
    if not np.isnan(r.oof): ax.plot(r.oof, yi + 0.14, "D", ms=7, color=CV, mec=SURF, mew=1.2, zorder=4)
    if not np.isnan(r.nested): ax.plot(r.nested, yi + 0.14, "D", ms=7, mfc=SURF, mec=CVTXT, mew=1.6, zorder=5)
    ax.text(1.015, yi, r.verdict, va="center", ha="left", fontsize=11, color=INK, transform=ax.get_yaxis_transform(), clip_on=False,
            fontweight="bold" if r.verdict.startswith("real") else "normal")
ax.set_yticks(yy); ax.set_yticklabels([f"{a}\n{b}" for a, b in zip(VT.notebook, VT.step)], fontsize=11); ax.tick_params(axis="y", length=0)
zero_line(ax, "x"); ax.xaxis.set_major_formatter(E5); ax.set_xlim(-32, 52); ax.grid(axis="y", visible=False)
ax.set_xlabel("change vs the file the step was built on (x 1e-5)")
ax.legend(handles=[Line2D([], [], marker="o", ls="", color=PUB, label="public change"),
                   Line2D([], [], marker="o", ls="-", color=PRIV, label="expected private change, 90% interval"),
                   Line2D([], [], marker="D", ls="", color=CV, label="OOF change (in-sample)"),
                   Line2D([], [], marker="D", ls="", mfc=SURF, mec=CVTXT, mew=1.6, label="authors' nested CV")],
          loc="upper center", bbox_to_anchor=(0.42, -0.16), ncol=4, fontsize=10.5)
headline(fig, "What each new notebook's public steps are likely to keep on private",
         "Expected private change = public change minus the give-back of picking the best public probe, simulated on the OOF rows (real 20/80 sizes)")
plt.show()
````

**Takeaway:** steps with OOF behind them keep most of their value; steps picked by the board alone are expected to give part of it back.

**Reading the verdicts.**

* **goodpjw2008's mixes offer real signal.** The bold mix gains on the OOF rows in all five folds and the authors' own
  nested CV keeps +19.5 of it. On the board alone, +16 is only about 1.3 noise SDs for a move this size, and keeping
  the best of five probes costs about six points in expectation, so the interval is wide; the OOF evidence is what
  makes it convincing. This is the strongest new public evidence since v3.
* **Areeb Ali's half step is small but real on OOF; the board cannot see it.** On OOF it adds about four points, up on
  5 of 5 folds (in-sample, one fold carrying about half of it). A nested choice of the step size also gains, on 4 of 5 folds,
  but prefers a longer step (t around 1.5), as the authors found on CV. The public +2 is the best of eight numbers with
  a noise of about three points, so the public score itself is no evidence: corrected for the selection, its expected
  private value is about -3 (the interval is wide because several of the eight probes are large moves, and any of them
  can win a random split). Take the file for the bold mix it contains, not for the last point.
* **The base switch has no OOF.** Aman Atar's 0.95370 file is a strong public file whose private value nobody can
  check from public data.
* **The two Part-3 probes on top of 0.95388** (Sameer Shaikh's team, both lost on the board) lose or barely move on
  OOF too: 5% more GP + DVAE loses on 4 of 5 folds, 5% more shallow tree gains under two points. In a file that already
  holds a member at a fitted weight, adding more of it does not help.

**What the error analysis says** (all on the OOF stand-in, so a guide rather than a forecast):

* **The top 1% is settled.** All four files catch the same share of frauds in their top 1%, 2% and 5% of rows (to a
  tenth of a percent). The steps between them hardly touch the order of the top rows.
* **The mixes gain lower down.** The step from the 0.95370 replay to the bold mix gains almost all of its AUC in rows
  the replay ranks *below* its top 5%: about a quarter of all frauds sit there, and the mix lifts them clearly more
  than the legit rows around them (left rank-shift panel). Its members carry signal about frauds that the base file
  ranks too low, which is also why the gain shows on every fold.
* **The change is spread over many rows.** The 1% of rows that move furthest are almost all legit rows in the middle
  of the score; they carry only a small part of the gain, and half of the rows must take the new ranks before most of
  it appears. These are not a few lucky rows.
* **The honest stack is still ahead on OOF.** Moving from the wide LM stack to the 0.95388 file's OOF counterpart
  loses, and again the loss sits below the top 1%, mostly in the 80&ndash;99% bands. The replay
  is an approximation (about 1% of the 0.95370 file has no OOF counterpart, and goodpjw2008 notes that the replay is
  if anything too hard on the file), so this is a measured hint, not a verdict on the file.
* **Calibration of the order.** A step that helps does not shift every fraud up; it shifts frauds up *relative to
  legit rows of the same band*. The rank-shift panels show that gap band by band. They are a description, not the
  whole story: in the third panel the 0.95388 file lifts low-ranked frauds more than legit rows *on average* and still
  loses AUC against the wide stack, because the AUC counts how many legit rows each fraud ends up above, row by row,
  not the average move.

<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">12 &middot; The 0.95390 curved ladder, step by step (v6)</div>

Thank you to **Aman Atar** for
[S6E11: 2nd-Order Curved Ladder Probit | LB 0.95390](https://www.kaggle.com/code/amanatar/s6e11-2nd-order-curved-ladder-probit-lb-0-95390)
(14 Nov), the new top public notebook. It builds on goodpjw2008's three mixes and Areeb Ali's 0.95388 file
(section 9) and prints every number it relies on, which is what makes the checks below possible.

**The recipe, as the notebook describes it.** The member weights are not published; the final file is read from the
author's dataset, which is not public.

```
z(v)    = Phi^-1((rank(v) - 0.5) / n)                         # probit of the rank instead of the logit
v1      = z_bold - z_balanced                                 # tangent: the last step of the ladder
v2      = (z_bold - z_balanced) - (z_balanced - z_careful)    # curvature: how the step itself changed
Sat B2  = 0.95388 file + a curved-ladder move + the wide LM stack            -> 0.95390
Sat B3  = another file                                                        -> 0.95390
twin    = 0.5 z(Sat B2) + 0.5 z(Sat B3)                       # midpoint of two files with the same score
Sat B4  = twin + a re-weighting of the 87 members, at right angles to the B2-B3 line   -> 0.95390 (published)
```

**The idea in plain words.** A ladder is a path of files: careful &rarr; balanced &rarr; bold &rarr; 0.95388. The usual
next rung repeats the last step in a straight line (the *tangent*). Aman Atar also measures how the step *changed*
from one rung to the next (a second difference, like acceleration) and follows a *curved* path. Everything is done
on the probit of the ranks rather than the logit: both keep each file's order, and they differ only in how strongly
the extreme tail counts when files are blended. Last, two files that scored the same 0.95390 are averaged, and a
re-weighting of the 87 members is added at right angles to the line joining them.

````python
# The author's printed OOF AUCs next to the ones recomputed here from public files (only the ladder files are public).
_rows = []
for k, pub in [("0.95370 file", "0.95370"), ("careful", "0.95376"), ("balanced", "0.95380"), ("bold", "0.95386"), ("0.95388", "0.95388"),
               ("Sat B2", "0.95390"), ("twin (B2 + B3) / 2", "not stated"), ("Sat B4 (published)", "0.95390")]:
    ours = _ours.get(k)
    _rows.append({"file": k, "public LB": pub, "OOF AUC, author's log": f"{AUTH390[k]:.6f}",
                  "OOF AUC, recomputed here": f"{ours:.6f}" if ours is not None else ("replay: %.6f" % OOFAUC["390"] if k.startswith("Sat B4") else "not public"),
                  "folds up vs 0.95388 (author)": "&ndash;" if k not in AUTH390_FOLDS or k == "0.95388" else
                  f"{sum(a > b for a, b in zip(AUTH390_FOLDS[k], AUTH390_FOLDS['0.95388']))}/5"})
table(pd.DataFrame(_rows), "The 0.95390 ladder: the author's numbers and ours",
      "All OOF numbers are in-sample: the weights of every file saw all five folds. The replay of the published file "
      f"reproduces {STEP_R2:.0%} of its step from the 0.95388 file (test rows, no labels), so its OOF AUC is close to, "
      "not equal to, the author's.", right=["public LB", "OOF AUC, author's log", "OOF AUC, recomputed here", "folds up vs 0.95388 (author)"])
````

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 5.6), gridspec_kw=dict(width_ratios=[1.35, 1]))
P_ = {k: np.array(v[:2]) for k, v in GEO.items()}
ss = np.linspace(-1.15, 1.75, 200)
_a1, _a2 = (P_["bold"] - P_["careful"]) / 2, (P_["bold"] + P_["careful"]) / 2 - P_["balanced"]
par = np.array([P_["balanced"] + s * _a1 + s * s * _a2 for s in ss])
ax.plot(par[:, 0], par[:, 1], color=C[6], lw=1.8, ls=(0, (5, 3)), zorder=2)
ax.plot([P_["balanced"][0], P_["bold"][0] + 1.25 * (P_["bold"][0] - P_["balanced"][0])], [0, 0], color=MUTED, lw=1.6, ls=(0, (2, 2)), zorder=1)
path = ["careful", "balanced", "bold", "0.95388"]
for a, b in zip(path[:-1], path[1:]):
    ax.annotate("", xy=P_[b], xytext=P_[a], arrowprops=dict(arrowstyle="-|>", color=PUB, lw=2, mutation_scale=16), zorder=3)
ax.annotate("", xy=P_["0.95390"], xytext=P_["0.95388"], arrowprops=dict(arrowstyle="-|>", color=PRIV, lw=2.4, mutation_scale=18), zorder=4)
LBL = {"careful": ("careful\n0.95376", (12, -2), "left"), "balanced": ("balanced\n0.95380", (-8, 14), "right"),
       "bold": ("bold\n0.95386", (0, 14), "center"), "0.95388": ("0.95388", (0, 12), "center"), "0.95390": ("0.95390\n(published)", (10, -6), "left")}
for k, (t, off, ha) in LBL.items():
    ax.plot(*P_[k], "o", ms=10 if k == "0.95390" else 8, color=PRIV if k == "0.95390" else NAVY, mec=SURF, mew=1.5, zorder=5)
    ax.annotate(t, P_[k], xytext=off, textcoords="offset points", ha=ha, va="bottom" if off[1] > 0 else "top", fontsize=11, color=INK,
                fontweight="bold" if k == "0.95390" else "normal")
ax.text(par[-1, 0] + 0.5, par[-1, 1], "curved path\n(parabola through\ncareful, balanced, bold)", fontsize=10.5, color=C[6], va="center")
ax.text(P_["bold"][0] + 1.25 * (P_["bold"][0] - P_["balanced"][0]), 0.6, "straight tangent", fontsize=10.5, color=INK2, ha="right", va="bottom")
ax.text(0.02, 0.04, f"the 0.95390 file also leaves this plane by {GEO['0.95390'][2]:.1f}\n(0.95388: {GEO['0.95388'][2]:.1f}); distances in RMS probit units x 1e-3",
        transform=ax.transAxes, fontsize=10.5, color=INK2, va="bottom")
ax.set_aspect("equal", adjustable="datalim"); ax.set_xlabel("along the tangent  bold - balanced  (x 1e-3)"); ax.set_ylabel("across it (x 1e-3)")
panel_title(ax, "A  The ladder bends, and the 0.95390 step follows the bend", "files on the test rows, in the probit-of-rank space the author uses")
u = 1 - np.logspace(-5, np.log10(0.5), 300)
ax2.plot(1 - u, norm.ppf(u), color=PUB, lw=2.4, label="probit  z = Φ⁻¹(u)")
ax2.plot(1 - u, logit(u) / 1.7, color=PRIV, lw=2.4, ls=(0, (5, 2)), label="logit(u) / 1.7")
ax2.axvspan(1e-5, y.mean(), color=BAND, zorder=0, lw=0)
ax2.text(0.012, 0.3, f"top {y.mean():.2%} of rows:\nwhere most frauds sit", fontsize=10.5, color=INK2, va="bottom", ha="left")
ax2.set_xscale("log"); ax2.invert_xaxis(); ax2.set_xlim(0.5, 1e-5)
ax2.set_xlabel("share of rows ranked above (1 - u)"); ax2.set_ylabel("value on the blending scale")
ax2.legend(loc="upper left", fontsize=11)
panel_title(ax2, "B  Probit vs logit: same order", f"a 50/50 blend built on either scale: 1 - ρ = {PROBIT_VS_LOGIT:.0e}")
headline(fig, "The idea: follow the ladder's curve, on the probit scale",
         "Left: the public ladder files and the 0.95390 file in two dimensions. Right: the two blending scales agree except in the far tail.")
plt.show()
````

**Takeaway:** the 0.95390 step follows the bend of the ladder, and the probit-versus-logit choice barely changes the order.

**How to read it.** The ladder does bend: careful &rarr; balanced &rarr; bold is not a straight line, and the 0.95390
file sits near the continuation of the parabola through the three mixes, not on the straight tangent. The author's
"orthogonal" curvature is in fact mostly aligned with the tangent (cosine **0.75**, the same number his notebook
prints), so most of the second-order term is *acceleration along the ladder*, plus a sideways bend. The choice of
probit over logit barely matters for these files: built on either scale, the same 50/50 blend has the same order up
to 1 &minus; &rho; of about 3e-11, far below any noise. Last, the "twin": two files with **equal public scores** do not mark a maximum between
them; with a public noise of about five points per step of this size, many nearby files would also score 0.95390.
Averaging two similar files is still a sensible, mild hedge.

````python
# Verdict of the step, in parts. Parts are found WITHOUT labels on the test rows (the replayed step regressed on the
# ladder's tangent and curvature) and applied to the OOF rows. In-sample = the weights saw all five folds.
_lab = {"the whole step": "whole step  0.95388 -> 0.95390", "along the ladder (tangent)": "part: along the tangent",
        "curvature (2nd order)": "part: curvature", "off the ladder plane (re-weighting)": "part: off the ladder plane"}
_r12 = []
for n in ["the whole step", "curvature (2nd order)", "along the ladder (tangent)", "off the ladder plane (re-weighting)"]:
    g = PART_GAIN[n]
    _r12.append({"move on the OOF rows": _lab[n], "kind": "in-sample", "change, all folds": f"{g.mean():+.1f}", "folds up": f"{(g > 0).sum()}/5",
                 "per fold": "  ".join(f"{v:+.1f}" for v in g)})
_r12.append({"move on the OOF rows": "step size along the whole step", "kind": "nested", "change, all folds": f"{T3_OUT.mean():+.1f}",
             "folds up": f"{(T3_OUT > 0).sum()}/5", "per fold": "  ".join(f"{v:+.1f}" for v in T3_OUT)})
for lam, g in zip(LAMS, RIDGE_OUT):
    _r12.append({"move on the OOF rows": f"re-weighting the 87 members, penalty {lam:g}", "kind": "nested", "change, all folds": f"{g.mean():+.1f}",
                 "folds up": f"{(g > 0).sum()}/5", "per fold": "  ".join(f"{v:+.1f}" for v in g)})
table(pd.DataFrame(_r12), "The 0.95390 step on the OOF rows (changes vs the 0.95388 file, x 1e-5)",
      f"{SHARE_PLANE:.0%} of the step lies in the plane of the ladder: {_ab[1]:.2f} x the curvature and {_ab[0]:+.2f} x the tangent. "
      "Nested = chosen on four folds and scored on the fifth: the step size along the published step (grid 0 to 3), or a ridge "
      "logistic re-weighting of the same 87 members on top of the 0.95388 file (all three penalties shown, none selected).",
      right=["change, all folds", "folds up", "per fold"])
````

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 6.0), gridspec_kw=dict(width_ratios=[1.45, 1]))
ROWS12 = [(_lab[n], g, "in") for n, g in PART_GAIN.items()]
ROWS12 = [ROWS12[3], ROWS12[1], ROWS12[0], ROWS12[2]] + [(f"nested re-weighting, penalty {lam:g}", g, "nest") for lam, g in zip(LAMS, RIDGE_OUT)]
yy = np.arange(len(ROWS12))[::-1]
for yi, (lab, g, kind) in zip(yy, ROWS12):
    up = (g > 0).sum(); col = CV if (up >= 4 and g.mean() > 0) else PRIV if up <= 1 else MUTED
    ax.plot([g.min(), g.max()], [yi, yi], color=RULE, lw=2.5, solid_capstyle="round", zorder=1)
    ax.scatter(g, np.full(5, yi), s=34, color=col, alpha=0.85, edgecolor="none", zorder=3, marker="o" if kind == "in" else "D")
    ax.plot(g.mean(), yi, "|", ms=18, mew=2.6, color=INK, zorder=4)
    ax.text(1.01, yi, f"{g.mean():+.1f}  {up}/5", va="center", ha="left", fontsize=11, color=CVTXT if col == CV else INK2,
            fontweight="bold" if col == CV else "normal", transform=ax.get_yaxis_transform(), clip_on=False)
ax.axhline(yy[3] - 0.5, color=RULE, lw=1)
ax.set_yticks(yy); ax.set_yticklabels([r[0] for r in ROWS12], fontsize=11); ax.tick_params(axis="y", length=0)
zero_line(ax, "x"); ax.xaxis.set_major_formatter(E5); ax.grid(axis="y", visible=False)
ax.set_xlabel("OOF AUC change vs the 0.95388 file (x 1e-5)")
ax.text(0.99, yy[3] - 0.42, "above: in-sample (circles)", transform=ax.get_yaxis_transform(), ha="right", va="bottom", fontsize=10.5, color=INK2)
ax.text(0.99, yy[3] - 0.58, "below: nested, fitted on 4 folds (diamonds)", transform=ax.get_yaxis_transform(), ha="right", va="top", fontsize=10.5, color=INK2)
panel_title(ax, "The gain comes from the curvature part", "five marks = five folds; black tick = mean")
for f in range(5): ax2.plot(TGRID3, TG3[:, f], color=C[[0, 6, 3, 4, 5][f]], lw=1.4, alpha=0.8, label=f"fold {f}")
ax2.plot(TGRID3, TG3.mean(1), color=INK, lw=2.6, label="mean")
ax2.axvline(1.0, color=PRIV, lw=1.4, ls=(0, (3, 2)))
ax2.text(1.06, TG3.max() * 0.62, "the published\nstep (t = 1)", color=PRIV, fontsize=10.5, va="top")
zero_line(ax2); ax2.yaxis.set_major_formatter(E5)
ax2.set_xlabel("t in  0.95388 + t (0.95390 - 0.95388)"); ax2.set_ylabel("OOF change per fold (x 1e-5)")
ax2.legend(loc="upper left", fontsize=10, ncol=2)
panel_title(ax2, "Longer steps keep gaining", "the direction was fitted on these rows")
headline(fig, "On the OOF rows the 0.95390 step gains on every fold, through its curvature part",
         "Replayed OOF of the published file; parts found on the test rows without labels; nested rows chosen on four folds and scored on the fifth")
plt.show()
````

**Takeaway:** the 0.95390 step is real on OOF, and its curvature part carries it.

**What the checks say.**

* **The step is real on OOF, and it is the curvature that carries it.** On the replayed OOF rows the whole step gains
  on 5 of 5 folds (the author reports the same on his own OOF). Split without labels, the curvature part alone gains on
  every fold, while the tangent part and the re-weighting outside the ladder's plane lose on their own.
* **These are in-sample numbers.** The author chose the weights on the same OOF rows, so the direction "keeps
  gaining" the further one goes along it (right panel): that is what a fitted direction does on its own rows, not proof
  that longer is better. The honest version is the nested re-weighting: fitted on four folds, a re-weighting of the same
  87 members gains on the fifth on **4 of 5 folds** (fold 1 loses every time). So the *kind* of move the 0.95390 file
  makes has support out of fold.
* **One caution, as in section 11.** The OOF side of the 0.95370 file is goodpjw2008's replay (R&sup2; 0.99), which
  the author of the replay calls too hard on the file. Moves away from it, towards the public members, may look better
  on these rows than on the real test rows.
* **The public +2 says nothing by itself.** It is about 0.4 noise SD for a step of this size, and the author reports at
  least three files at 0.95390; the expected private value of the public number is shown in section 9's verdict figure.

<div style="background: #0b1f4b; color: #ffffff; padding: 9px 16px; border-radius: 8px; font-size: 20px; font-weight: 700;">13 &middot; Can one honest step improve the best public file?</div>

This section's extension starts from **goodpjw2008's 0.95417** file (since v7; v4 and v5
started from Areeb Ali's 0.95388, v6 from Aman Atar's 0.95390, and no step passed there). On top of it we allow **at most
two steps**, and only steps that pass a rule fixed in v4, before looking at any result:

1. **Candidates.** A small step `B + w (M - B)` in logit-rank space towards (or away from) one public member `M` that
   has honest OOF on the shared five folds and is **not** one of the 87 members already in the mix: the 17 single
   models of mathsonchain's dataset, goodpjw2008's wide stack without LMs and his EBM bag. `w` is one of
   &plusmn;0.01, &plusmn;0.02, &plusmn;0.03, &plusmn;0.05. That is 152 candidate steps.
2. **Nested choice.** For each of the five folds in turn, the other four folds pick the best candidate (member and
   weight); the held-out fold scores it. The OOF side of the 0.95417 file is exact: its steps are moves of public members.
3. **Gate.** Accept a step only if it gains on **at least 4 of 5** held-out folds and on average. The public board is
   never consulted.

If a step passes, the extended 0.95417 file is written as `submission_extended.csv`, and its fold verdicts are shown below.
Since v12, `submission.csv` is the public-board file of [section v12](#s17) (public 0.95424); the v11 file of
[section v11](#s16) (public 0.95422) is `submission_v11_steered.csv`, and `submission_cv_backed.csv` the CV-backed alternative; in v10 it was BhuvanNR's step-8 file (public 0.95419), whose full
analysis will follow in a future version. **Our own CV-based reading favours
the CV-led file** (section v7: it is ahead of the 0.95417 file on OOF), so the notebook also writes
`submission_cv.csv`, the author's CV-led file rebuilt byte for byte, as a second output. The public best is written only
as a credited rebuild.

````python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw=dict(width_ratios=[1.35, 1]))
rd = EXT[0]
cols = [CV if g > 0 else PRIV for g in rd["outer"]]
ax.bar(np.arange(5), rd["outer"], color=cols, width=0.6, zorder=3)
for f, (g, (m, w)) in enumerate(zip(rd["outer"], rd["picks"])):
    ax.text(f, g + (0.4 if g >= 0 else -0.4), f"{g:+.1f}", ha="center", va="bottom" if g >= 0 else "top", fontsize=12, fontweight="bold", color=INK)
ylo = min(rd["outer"].min(), 0) - 7.5; yhi = max(rd["outer"].max(), 0) + 3
ax.set_ylim(ylo, yhi)
for f, (m, w) in enumerate(rd["picks"]):
    ax.text(f, ylo + 0.4, f"{m.split(' (')[0].replace(', ', ',\n')}\nw = {w:+.2f}", ha="center", va="bottom", fontsize=10.5, color=INK2)
zero_line(ax); ax.set_xticks(range(5)); ax.set_xticklabels([f"held-out fold {f}" for f in range(5)], fontsize=11)
ax.yaxis.set_major_formatter(E5); ax.set_ylabel("AUC change on the held-out fold (x 1e-5)"); ax.grid(axis="x", visible=False)
panel_title(ax, f"Step 1: {(rd['outer'] > 0).sum()}/5 folds up, mean {rd['outer'].mean():+.1f}: "
                f"{'gate passed' if rd['passed'] else 'gate not passed'}", "the step each set of four folds picked (bottom) and how it did on the fifth")
vals = [rd["final_in"], rd["outer"].mean()]
ax2.bar([0, 1], vals, color=[MUTED, CV if vals[1] > 0 else PRIV], width=0.55, zorder=3)
for i, v in enumerate(vals): ax2.text(i, v + (0.25 if v >= 0 else -0.25), f"{v:+.1f}", ha="center", va="bottom" if v >= 0 else "top", fontsize=13, fontweight="bold", color=INK)
zero_line(ax2); ax2.set_xticks([0, 1]); ax2.set_xticklabels([f"best of {N_CAND} steps,\nchosen and scored\non the same five folds", "the same search,\nnested (chosen on four,\nscored on the fifth)"], fontsize=11)
ax2.yaxis.set_major_formatter(E5); ax2.set_ylabel("OOF change (x 1e-5)"); ax2.grid(axis="x", visible=False)
lo2 = min(vals) - 2; ax2.set_ylim(min(lo2, -2), max(vals) + 2)
panel_title(ax2, "Selection inflates CV too", "the winner's curse of section 4, inside cross-validation")
headline(fig, "No candidate step passes the nested gate: the output stays the 0.95417 file" if N_STEPS == 0
         else f"{N_STEPS} step{'s' if N_STEPS > 1 else ''} pass{'' if N_STEPS > 1 else 'es'} the nested gate and {'are' if N_STEPS > 1 else 'is'} written as submission_extended.csv; the 0.95417 file itself is not changed",
         f"{N_CAND} small steps towards or away from {len(EXT_NAMES)} public members with honest OOF, chosen inside the folds; the public board is never used")
plt.show()
````

**Takeaway:** 2 honest steps pass the gate on top of the 0.95417 file (round 1: +0.05 towards lgb_glmboost (mathsonchain), held-out gain +4.9 on average, 4/5 folds up; round 2: +0.05 towards wide stack, no LM (goodpjw2008), held-out gain +4.4 on average, 4/5 folds up); they move it back towards the OOF optimum (+12.6 points in-sample). They are written as a separate file, submission_extended.csv.
