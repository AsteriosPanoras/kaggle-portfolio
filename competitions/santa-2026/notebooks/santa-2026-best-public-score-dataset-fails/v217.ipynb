```python
# Reusable helpers. Copy these into your own notebook: nothing here depends on this competition's files except the CSV format.
import csv, glob, json, os, time
csv.field_size_limit(1 << 30)

# --- where the data lives (Kaggle attaches inputs under /kaggle/input) ---------------------------------------------
ROOTS = ['/kaggle/input'] + [p for p in os.environ.get('SANTA_INPUT', '').split(':') if p]
DATASET = 'santa-2026-elf-programs-and-experiments'   # the only input this notebook reads
def find(name):
    for root in ROOTS:
        hits = sorted(glob.glob(f'{root}/**/{name}', recursive=True), key=lambda h: (DATASET not in h and 'release/ds' not in h, h))
        if hits: return hits[0]
    raise FileNotFoundError(f'{name}: attach the dataset {DATASET}')

def load_puzzles(path):
    """puzzles.csv -> {id: {start: [..], targets: [..], limit: int, family: str}}"""
    out = {}
    for r in csv.DictReader(open(path)):
        out[r['id']] = dict(start=json.loads(r['starting_values']), targets=json.loads(r['targets']),
                            limit=int(r['value_limit']), family=r.get('family', ''))
    return out

def read_programs(path):
    """Any CSV with the columns `id,program` (a submission, best_programs.csv, ...) -> {id: [[op, i, j, d], ...]}"""
    return {r['id']: json.loads(r['program']) for r in csv.DictReader(open(path))}

# --- helper 1: the exact replay (the verifier) ---------------------------------------------------------------------
def apply(op, a, b):
    return a + b if op == '+' else a - b if op == '-' else a * b if op == '*' else a // b

def replay(p, prog):
    """Strict replay of one program under the competition rules. Returns its length, or raises ValueError.
    Rules: 48 registers; every read register must hold a value; values stay positive integers <= value_limit;
    subtraction needs a > b; division must be exact; every target must appear in some register at some moment."""
    R = [None] * 48
    for k, v in enumerate(p['start']): R[k] = v
    seen = set(p['start']); L = p['limit']
    if len(prog) > 4096: raise ValueError('too long')
    for op, i, j, d in prog:
        if op not in ('+', '-', '*', '/'): raise ValueError(f'bad operator {op!r}')
        if not all(isinstance(x, int) and 0 <= x < 48 for x in (i, j, d)): raise ValueError('register out of range')
        a, b = R[i], R[j]
        if a is None or b is None: raise ValueError('undefined register')
        if op == '-' and a <= b: raise ValueError('non-positive subtraction')
        if op == '/' and a % b: raise ValueError('inexact division')
        v = apply(op, a, b)
        if not 0 < v <= L: raise ValueError('value out of range')
        R[d] = v; seen.add(v)
    if not set(p['targets']) <= seen: raise ValueError('missing target')
    return len(prog)

def is_valid(p, prog):
    try: replay(p, prog); return True
    except (ValueError, TypeError, IndexError): return False

# --- helper 2: per-puzzle merge of any number of `id,program` CSVs --------------------------------------------------
def merge_programs(paths, puzzles):
    """Keep, for every puzzle, the SHORTEST program that replays correctly across all the files in `paths`.
    Ties keep the earlier file. Returns (best, owner): best = {id: program}, owner = {id: file it came from}.
    Candidates that are not shorter than the current best are skipped before the (slower) replay."""
    best, owner = {}, {}
    for path in paths:
        for pid, prog in read_programs(path).items():
            if pid not in puzzles or (pid in best and len(prog) >= len(best[pid])): continue
            if is_valid(puzzles[pid], prog):
                best[pid], owner[pid] = prog, os.path.basename(path)
    return best, owner

# --- helper 3: compare two submissions per puzzle -------------------------------------------------------------------
def compare_submissions(a, b, puzzles=None, names=('A', 'B')):
    """a, b: {id: program}. Returns a table with one row per puzzle where the lengths differ (and prints the totals).
    If `puzzles` is given, an invalid program counts as missing (None) instead of trusting its length."""
    import pandas as pd
    rows = []
    for pid in sorted(set(a) | set(b)):
        la = len(a[pid]) if pid in a and (puzzles is None or is_valid(puzzles[pid], a[pid])) else None
        lb = len(b[pid]) if pid in b and (puzzles is None or is_valid(puzzles[pid], b[pid])) else None
        rows.append((pid, la, lb))
    t = pd.DataFrame(rows, columns=['id', f'len_{names[0]}', f'len_{names[1]}'])
    t['diff'] = t.iloc[:, 2] - t.iloc[:, 1]            # negative: B is shorter
    both = t.dropna()
    print(f'{names[0]}: {int(both.iloc[:, 1].sum()):,}   {names[1]}: {int(both.iloc[:, 2].sum()):,}   '
          f'(B - A = {int(both["diff"].sum()):+,} over {len(both)} puzzles; B shorter in {(both["diff"] < 0).sum()}, '
          f'A shorter in {(both["diff"] > 0).sum()}, {(both["diff"] == 0).sum()} equal)')
    return t[(t['diff'] != 0) | t['diff'].isna()].reset_index(drop=True)
```

```python
# Load the puzzles and the best programs, then replay every one of the 1,000 programs under the competition rules.
P = load_puzzles(find('puzzles.csv'))
programs = read_programs(find('best_programs.csv'))

t0 = time.time()
TOTAL = sum(replay(P[k], programs[k]) for k in P)          # raises if any program breaks a rule
assert len(programs) == 1000 and set(programs) == set(P)
print('Inputs: only the dataset', DATASET)
print(f'All 1,000 programs replay correctly: {TOTAL:,} instructions ({time.time() - t0:.1f} s).')
```

```python
if WRITE_SUBMISSION:
    with open('/kaggle/working/submission.csv', 'w', newline='') as f:
        w = csv.writer(f); w.writerow(['id', 'program'])
        for pid in sorted(programs): w.writerow([pid, json.dumps(programs[pid], separators=(',', ':'))])
    print(f'submission.csv written: {TOTAL:,} instructions')
else:
    print('WRITE_SUBMISSION is False: no submission file in this output. Copy & Edit and set it to True.')
```

```python
# Using the helpers on other files. Point `paths` at any number of `id,program` CSVs (yours, a public notebook's output, ...).
# Here the only file is the dataset's own best_programs.csv, so the merge must reproduce it exactly and the comparison is empty.
paths = [find('best_programs.csv')]
merged, owner = merge_programs(paths, P)
print(f'merge_programs: {len(merged)} puzzles, {sum(len(v) for v in merged.values()):,} instructions')
diff = compare_submissions(programs, merged, puzzles=P, names=('dataset', 'merged'))
print(f'puzzles where the two differ: {len(diff)}')
```

```python
# Analysis setup: plotting style and the analysis tables from the dataset.
import math
from collections import Counter, defaultdict
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
# One figure style for light AND dark pages: solid white figures with dark text, so they never depend on the page colour.
INK, MUTED, GRID = '#14181f', '#4a5160', '#dfe3ea'
GREEN, NAVY, RED, GOLD, BLUE = '#1f6f50', '#0b3d2e', '#b5452f', '#b07a00', '#2a6fb0'
plt.rcParams.update({'figure.facecolor': 'white', 'axes.facecolor': 'white', 'savefig.facecolor': 'white',
    'savefig.transparent': False, 'figure.dpi': 110, 'font.size': 12, 'axes.titlesize': 14.5, 'axes.titleweight': 'bold',
    'axes.labelsize': 12, 'axes.edgecolor': '#333333', 'axes.labelcolor': INK, 'xtick.color': INK, 'ytick.color': INK,
    'text.color': INK, 'axes.spines.top': False, 'axes.spines.right': False, 'legend.frameon': False, 'legend.fontsize': 11})
PR = load_puzzles(find('practice_puzzles.csv'))
best = pd.read_csv(find('best_programs.csv'))
experiments = pd.read_csv(find('experiments.csv'))
lineage = pd.read_csv(find('public_lineage.csv'))
headroom = pd.read_csv(find('headroom_independent_build.csv'))
structure = pd.read_csv(find('practice_structure_costs.csv'))
```

```python
# Per-target accounting used by the figures: for every target, the instructions spent since the previous new target.
rows = []; fan = Counter(); helper_share = []
for pid, prog in programs.items():
    p = P[pid]; T = set(p['targets']); seen = set(p['start']); R = [None] * 48; prod = [None] * 48
    for k, v in enumerate(p['start']): R[k] = v; prod[k] = -1
    last = 0; rank = 0; reads = Counter(); is_target = []
    for n, (op, i, j, d) in enumerate(prog):
        for s in {prod[i], prod[j]}:
            if s is not None and s >= 0: reads[s] += 1
        v = apply(op, R[i], R[j]); R[d] = v; prod[d] = n
        new = v in T and v not in seen
        is_target.append(new)
        if new:
            rows.append((pid, rank, v.bit_length(), n + 1 - last, len(p['start']))); last = n + 1; rank += 1
        seen.add(v)
    for n in range(len(prog)): fan[min(reads[n], 6)] += 1
    helper_share.append(sum(1 for n in range(len(prog)) if not is_target[n] and reads[n] >= 2) / len(prog))
gaps = pd.DataFrame(rows, columns=['id', 'rank', 'bits', 'cost', 'n_start'])
```

```python
# First-screen figure: where the instructions go.
b = (gaps.bits // 10 * 10).clip(lower=10, upper=50)
first = gaps['rank'] == 0
groups = [('first\ntargets', gaps[first].cost.sum())]
for lo in (10, 20, 30, 40, 50):
    m = (~first) & (b == lo)
    groups.append((f'{lo}-{lo+9} bit\ntargets' if lo < 50 else '50+ bit\ntargets', gaps[m].cost.sum()))
cols = [RED, '#cfe6da', '#9cc9b4', '#6fae92', GREEN, NAVY]
dark = (RED, GREEN, NAVY)
fig, ax = plt.subplots(figsize=(14, 3.7))
left = 0
for (name, v), c in zip(groups, cols):
    ax.barh(0, v, left=left, color=c, height=0.55, edgecolor='white')
    if v / TOTAL > 0.08:
        ax.text(left + v / 2, 0, f'{name}\n{v:,} ({v / TOTAL:.0%})', ha='center', va='center', fontsize=11.5,
                color='white' if c in dark else INK, fontweight='bold')
    else:
        ax.annotate(f'{name.replace(chr(10), " ")}: {v:,} ({v / TOTAL:.0%})', (left + v / 2, 0.28), xytext=(left + v / 2, 0.45),
                    ha='center', fontsize=11, arrowprops=dict(arrowstyle='-', color=MUTED))
    left += v
ax.set_xlim(0, TOTAL); ax.set_ylim(-0.4, 0.6); ax.set_yticks([]); ax.spines['left'].set_visible(False)
ax.set_title(f'Where the {TOTAL:,} instructions go: an expensive first target in every puzzle, then the big targets', loc='left')
ax.set_xlabel('instructions (each segment = instructions spent between consecutive new targets)')
plt.tight_layout(); plt.show()
```

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.65; margin: 8px 2px 12px 2px;">Each of the 1,000 puzzles gives a few small <b>starting values</b> in a bank of 48 registers and up to 32 large <b>targets</b> (up to 10<sup>18</sup>). A program is a list of instructions <code style="background: #e8edf5; color: #14181f; padding: 1px 6px; border-radius: 4px; font-size: 92%;">["op", i, j, d]</code>: read registers <i>i</i> and <i>j</i>, apply + &minus; &times; &divide;, write register <i>d</i>. Values must stay positive integers no larger than the puzzle's <code style="background: #e8edf5; color: #14181f; padding: 1px 6px; border-radius: 4px; font-size: 92%;">value_limit</code>, division must be exact, and every target must appear in some register at least once. <b>The score is the total number of instructions over all 1,000 programs; lower is better.</b> There is no hidden test set: the number you compute locally with <code style="background: #e8edf5; color: #14181f; padding: 1px 6px; border-radius: 4px; font-size: 92%;">replay</code> is the leaderboard number.</div>


```python
# Figure: the Data page's worked example, drawn as a register timeline.
steps = [('/', 0, 0, 1), ('-', 0, 1, 2), ('+', 0, 2, 3), ('+', 0, 3, 4)]
R = [3] + [None] * 4; targets = {1, 2, 5, 8}
fig, ax = plt.subplots(figsize=(13, 3.6))
for col in range(5):
    if col: op, i, j, d = steps[col - 1]; R[d] = apply(op, R[i], R[j])
    for r in range(5):
        v = R[r]
        fc = '#ffffff' if v is None else ('#ffe08a' if v in targets else '#dcebe3')
        ax.add_patch(FancyBboxPatch((col * 2.6, -r), 2.2, 0.8, boxstyle='round,pad=0.02', fc=fc, ec='#333333', lw=1))
        ax.text(col * 2.6 + 1.1, -r + 0.4, '' if v is None else str(v), ha='center', va='center', fontsize=14,
                fontweight='bold' if col and steps[col - 1][3] == r else 'normal')
    ax.text(col * 2.6 + 1.1, 1.25, 'start' if not col else
            f'step {col}: r{steps[col-1][1]} {steps[col-1][0]} r{steps[col-1][2]} -> r{steps[col-1][3]}',
            ha='center', fontsize=11.5)
for r in range(5): ax.text(-0.5, -r + 0.4, f'r{r}', ha='right', va='center', fontsize=12, color=MUTED)
ax.set_xlim(-1.2, 13.2); ax.set_ylim(-4.4, 1.8); ax.axis('off')
ax.set_title('A 4-instruction program: start [3] ends with the targets 1, 2, 5 and 8 (yellow) in registers', loc='left')
plt.tight_layout(); plt.show()
```

```python
# Figure: the families (practice set). Left: how close sorted targets are to each other. Right: how many targets can be
# built in <= 4 instructions from the starting values plus the smaller targets (exact search; E09 in section 6).
fams = ['independent', 'shared', 'multiplicative', 'mixed', 'tiny']; colors = [NAVY, GREEN, GOLD, BLUE, RED]
fig, ax = plt.subplots(1, 2, figsize=(15, 4.5))
near = []
for f in fams:
    rel = []
    for p in PR.values():
        if p['family'] != f: continue
        t = sorted(set(p['targets']))
        rel += [(b - a) / b for a, b in zip(t, t[1:])]
    near.append(100 * np.mean(np.array(rel) < 1e-4) if rel else 0)
ax[0].bar(fams, near, color=colors)
for k, v in enumerate(near): ax[0].text(k, v + 0.5, f'{v:.0f}%', ha='center', fontsize=12)
ax[0].set(title='Only "shared" puzzles hold near-duplicate targets', ylabel='% of neighbouring target pairs\nwithin 0.01% of each other', ylim=(0, max(near) * 1.25 + 1))
s = structure.copy(); s['cheap'] = s.cost_from_starts_and_earlier_targets.notna()
share = s.groupby('family').cheap.mean().reindex(fams)
ax[1].bar(fams, share * 100, color=colors)
for k, v in enumerate(share): ax[1].text(k, v * 100 + 1, f'{v:.0%}', ha='center', fontsize=12)
ax[1].set(title=f'Few targets are cheap to reach ({s.cheap.mean():.0%} overall)', ylabel='% of targets within 4 instructions\nof starts + smaller targets', ylim=(0, 115))
for a in ax: a.grid(axis='y', color=GRID); a.set_axisbelow(True); a.tick_params(axis='x', labelsize=11)
plt.tight_layout(); plt.show()
```

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #14181f; background: #e8f0fb; border-left: 6px solid #2a6fb0; padding: 12px 18px; border-radius: 10px; margin: 10px 0;"><b>A counting argument.</b> One instruction combines two of at most 48 register values with one of four operators, so it can reach at most about 2 &middot; 48<sup>2</sup> &asymp; 4,608 new values, about 12 bits of information.
