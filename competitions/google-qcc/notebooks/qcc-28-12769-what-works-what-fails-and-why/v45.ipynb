## The ideas, in plain words

What each change does and why it saves cost. These are the authors' own descriptions, shortened; the Almond Labs notebook has the full derivations.

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #51627a; border-radius: 8px; padding: 12px 16px; margin-top: 8px; max-width: 100%; overflow-wrap: anywhere; border: 1px solid #c9d2e0;"><b><code style="background: #e8edf5; color: #14181f; padding: 1px 5px; border-radius: 4px; font-size: 92%; overflow-wrap: anywhere;">LEND29</code>: lending a wire (invsqrt, -0.5 Toffoli per shot).</b> One wire can be computed from other wires by a simple parity rule, so it can be "lent" to a later step without spending a Toffoli to build it, and returned afterwards. The authors measure the mean saving as 0.5 Toffoli per shot. Checked by the authors over all 3,133,452 inputs.</div>

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #51627a; border-radius: 8px; padding: 12px 16px; margin-top: 8px; max-width: 100%; overflow-wrap: anywhere; border: 1px solid #c9d2e0;"><b><code style="background: #e8edf5; color: #14181f; padding: 1px 5px; border-radius: 4px; font-size: 92%; overflow-wrap: anywhere;">livescan</code> + <code style="background: #e8edf5; color: #14181f; padding: 1px 5px; border-radius: 4px; font-size: 92%; overflow-wrap: anywhere;">LDH</code>: holding a product in an idle wire (DQI, -32.8 Toffolis per shot).</b> A scan finds wires that are provably zero for a long stretch (58,431 intervals). A product that the circuit would compute twice is kept in such an idle wire instead of being recomputed. No extra qubit is needed because the wire was idle anyway. The result had 0 value and 0 phase errors on both local verifier salts.</div>

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #51627a; border-radius: 8px; padding: 12px 16px; margin-top: 8px; max-width: 100%; overflow-wrap: anywhere; border: 1px solid #c9d2e0;"><b>Recompiled angle loaders (chemistry, -6 Toffolis).</b> The chemistry circuit loads rotation angles from a table through small loader circuits. Three loaders were rebuilt in a cheaper cubic form, and the table words they read were edited only in cells that cannot change the result (the authors check that the two-body vectors stay identical).</div>

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #51627a; border-radius: 8px; padding: 12px 16px; margin-top: 8px; max-width: 100%; overflow-wrap: anywhere; border: 1px solid #c9d2e0;"><b>Weight refit and gauge polish (chemistry, about +0.0023 points).</b> The weights can be rotated by an orthogonal "gauge" without changing the physics; a constrained optimiser then lowers the cost factor lambda_eff from 46.9321 (ThemBones) to 46.9218 while the three accuracy limits still hold.</div>

<div style="font-family: 'Segoe UI', Roboto, 'Helvetica Neue', Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #51627a; border-radius: 8px; padding: 12px 16px; margin-top: 8px; max-width: 100%; overflow-wrap: anywhere; border: 1px solid #c9d2e0;"><b>A ticket re-rolls the DQI draw.</b> Adding a zero-cost tag line to the DQI file changes its hash and therefore the random test shots, without changing the circuit. Teams verify several tickets and ship one with no failed shot (a failed shot raises the cost by about 0.3 to 0.5 percent).</div>

```python
import zipfile, hashlib, pathlib, shutil
WRITE_SUBMISSION = False  # set False in the public version: nothing is written then
# Per task, the cheapest public files: invsqrt + chemistry from Rithvik Redrouthu's set, DQI from tonylica's set.
HASH = {"R_invsqrt": "d2148a7c", "R_npz": "e259454d", "R_angle0": "8df36485", "T_dqi": "4b572141"}
sha = lambda b: hashlib.sha256(b).hexdigest()
def hz(z, n):
    h = hashlib.sha256()
    with z.open(n) as f:
        for b in iter(lambda: f.read(1 << 23), b""): h.update(b)
    return h.hexdigest()
R = T = None
for p in sorted(pathlib.Path("/kaggle/input").rglob("submission*.zip")):
    try:
        z = zipfile.ZipFile(p)
        if len(z.namelist()) != 18: continue
        if hz(z, "invsqrt.kmb").startswith(HASH["R_invsqrt"]) and hz(z, "dfthcsos/submission.npz").startswith(HASH["R_npz"]) and hz(z, "dfthcsos/angle_k_0.kmb").startswith(HASH["R_angle0"]): R = (p, z)
        elif hz(z, "dqi.kmb").startswith(HASH["T_dqi"]): T = (p, z)
    except Exception as e: print("skip", p, e)
assert R and T, "both source file sets are needed (Add Input: the two source notebooks)"
print("invsqrt + chemistry from", R[0].parent.name, "| DQI from", T[0].parent.name)
names = R[1].namelist()
if WRITE_SUBMISSION:
    with zipfile.ZipFile("/kaggle/working/submission.zip", "w", zipfile.ZIP_DEFLATED) as out:
        for n in names:
            src = T[1] if n == "dqi.kmb" else R[1]
            with src.open(n) as fi, out.open(zipfile.ZipInfo(n), "w", force_zip64=True) as fo: shutil.copyfileobj(fi, fo, 1 << 23)
    with zipfile.ZipFile("/kaggle/working/submission.zip") as z:
        assert len(z.namelist()) == 18 and hz(z, "dqi.kmb").startswith(HASH["T_dqi"]) and hz(z, "invsqrt.kmb").startswith(HASH["R_invsqrt"])
    print("written /kaggle/working/submission.zip, 18 members, hashes checked")
```
