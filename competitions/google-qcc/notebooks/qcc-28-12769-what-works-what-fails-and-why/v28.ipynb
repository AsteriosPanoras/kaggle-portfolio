**Files** (Apache-2.0 public work): invsqrt `ecf908fe` (Leo Tang's v59 of maoxin123's 44-qubit circuit; generator line Syed Asad Ali), DQI `4471b493` (Koushik Rudra: 18 pair-window retentions on ozertuu's merge `e208d350`, which builds on Syed Asad Ali, Balram Srinivasan, Grant Prater, boristown, manwithacat, vumaniacom, Atakan Aldemir), chemistry `6ef06081` (Manvender Singh's tight-margin refit of leotang09's weights; loaders by Lera / marculera and maoxin123).


```python
import zipfile, hashlib, pathlib, shutil
WANT = {"invsqrt.kmb": "ecf908fe", "dqi.kmb": "4471b493", "dfthcsos/submission.npz": "6ef06081", "dfthcsos/angle_k_0.kmb": "ed648708"}
ZIP_SHA = "5a7d4c93c07a"
src = next(p for p in pathlib.Path("/kaggle/input").rglob("submission.zip") if "v60" in str(p).lower())
sha = lambda b: hashlib.sha256(b).hexdigest()
whole = sha(src.read_bytes())
print("source:", src, whole[:12])
z = zipfile.ZipFile(src)
assert len(z.namelist()) == 18, z.namelist()
for n, pre in WANT.items():
    h = sha(z.read(n)); print(f"{n:28s} {h[:12]}"); assert h.startswith(pre), (n, h[:8])
assert whole.startswith(ZIP_SHA), "unexpected submission.zip"
shutil.copyfile(src, "/kaggle/working/submission.zip")
print("written /kaggle/working/submission.zip", pathlib.Path("/kaggle/working/submission.zip").stat().st_size, "bytes")
```
