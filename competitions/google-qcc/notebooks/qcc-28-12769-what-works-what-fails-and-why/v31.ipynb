<div style="font-size: 16px; margin-top: 8px; color: #ffffff;">This fast version unpacks the current public frontier file set (ozertuu's dataset <i>quantum-sim-linalg-cache</i>), checks every file by SHA-256 and writes <code>submission.zip</code> in seconds. It does not re-run the verifiers. The full analysis version follows.</div>

**Files** (public work, credited in the datasets' licence notices): invsqrt `acc7b9c8` (maoxin123's 44-qubit circuit line, Syed Asad Ali generator, Ali Haydar Ozdag table, sub-ULP angle calibration), DQI `f56b04c7` (Syed Asad Ali line, Hayato Hikawa retention, Koushik Rudra pair-window, goodpjw2008 swap-collapse), chemistry `6ef06081` (Manvender Singh refit; loaders by Lera / marculera and maoxin123).


```python
import zipfile, hashlib, pathlib, lzma
WANT = {"invsqrt.kmb": "acc7b9c8", "dqi.kmb": "f56b04c7", "dfthcsos/submission.npz": "6ef06081"}
sha = lambda b: hashlib.sha256(b).hexdigest()
root = next(p for p in pathlib.Path("/kaggle/input").rglob("dqi.kmb.xz") if "linalg" in str(p)).parent
print("source:", root)
files = {}
for n in ["invsqrt.kmb", "dfthcsos/submission.npz"] + [f"dfthcsos/angle_k_{i}.kmb" for i in range(15)]:
    files[n] = (root / n).read_bytes()
files["dqi.kmb"] = lzma.decompress((root / "dqi.kmb.xz").read_bytes())
for n, pre in WANT.items():
    h = sha(files[n]); print(f"{n:28s} {h[:12]}"); assert h.startswith(pre), (n, h[:8])
assert len(files) == 18
with zipfile.ZipFile("/kaggle/working/submission.zip", "w", zipfile.ZIP_DEFLATED, compresslevel=6) as z:
    for n in sorted(files):
        zi = zipfile.ZipInfo(n, date_time=(2026, 11, 19, 0, 0, 0)); zi.compress_type = zipfile.ZIP_DEFLATED
        z.writestr(zi, files[n])
p = pathlib.Path("/kaggle/working/submission.zip"); print("written", p.stat().st_size, "bytes", sha(p.read_bytes())[:12])
```
