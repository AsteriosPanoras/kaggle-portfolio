**Files** (public work, credited in the source notebooks): invsqrt `d0a9a694` (74,339.71; Syed Asad Ali, Ali Haydar Ozdag, maoxin123, Jeff Emerson, Leo Tang; measured in Mizero Lucky Gall's notebook), DQI `db0a845f` (345,110,199,047.76; swap-collapse + leaf-port by goodpjw2008 on the Syed Asad Ali / Hayato Hikawa / Koushik Rudra / manwithacat line), chemistry `6ef06081` (Manvender Singh refit; loaders by Lera / marculera and maoxin123).


```python
import zipfile, hashlib, pathlib, shutil
WANT = {"invsqrt.kmb": "d0a9a694", "dqi.kmb": "db0a845f", "dfthcsos/submission.npz": "6ef06081", "dfthcsos/angle_k_0.kmb": "ed648708"}
sha = lambda b: hashlib.sha256(b).hexdigest()
src = None
for p in pathlib.Path("/kaggle/input").rglob("submission.zip"):
    try:
        z = zipfile.ZipFile(p)
        if len(z.namelist()) == 18 and all(sha(z.read(n)).startswith(h) for n, h in WANT.items()): src = p; break
    except Exception as e: print("skip", p, e)
assert src, "source submission.zip not found"
z = zipfile.ZipFile(src)
for n, h in WANT.items(): print(f"{n:28s} {sha(z.read(n))[:12]}")
shutil.copyfile(src, "/kaggle/working/submission.zip")
p = pathlib.Path("/kaggle/working/submission.zip"); print("source:", src); print("written", p.stat().st_size, "bytes", sha(p.read_bytes())[:12])
```
