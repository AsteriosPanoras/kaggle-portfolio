## The score, task by task: the official verifiers
The next cell runs the three **unmodified official verifiers** of the competition (from the competition data) on the exact files above, at their default salt, and computes the score as the board does: the geometric mean of reference cost / cost over the three tasks. The invsqrt run checks every input and takes about 15 minutes on a Kaggle CPU; the whole page runs in about an hour.

```python
COMP = next(p for p in [os.environ.get("QCC_COMP", ""), "/kaggle/input/competitions/google-verifiable-quantum-circuit-compilation",
                        "/kaggle/input/google-verifiable-quantum-circuit-compilation"] if p and os.path.isdir(p + "/verifiers"))
VER, KICK = COMP + "/verifiers", COMP + "/kickmix"
WORK = pathlib.Path(os.environ.get("QCC_WORK", "/tmp/qcc_work")); shutil.rmtree(WORK, ignore_errors=True); WORK.mkdir(parents=True)
for n in SET: zipfile.ZipFile(SRC[n]).extract(n, WORK)
ENV = dict(os.environ, PYTHONPATH=os.pathsep.join([KICK, VER] + ([os.environ["PYTHONPATH"]] if os.environ.get("PYTHONPATH") else [])))
def run_json(args, timeout=7200):
    r = subprocess.run([sys.executable] + [str(x) for x in args], cwd=VER, env=ENV, capture_output=True, text=True, timeout=timeout)
    out = r.stdout + r.stderr; dec = json.JSONDecoder(); payload = None
    for i, ch in enumerate(out):
        if ch == "{":
            try:
                obj, _ = dec.raw_decode(out[i:])
                if isinstance(obj, dict) and "status" in obj: payload = obj
            except json.JSONDecodeError: continue
    return payload or {"status": "NO_JSON", "tail": out[-800:]}
REF = {"invsqrt": 2076774.233715282, "dqi": 24439463583672.93, "dfthcsos": 3355599588.3155375}
RES = {}
RES["dfthcsos"] = run_json(["dfthcsos/verify.py", WORK / "dfthcsos"])
r = RES["dfthcsos"]; log("chemistry", r.get("status"), "cost", r.get("score"), "lambda_eff", r.get("lambda_eff"), "T", r.get("toffoli_count"))
RES["invsqrt"] = run_json(["invsqrt/verify_submission_invsqrt.py", WORK / "invsqrt.kmb", "--max-workers", str(os.cpu_count() or 4)] + (["--max-x", "20000"] if QUICK else []))
r = RES["invsqrt"]; log("invsqrt", r.get("status"), "cost", r.get("score"), "Toffolis/shot", r.get("toffoli_count"))
RES["dqi"] = run_json(["dqi/verify_submission_dqi.py", WORK / "dqi.kmb"] + (["--num-shots", "4096"] if QUICK else []))
r = RES["dqi"]; log("dqi (default salt)", r.get("status"), "cost", r.get("score"), "value errors", r.get("value_errors"), "phase errors", r.get("phase_errors"))
for k in RES: assert RES[k].get("status") == "PASS", (k, RES[k])
ratio = {k: REF[k] / RES[k]["score"] for k in REF}
GM = math.prod(ratio.values()) ** (1 / 3)
i, d, c = RES["invsqrt"], RES["dqi"], RES["dfthcsos"]
extra = {"invsqrt": f"{i.get('toffoli_count'):.4f} Toffolis and {i.get('t_count', float('nan')):.4f} T per shot, {i.get('num_qubits')} qubits",
         "dqi": f"{d.get('toffoli_count'):,.1f} Toffolis per shot, {d.get('num_qubits')} qubits, {d.get('value_errors')} value and {d.get('phase_errors')} phase errors in {d.get('verified_shots', 0):,} shots",
         "dfthcsos": f"lambda_eff {c.get('lambda_eff'):.9f}, {c.get('toffoli_count'):,.0f} Toffolis per block, {c.get('num_qubits')} qubits"}
name = {"invsqrt": "invsqrt", "dqi": "DQI", "dfthcsos": "chemistry"}
rows = "".join("<tr>" + cell(name[k]) + cell(f"{RES[k]['score']:,.2f}") + cell(f"{ratio[k]:.5f}") + cell(extra[k]) + "</tr>" for k in REF)
rows += "<tr>" + cell("<b>geometric mean</b>", "#e9f7ef") + cell("", "#e9f7ef") + cell(f"<b>{GM:.5f}</b>", "#e9f7ef") + cell("default salt of the verifiers", "#e9f7ef") + "</tr>"
display(HTML('<table style="border-collapse:collapse;color:#111111;background:#ffffff;">' + "<tr>" + th("task") + th("cost (official verifier)") + th("reference / cost") + th("what the verifier measured") + "</tr>" + rows + "</table>"))
```

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #f4f6fa; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>What is 0.01 point worth?</b> The score is a geometric mean, so it moves like cost to the power minus one third: a relative cost cut of 3 &times; 0.01 / 33.2, about 0.09 percent on one task, adds 0.01 points. The next cell turns that into units of each task, to first order and using the numbers the verifiers just printed. It is a guide for choosing where to look, not a promise.</div>

```python
rel = 3 * 0.01 / GM                         # relative cost cut of one task that adds 0.01 points
q_inv = i.get("num_qubits") or 37
tb = c.get("toffoli_count"); leff = c.get("lambda_eff")
rows = (
  "<tr>" + cell("invsqrt") + cell(f"{rel * 100:.3f} %") + cell(f"{rel * RES['invsqrt']['score'] / q_inv:.2f} of {RES['invsqrt']['score'] / q_inv:,.1f} cost units per shot, or {rel * q_inv:.3f} of a qubit (the cost is qubits times the per-shot price of Toffolis, T gates and rotations)") + "</tr>"
  "<tr>" + cell("DQI") + cell(f"{rel * 100:.3f} %") + cell(f"about {rel * d.get('toffoli_count'):,.0f} of {d.get('toffoli_count'):,.0f} Toffolis per shot, or {rel * d.get('num_qubits'):.2f} of {d.get('num_qubits')} qubits, or {rel / 0.0038:.2f} of one failed shot (in the salt table one failed shot costs about 0.4 percent of the DQI cost, about 0.04 points)") + "</tr>"
  "<tr>" + cell("chemistry") + cell(f"{rel * 100:.3f} %") + cell(f"{rel * tb:.1f} of {tb:,.0f} Toffolis per block, or {rel * leff:.4f} off lambda_eff = {leff:.6f}, or {rel * c.get('num_qubits'):.2f} of {c.get('num_qubits')} qubits") + "</tr>")
display(HTML('<table style="border-collapse:collapse;color:#111111;background:#ffffff;">' + "<tr>" + th("task") + th("cost cut for +0.01") + th("in the units of the task") + "</tr>" + rows + "</table>"))
```

## How much the DQI score depends on the shot draw
The DQI verifier draws 102,400 random test shots. The seed comes from the SHA-256 of the file and a salt; the board uses a salt we do not know. A shot on which the decoder fails raises the cost, because the verifier prices the repetitions needed to make up for failures. The next cell runs the same DQI file at the default salt and at five more salts (<code>s1</code> to <code>s5</code>) and shows the score each draw gives. The chemistry cost does not depend on the salt and the invsqrt run checks every input; only the DQI moves.

```python
import matplotlib.pyplot as plt
SALTS = ["", "s1", "s2", "s3", "s4", "s5"]
if QUICK: SALTS = ["", "s1"]
runs = {}
for s in SALTS:
    r = RES["dqi"] if s == "" else run_json(["dqi/verify_submission_dqi.py", WORK / "dqi.kmb", "--salt", s] + (["--num-shots", "4096"] if QUICK else []))
    assert r.get("status") == "PASS", (s, r)
    runs[s or "default"] = r
    log("dqi salt", repr(s), "value errors", r.get("value_errors"), "cost", f"{r['score']:,.0f}")
base = ratio["invsqrt"] * ratio["dfthcsos"]
gms = {k: (base * REF["dqi"] / r["score"]) ** (1 / 3) for k, r in runs.items()}
rows = "".join("<tr>" + cell(k) + cell(str(r.get("value_errors"))) + cell(str(r.get("phase_errors"))) + cell(f"{r['score'] / 1e9:,.3f}") + cell(f"{REF['dqi'] / r['score']:.4f}") + cell(f"{gms[k]:.5f}") + "</tr>" for k, r in runs.items())
display(HTML('<table style="border-collapse:collapse;color:#111111;background:#ffffff;">' + "<tr>" + th("salt") + th("value errors") + th("phase errors") + th("DQI cost (10^9)") + th("DQI reference / cost") + th("score of the three files") + "</tr>" + rows + "</table>"))
vals = list(gms.values())
print(f"score over {len(vals)} draws: lowest {min(vals):.5f}, highest {max(vals):.5f}, mean {sum(vals) / len(vals):.5f}")
if BOARD_SCORE:
    need = BOARD_SCORE ** 3 / base
    print(f"the board score {BOARD_SCORE} of these bytes implies a DQI cost of {REF['dqi'] / need / 1e9:,.3f}e9 (reference / cost {need:.4f}) on the board's own draw")
fig, ax = plt.subplots(figsize=(7, 3.4))
ks = list(gms); ax.bar(ks, [gms[k] for k in ks], color="#1d5aa6", width=0.6)
lo = min(vals + ([BOARD_SCORE] if BOARD_SCORE else [])); ax.set_ylim(lo - 0.05, max(vals + ([BOARD_SCORE] if BOARD_SCORE else [])) + 0.03)
if BOARD_SCORE: ax.axhline(BOARD_SCORE, color="#c0392b", lw=1.6, ls="--"); ax.text(len(ks) - 0.5, BOARD_SCORE + 0.003, f"board {BOARD_SCORE}", ha="right", color="#c0392b", fontsize=9)
for i_, k in enumerate(ks): ax.text(i_, gms[k] + 0.002, f"{runs[k].get('value_errors')} err.", ha="center", fontsize=8.5, color="#111111")
ax.set_ylabel("score of the three files"); ax.set_xlabel("salt of the DQI draw (errors = failed shots out of 102,400)")
ax.set_facecolor("#ffffff"); fig.patch.set_facecolor("#ffffff")
for sp in ("top", "right"): ax.spines[sp].set_visible(False)
plt.tight_layout(); plt.show()
```

## The ideas, in plain words
What each last-step change does and why it saves cost. These are the authors' own descriptions, shortened; their notebooks have the derivations. We ran the verifiers on the results, but we did not re-derive the ideas.

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>Three T gates fewer (invsqrt, <a href="https://www.kaggle.com/code/varianceofx/vqcc-three-t-phase-compiler" style="color: #0b3d91; background: transparent;">varianceofx</a>).</b> Two wires later in the circuit hold values u and v with u - v = a(1 - b) mod 4. An S-dagger on u and an S on v then give the same phase as the T gates did, and no T gate is needed. The Toffoli count is unchanged; the T count per shot goes from 8.5 to 5.5 (cost 60,955 to 60,927.6).</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>A last verified cut of the invsqrt (<a href="https://www.kaggle.com/code/prvsiyan/verifier-guided-quantum-circuit-optimization" style="color: #0b3d91; background: transparent;">prvsiyan</a>).</b> The file <code>6b592145</code> has a mean of 1,406.31156 Toffolis per shot against 1,406.31243 for <code>bb676b95</code> and costs 0.03 less. The notebook of alexchilton and our own reading do not trace what changed; the authors report that it passes on all 200,540,928 inputs.</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>Zero-test cuts (DQI, <a href="https://www.kaggle.com/code/goodpjw2008/qcc-dqi-zero-test-cuts-lb-33-17067" style="color: #0b3d91; background: transparent;">goodpjw2008</a>, <a href="https://www.kaggle.com/code/evgendvorkin/google-quantum-ai-deep-eda-with-analogies" style="color: #0b3d91; background: transparent;">evgendvorkin</a>, <a href="https://www.kaggle.com/code/ozertuu/google-quantum-ai-vqcc-33-17067-sota-apex" style="color: #0b3d91; background: transparent;">ozertuu</a>).</b> A Toffoli-like gate can become a cheaper gate when one of its controls is implied by the others on the shots that reach it. The decoder tests a 24-bit value for zero with a chain of AND gates; in positions 17 to 24 the extra control was found to be implied on the shots checked, so the AND (CCX) is replaced by a plain CX. The price is that the circuit no longer checks cases that almost never occur: a rare failed shot raises the cost. In an earlier run of this page on <code>6f78ecdd</code> the default salt drew 4 failed shots out of 102,400 and three of five other salts drew none.</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>Shorter chemistry loaders (<a href="https://www.kaggle.com/code/goodpjw2008/qcc-shorter-chemistry-loaders-lb-33-25057" style="color: #0b3d91; background: transparent;">goodpjw2008</a>).</b> A loader XORs 35 bits into the word registers for 1,219 addresses. Every Toffoli adds one product of two linear functions of the address to a set of outputs. goodpjw2008 searches for loaders whose products cover less. One pass goes from 3,357 Toffolis (maoxin123) to 3,310 in version 19, and since the loaders run twice the chemistry T count goes from 13,214 to 13,120.</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>Dead words (<a href="https://www.kaggle.com/code/goodpjw2008/qcc-shorter-chemistry-loaders-lb-33-25057" style="color: #0b3d91; background: transparent;">goodpjw2008</a>, versions 18 and 19).</b> Each of the 91,425 words is a rotation angle. 2,400 of them rotate an orbital whose amplitude is exactly zero, so they may take any value without changing the physics. Version 19 gives new values to 2,261 of them; each gives a few free bits to the search for the loaders. The words are seen by two loaders (the one that loads and the one that removes), so the loaders of a group are solved one after the other.</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>A continued weight refit (<a href="https://www.kaggle.com/code/goodpjw2008/qcc-shorter-chemistry-loaders-lb-33-25057" style="color: #0b3d91; background: transparent;">goodpjw2008</a> on <a href="https://www.kaggle.com/code/evgendvorkin/google-quantum-ai-deep-eda-with-analogies" style="color: #0b3d91; background: transparent;">evgendvorkin</a>'s table).</b> The chemistry cost is lambda_eff times the Toffoli count. With the words fixed, the weights can be re-optimised against the official accuracy limits; lambda_eff is now 46.914304.</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; font-size: 15.5px; line-height: 1.6; color: #111111; background: #ffffff; border-left: 6px solid #1d5aa6; border-radius: 8px; padding: 12px 16px; max-width: 100%; overflow-wrap: anywhere;"><b>Another DQI file (<a href="https://www.kaggle.com/code/logiclab/quantum-circuit-compilation-v58-submission" style="color: #0b3d91; background: transparent;">logiclab</a>).</b> <code>b183c793</code> is published without a description; the title of its notebook says 10.5 Toffolis per shot fewer than <code>6f78ecdd</code>.
